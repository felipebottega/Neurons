# Neurons — Diagnóstico da função $H$ e da cobertura de conexões

Este notebook foi construído a partir das implementações **`src/functions_native.cpp`**, **`src/brain_builder_native.cpp`** e **`src/genome_native.cpp`** do projeto NeuronsCpp. Ele investiga por que a proporção de neurônios isolados **antes do resgate** cresce com o tamanho do cérebro.

**Perguntas principais**

1. A própria função $H(A_{\rm origem},A_{\rm destino})$ favorece alguns destinos sistematicamente?
2. Os valores espaciais de $A$ tornam muitos neurônios quase indistinguíveis para a seleção?
3. Escolher os **$N_{\max}$ maiores valores positivos de $H$** concentra mais entradas que um sorteio uniforme entre candidatos com $H>0$?
4. A concentração aumenta com as dimensões mesmo quando $N_{\max}$ também cresce?

**Escopo:** simula a **geração inicial**, a **remoção de conexões proibidas** e a **contagem de isolados antes do resgate**. Não simula o resgate, plasticidade ou dinâmica. Os dados gerados aqui não substituem uma execução no Godot: NumPy e C++ podem divergir minimamente em arredondamentos `float32` e desempates.

**Requisitos:** Python 3.10+, `numpy`, `pandas`, `matplotlib`, `IPython` (ambiente Jupyter típico). Sem Godot, compilação ou `scipy`.

## 0. Configurações

Deixe `PROFILE = "full"` para analisar as quatro escalas; esse modo avalia **todos os neurônios de origem** e pode demorar, especialmente em 42 × 42 × 10. `"quick"` é um teste rápido do notebook. Ajuste `REPLICATES_FULL` para mais amostras quando quiser resultados estatisticamente mais estáveis.

**Importante:** não extrapole os resultados de uma ou duas amostras para toda a população genética. As médias por dimensão são mostradas junto com o tamanho da amostra.

```python
import time, json, math, zipfile
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

PROFILE = "full"            # "quick" ou "full"
SEED = 20261024
K = 10
N_INPUTS = 10
N_OUTPUTS = 10
REPLICATES_FULL = {10: 4, 22: 3, 32: 2, 42: 2}
REPLICATES_QUICK = {10: 2, 22: 1}
SIZES = REPLICATES_FULL if PROFILE == "full" else REPLICATES_QUICK
OUTPUT_DIR = Path("neurons_h_results")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

# Opcional: análise de um genoma específico. Substitua as três strings
# pela sequência quaternária do seu genoma; deixe None para apenas genomas aleatórios.
CUSTOM_GENOME = None  # {"connections": "...64 dígitos...", "beta": "...8...", "max_connections": "...8..."}
CUSTOM_SIZE = (22, 22, 10)

# O controle aleatório seleciona N_max destinos dentre os mesmos candidatos com H > 0
# antes da filtragem de conexões proibidas (tal como o construtor faz).
RUN_POSITIVE_RANDOM_CONTROL = True
STORE_SOURCE_SAMPLES = 48

print("Perfil:", PROFILE, "| repetições:", SIZES)
print("Resultados:", OUTPUT_DIR.resolve())
```

## 1. Reprodução de $A$, $\beta$, $N_{\max}$ e $H$

No código nativo:

$$
A(i,j,k)=\tanh\!\left[\frac1{\sqrt{64}}\sum_{r,s,q=0}^{3}
\left(\frac{2x_{q+4s+16r}}3-1\right)
\cos\!\left(\frac{\pi ri}{I-1}\right)
\cos\!\left(\frac{\pi sj}{J-1}\right)
\cos\!\left(\frac{\pi qk}{K-1}\right)\right].
$$

$$
H(a,b)=\sin\left(\frac\pi{16}+\sum_{m=1}^4 c_m\cos(ma)\sin(mb)\right),\qquad
c_m=\frac{3\pi}{4\sum_{t=0}^{63}\sqrt{1+t}}
\sum_{t:x_t=m-1}\sqrt{1+t}.
$$

Observe que **$H(a,b)\neq H(b,a)$**, em geral. O gene `connections` determina **tanto o campo $A$** quanto os coeficientes de $H$. A seleção posterior usa $H>0$ e os maiores valores disponíveis.

Para velocidade, pré-calculamos seno/cosseno dos valores de $A$, mas **não alteramos a expressão matemática de $H$**. Pequenas diferenças em `float32` em relação ao C++ são possíveis.

```python
PI32 = np.float32(np.pi)
W8 = np.sqrt(np.arange(1, 9, dtype=np.float32))
W64 = np.sqrt(np.arange(1, 65, dtype=np.float32))

def random_genome(seed):
    rng = np.random.default_rng(seed)
    # O construtor C++ gera dígitos independentes e uniformes de 0 a 3.
    return {"connections": rng.integers(0,4,64,dtype=np.int8),
            "beta": rng.integers(0,4,8,dtype=np.int8),
            "max_connections": rng.integers(0,4,8,dtype=np.int8)}

def parse_genome(dictionary):
    lengths = {"connections": 64, "beta": 8, "max_connections": 8}
    ans = {}
    for name, count in lengths.items():
        value = dictionary[name]
        if isinstance(value, str):
            if len(value) != count or not set(value) <= set("0123"):
                raise ValueError(f"{name}: esperado {count} dígitos quaternários")
            ans[name] = np.array([int(ch) for ch in value], dtype=np.int8)
        else:
            ans[name] = np.asarray(value, dtype=np.int8)
            if ans[name].shape != (count,) or not np.isin(ans[name], [0,1,2,3]).all():
                raise ValueError(f"{name}: formato inválido")
    return ans

def genome_text(g):
    return {key: "".join(map(str, val)) for key, val in g.items()}

def beta_max(g, I, J):
    beta = np.float32(1.0 + np.sum(W8*g["beta"],dtype=np.float32)
                     / np.sum(3*W8,dtype=np.float32) * np.float32((I*J/16.)**(1/3)))
    maxconn = int(np.floor(np.sum(W8*np.power(beta, g["max_connections"].astype(np.float32)),dtype=np.float32)))
    return float(beta), maxconn

def h_coefficients(g):
    digits = g["connections"]
    return np.array([np.sum(W64[digits == d],dtype=np.float32) for d in range(4)],dtype=np.float32) * (np.float32(3)*PI32 / (np.float32(4)*np.sum(W64,dtype=np.float32)))

def make_A(g, I, J, K):
    # A[k,j,i], linearizado como i + I*j + I*J*k.
    coeff = (np.float32(2)/np.float32(3))*g["connections"].astype(np.float32)-np.float32(1)
    cx = np.cos(PI32*np.arange(4,dtype=np.float32)[:,None]*
                (np.arange(I,dtype=np.float32)[None,:]/np.float32(I-1))).astype(np.float32)
    cy = np.cos(PI32*np.arange(4,dtype=np.float32)[:,None]*
                (np.arange(J,dtype=np.float32)[None,:]/np.float32(J-1))).astype(np.float32)
    cz = np.cos(PI32*np.arange(4,dtype=np.float32)[:,None]*
                (np.arange(K,dtype=np.float32)[None,:]/np.float32(K-1))).astype(np.float32)
    field = np.zeros((K,J,I), dtype=np.float32)
    for r in range(4):
        for s in range(4):
            for q in range(4):
                field += coeff[q + 4*s + 16*r] * cx[r][None,None,:] * cy[s][None,:,None] * cz[q][:,None,None]
    return np.tanh(field / np.float32(8)).astype(np.float32).ravel()

def h_direct(a, b, c):
    z = np.float32(np.pi/16)
    for m in range(1, 5):
        z = z + c[m-1]*np.cos(np.float32(m)*a)*np.sin(np.float32(m)*b)
    return np.sin(z)

def make_fast_h(A, coeffs):
    harmonics = np.arange(1,5,dtype=np.float32)[:,None]
    ac = (coeffs[:,None]*np.cos(harmonics*A[None,:])).astype(np.float32)
    bs = np.sin(harmonics*A[None,:]).astype(np.float32)
    return ac, bs

def h_from_cached(ac, bs, src, target_indices):
    z = np.full(len(target_indices), np.float32(np.pi/16), dtype=np.float32)
    for m in range(4):
        z += ac[m, src] * bs[m, target_indices]
    return np.sin(z).astype(np.float32)

# Testes de sanidade (independentes do Godot)
test_g = random_genome(7)
a_test = make_A(test_g, 5, 5, 4)
c_test = h_coefficients(test_g)
ac_test, bs_test = make_fast_h(a_test, c_test)
i_test = np.arange(0, len(a_test), 2, dtype=int)
assert np.all(np.abs(a_test) <= 1)
assert np.isclose(c_test.sum(), 3*np.pi/4, atol=1e-6)
assert np.allclose(h_from_cached(ac_test, bs_test, 0, i_test),
                   h_direct(a_test[0], a_test[i_test], c_test), atol=1e-6)
print("Testes locais de A e H: OK. Exemplo β, Nmax:", beta_max(test_g,10,10))
```

## 2. Visualização da função $H$ isoladamente

Os gráficos mostram $H$ sobre o domínio $A\in[-1,1]$, sua **assimetria direcional** e a parcela do plano em que $H>0$. Isto descreve a função, **não ainda a distribuição real dos neurônios no espaço**.

```python
demo_genome = parse_genome(CUSTOM_GENOME) if CUSTOM_GENOME is not None else random_genome(SEED)
demo_c = h_coefficients(demo_genome)
xx = np.linspace(-1,1,161,dtype=np.float32)
Hgrid = h_direct(xx[:,None], xx[None,:], demo_c)
Hswap = Hgrid - Hgrid.T
fig, axs = plt.subplots(1,3,figsize=(16,4.5),constrained_layout=True)
m = axs[0].imshow(Hgrid,origin="lower",extent=(-1,1,-1,1),vmin=-1,vmax=1,cmap="coolwarm",aspect="auto")
axs[0].set(xlabel="A destino",ylabel="A origem",title="H(origem, destino)");fig.colorbar(m,ax=axs[0])
m = axs[1].imshow(Hswap,origin="lower",extent=(-1,1,-1,1),cmap="coolwarm",aspect="auto")
axs[1].set(xlabel="A destino",ylabel="A origem",title="H(a,b) − H(b,a)");fig.colorbar(m,ax=axs[1])
axs[2].bar(np.arange(1,5),demo_c);axs[2].set(xticks=np.arange(1,5),xlabel="Harmônico m",ylabel="Coeficiente",title="Coeficientes genéticos de H")
plt.show()
print("Área com H > 0 na grade uniforme:",round(100*np.mean(Hgrid>0),2),"%")
print("Coeficientes:",np.round(demo_c,5))
print("Gene connections:",genome_text(demo_genome)["connections"])
```

## 3. Replica fiel das regras de seleção inicial

São reproduzidos os detalhes que podem afetar o isolamento:

- Todos os neurônios são ordenados por $A$ **dentro de cada camada** e depois intercalados por ranking entre as camadas; os primeiros viram `input`, os últimos `output`.
- Cada origem não `output` considera destinos com $|\Delta k|\le 2$, excluindo a si mesma.
- Avalia **todos** esses destinos, retém candidatos com $H>0$ e seleciona até $N_{\max}$ maiores valores **antes** da limpeza.
- Na limpeza: remove `* → input` e `input → output` (a origem `output` não gera conexões).
- Um neurônio interno/de saída isolado tem **nenhuma entrada**; um `input` isolado tem **nenhuma saída**.

**Controle:** mesma geometria, mesmo genoma, mesmo conjunto de destinos com $H>0$, mesmo $N_{\max}$ e mesma limpeza, mas substitui a escolha dos maiores $H$ por uma escolha **aleatória uniforme** entre os positivos. É um experimento contrafactual de diagnóstico, **não uma proposta de alterar o modelo**.

A simulação **não** executa o resgate. Consequentemente não usa o custo enorme de exportar milhões de dicionários do Godot.

```python
def flags_by_rank(A, I, J, K, n_inputs, n_outputs):
    N = len(A)
    if n_inputs+n_outputs > N:
        raise ValueError("Número de inputs+outputs maior que número de neurônios")
    by_layer = []
    for k in range(K):
        ix = np.arange(k*I*J,(k+1)*I*J,dtype=np.int32)
        by_layer.append(ix[np.argsort(A[ix],kind="quicksort")])
    # rank 0 de cada camada; depois rank 1 de cada camada; ...
    permutation = np.stack(by_layer,axis=1).ravel()
    is_input = np.zeros(N,dtype=bool); is_output = np.zeros(N,dtype=bool)
    is_input[permutation[:n_inputs]] = True
    is_output[permutation[-n_outputs:]] = True
    return is_input, is_output

def gini_nonnegative(data):
    x = np.sort(np.asarray(data,dtype=np.float64)); n=len(x)
    if n == 0 or np.sum(x) == 0: return 0.
    return float((2*np.dot(np.arange(1,n+1),x)/(n*np.sum(x))) - (n+1)/n)

def metrics(incoming, outgoing, is_input, is_output):
    internal = ~(is_input | is_output)
    counts = {"internal":int(np.count_nonzero((incoming==0)&internal)),
              "output":int(np.count_nonzero((incoming==0)&is_output)),
              "input":int(np.count_nonzero((outgoing==0)&is_input))}
    N=len(incoming)
    eligible = ~is_input
    vals = incoming[eligible]
    tenpct = max(1,int(np.ceil(.1*len(vals))))
    total = np.sum(vals)
    top_share = float(np.partition(vals,-tenpct)[-tenpct:].sum()/total) if total else np.nan
    return {"isolated_total":sum(counts.values()),"isolation_pct":100.*sum(counts.values())/N,
            "isolated_internal":counts["internal"],"isolated_output":counts["output"],"isolated_input":counts["input"],
            "mean_incoming":float(np.mean(vals)),"median_incoming":float(np.median(vals)),
            "gini_incoming":gini_nonnegative(vals),"top_10pct_incoming_share":top_share,
            "zero_incoming_eligible_pct":100.*np.mean(vals==0),"total_valid_connections":int(np.sum(vals))}

def simulate(g, size, seed=0, baseline=True, keep_pairs=48, verbose=False):
    I,J,K = map(int,size)
    N=I*J*K
    tic=time.perf_counter()
    A=make_A(g,I,J,K)
    c=h_coefficients(g)
    beta, max_conn=beta_max(g,I,J)
    is_input,is_output=flags_by_rank(A,I,J,K,N_INPUTS,N_OUTPUTS)
    ac,bs=make_fast_h(A,c)
    # A lista de índices por camada já está linearizada na mesma ordem espacial do C++.
    sources=np.flatnonzero(~is_output)
    inc=np.zeros(N,dtype=np.int32); out=np.zeros(N,dtype=np.int32)
    inc_null=np.zeros(N,dtype=np.int32); out_null=np.zeros(N,dtype=np.int32)
    rng=np.random.default_rng(seed+800000)
    chosen_sources=set(rng.choice(sources,size=min(keep_pairs,len(sources)),replace=False).tolist())
    selected_sets={}
    # Valor de H em candidaturas amostradas, opcionalmente para interpretação dos gráficos.
    sample_h=[], []
    n_pos=0; n_eligible=0; removed=0; preclean=0
    per_layer_targets = [np.arange(k*I*J,(k+1)*I*J,dtype=np.int32) for k in range(K)]
    windows = [np.concatenate(per_layer_targets[max(0,k-2):min(K,k+3)]) for k in range(K)]
    for step,src in enumerate(sources):
        layer=int(src//(I*J))
        all_targets=windows[layer]
        targets=all_targets[all_targets!=src] # remove só a autoconexão
        h=h_from_cached(ac,bs,int(src),targets)
        positive=np.flatnonzero(h>0)
        n_pos+=len(positive);n_eligible+=len(targets)
        k_take=min(max_conn,len(positive))
        if k_take == 0:
            continue
        if k_take < len(positive):
            local=positive[np.argpartition(h[positive], -k_take)[-k_take:]]
        else:
            local=positive
        picks=targets[local]
        preclean+=len(picks)
        good=(~is_input[picks]) & (~(is_input[src]&is_output[picks]))
        removed+=len(picks)-int(good.sum())
        kept=picks[good]
        inc[kept]+=1 # um destino aparece no máximo 1 vez por origem
        out[src]+=len(kept)
        if src in chosen_sources:
            selected_sets[int(src)]=set(kept.tolist())
        if baseline:
            idx = rng.choice(positive,size=k_take,replace=False)
            random_picks=targets[idx]
            random_good=(~is_input[random_picks]) & (~(is_input[src]&is_output[random_picks]))
            selected=random_picks[random_good]
            inc_null[selected]+=1
            out_null[src]+=len(selected)
        if verbose and ((step+1)%max(1,(len(sources)//4))==0):
            print(f"   {step+1:,}/{len(sources):,} origens ({time.perf_counter()-tic:.1f} s)")
    obs=metrics(inc,out,is_input,is_output)
    obs_null=metrics(inc_null,out_null,is_input,is_output) if baseline else None
    within,between=[],[]
    sl=list(selected_sets)
    for i,src in enumerate(sl):
        for dst in sl[i+1:]:
            left,right=selected_sets[src], selected_sets[dst]
            if not left or not right: continue
            overlap=len(left & right)/min(len(left),len(right))
            if src//(I*J)==dst//(I*J): within.append(overlap)
            else: between.append(overlap)
    result={"size":size,"N":N,"beta":beta,"Nmax":max_conn,"coeff":c,
            "A":A,"is_input":is_input,"is_output":is_output,"incoming":inc,"outgoing":out,
            "incoming_null":inc_null,"outgoing_null":out_null,
            "metrics":obs,"null_metrics":obs_null,
            "positive_candidate_pct":100*n_pos/n_eligible,"removed_after_topk":removed,
            "preclean_connections":preclean,"selection_removed_pct":100*removed/preclean if preclean else 0.,
            "same_layer_overlap":float(np.mean(within)) if within else np.nan,
            "different_layer_overlap":float(np.mean(between)) if between else np.nan,
            "seconds":time.perf_counter()-tic,"genome":genome_text(g)}
    return result

# Verificação: num caso onde Nmax é muito maior que os candidatos positivos,
# os dois métodos convergem (salvo ordem), pois ambos devem escolher todos os positivos.
smoke_g=random_genome(123)
smoke=simulate(smoke_g,(4,4,3),seed=1,baseline=True,keep_pairs=8)
assert np.array_equal(smoke["incoming"],smoke["incoming_null"])
assert smoke["metrics"]["isolation_pct"]==smoke["null_metrics"]["isolation_pct"]
print("Teste da seleção / limpeza / controle: OK", "| Nmax =",smoke["Nmax"])
```

## 4. Executar os experimentos

Cada execução simula **todas as origens** e **todos os candidatos permitidos**, inclusive nos cérebros maiores. Portanto, a taxa de isolamento obtida é a taxa **efetiva antes do resgate**, não uma estimativa criada a partir de uma pequena amostra de origens. O tempo pode ser alto: prefira CPU disponível e acompanhe o contador de execuções.

```python
experiments=[]
run_start=time.perf_counter()
for I, reps in SIZES.items():
    for rep in range(reps):
        local_seed = SEED + I*10000 + rep
        genome = random_genome(local_seed)
        label=f"{I}x{I}x{K} / rep {rep+1}/{reps}"
        print("Iniciando",label,flush=True)
        res=simulate(genome,(I,I,K),seed=local_seed,baseline=RUN_POSITIVE_RANDOM_CONTROL,
                     keep_pairs=STORE_SOURCE_SAMPLES,verbose=False)
        res["label"]=label;res["seed"]=local_seed
        experiments.append(res)
        print(f"  Nmax={res['Nmax']}, isolados={res['metrics']['isolation_pct']:.2f}%"
              + (f", controle={res['null_metrics']['isolation_pct']:.2f}%" if res['null_metrics'] else "")
              + f", tempo={res['seconds']:.1f}s",flush=True)
if CUSTOM_GENOME is not None:
    cg=parse_genome(CUSTOM_GENOME)
    cr=simulate(cg,CUSTOM_SIZE,seed=SEED+999,baseline=RUN_POSITIVE_RANDOM_CONTROL,
                keep_pairs=STORE_SOURCE_SAMPLES)
    cr['label']='CUSTOM';cr['seed']=SEED+999
    experiments.append(cr)
    print(f"CUSTOM: isolados={cr['metrics']['isolation_pct']:.2f}%, Nmax={cr['Nmax']}")
print(f"Concluído: {len(experiments)} cérebros em {time.perf_counter()-run_start:.1f}s")
```

## 5. Tabelas principais

Diferença `H − aleatório` positiva significa que **selecionar os melhores valores de H deixou mais neurônios isolados** do que escolher aleatoriamente entre todos os candidatos de H positivo, para o mesmo genoma e o mesmo Nmax.

**Limitação importante:** no controle, a quantidade de conexões após a limpeza pode ser um pouco diferente, porque destinos proibidos são escolhidos em proporções diferentes. Confira a coluna `valid_edges` antes de atribuir toda a diferença à concentração de H.

```python
rows=[]
for r in experiments:
    m=r['metrics']; m0=r['null_metrics']
    row={"label":r["label"],"I":r["size"][0],"N":r["N"],"seed":r["seed"],
         "beta":r["beta"],"Nmax":r["Nmax"],
         "H_positive_pct":r["positive_candidate_pct"],
         "isolation_H_pct":m["isolation_pct"],"isolated_internal":m["isolated_internal"],
         "isolated_output":m["isolated_output"],"isolated_input":m["isolated_input"],
         "valid_edges_H":m["total_valid_connections"],
         "gini_H":m["gini_incoming"],"top10_share_H":m["top_10pct_incoming_share"],
         "same_layer_overlap":r["same_layer_overlap"],
         "different_layer_overlap":r["different_layer_overlap"],
         "selection_removed_pct":r["selection_removed_pct"],
         "run_seconds":r["seconds"]}
    if m0:
        row.update(isolation_random_pct=m0["isolation_pct"],
                   difference_pp=m["isolation_pct"]-m0["isolation_pct"],
                   valid_edges_random=m0["total_valid_connections"],
                   gini_random=m0["gini_incoming"],
                   top10_share_random=m0["top_10pct_incoming_share"])
    rows.append(row)
df=pd.DataFrame(rows)
display(df.round(3))
summary=df.groupby(["I","N"],as_index=False).agg(
    n_brains=("N","size"),mean_rescue_H=("isolation_H_pct","mean"),
    sd_rescue_H=("isolation_H_pct","std"),mean_Nmax=("Nmax","mean"),
    mean_gini_H=("gini_H","mean"), mean_overlap=("same_layer_overlap","mean"))
if RUN_POSITIVE_RANDOM_CONTROL:
    null_summary=df.groupby(["I","N"],as_index=False).agg(
        mean_rescue_random=("isolation_random_pct","mean"),
        mean_difference_pp=("difference_pp","mean"),mean_gini_random=("gini_random","mean"))
    summary=summary.merge(null_summary,on=["I","N"])
print("Resumo por dimensão:")
display(summary.round(3))
df.to_csv(OUTPUT_DIR/"per_brain_metrics.csv",index=False)
summary.to_csv(OUTPUT_DIR/"per_size_summary.csv",index=False)
```

## 6. Escala: isolamento, heterogeneidade e competição por destinos

## 7. Diagnóstico detalhado de um cérebro

Escolhemos, por padrão, o **maior cérebro simulado**. Os gráficos permitem distinguir:

- Destinos com $A$ semelhante recebendo sistematicamente mais conexões.
- Padrões espaciais de concentração dentro das camadas.
- Muitos destinos com grau de entrada zero apesar de milhões de conexões emitidas.
- Maior desigualdade causada pela seleção dos máximos de $H$, quando comparada ao controle aleatório.

```python
chosen = max(experiments,key=lambda r:(r['N'],r['seed']))
I,J,KK=chosen['size']
A=chosen['A']; deg=chosen['incoming']; deg0=chosen['incoming_null']
valid=~chosen['is_input']
print(f"Selecionado: {chosen['label']} | seed={chosen['seed']} | Nmax={chosen['Nmax']}")
fig,axs=plt.subplots(1,3,figsize=(17,4.5),constrained_layout=True)
axs[0].hist(deg[valid],bins=60,alpha=.6,label='Maiores H')
if RUN_POSITIVE_RANDOM_CONTROL: axs[0].hist(deg0[valid],bins=60,alpha=.5,label='Aleatório dentre H>0')
axs[0].set(xlabel='Conexões recebidas',ylabel='Quantidade de destinos',title='Distribuição do grau de entrada')
axs[0].legend(fontsize=8)
axs[1].scatter(A[valid],deg[valid],s=5,alpha=.16)
axs[1].set(xlabel='A do neurônio de destino',ylabel='Entradas recebidas',title='Dependência entre A e conectividade')
order=np.argsort(A[valid])
a_sorted=A[valid][order]; d_sorted=deg[valid][order]
parts=np.array_split(np.arange(len(order)),30)
means=np.array([d_sorted[p].mean() for p in parts])
a_means=np.array([a_sorted[p].mean() for p in parts])
axs[2].plot(a_means,means,'-o',ms=3)
axs[2].set(xlabel='A do destino (30 grupos de igual tamanho)',ylabel='Entradas médias',title='Perfil de preferência por A')
for ax in axs:ax.grid(alpha=.2)
fig.savefig(OUTPUT_DIR/'detailed_indegree_vs_A.png',dpi=160)
plt.show()
```

```python
k_slice = KK//2
start = k_slice*I*J; end = start+I*J
fig,axs=plt.subplots(1,3,figsize=(15,4.4),constrained_layout=True)
for ax,values,title in zip(axs,[A[start:end],deg[start:end],deg0[start:end]],
                            ['Campo A','Entradas: maiores H','Entradas: aleatório com H>0']):
    im=ax.imshow(values.reshape(J,I),origin='lower',aspect='equal')
    ax.set(xlabel='i',ylabel='j',title=f'{title} (k={k_slice})')
    fig.colorbar(im,ax=ax)
fig.savefig(OUTPUT_DIR/'spatial_maps.png',dpi=160)
plt.show()
layer_table=[]
for k in range(KK):
    s=slice(k*I*J,(k+1)*I*J)
    v=~chosen['is_input'][s]
    layer_table.append({'layer_k':k,'A_mean':float(np.mean(A[s])),
         'isolated_H_pct':float(100*np.mean(deg[s][v]==0)),
         'isolated_random_pct':float(100*np.mean(deg0[s][v]==0)),
         'avg_incoming_H':float(np.mean(deg[s][v]))})
layer_df=pd.DataFrame(layer_table)
display(layer_df.round(3))
layer_df.to_csv(OUTPUT_DIR/'layers_detailed.csv',index=False)
```

## 8. Concentração, entradas nulas e sobreposição

A participação dos **10% de destinos mais conectados** mede concentração direta. O **Gini** vai de 0 (entradas perfeitamente uniformes) a próximo de 1 (muita concentração). Já a sobreposição mede a fração média de destinos comuns entre pares de origens selecionadas, normalizada pelo tamanho do menor conjunto. A sobreposição é calculada sobre apenas algumas origens escolhidas aleatoriamente para diagnóstico, enquanto o isolamento e o grau de entrada usam **todas** as origens.

```python
diag_cols=['label','Nmax','isolation_H_pct','gini_H','top10_share_H','same_layer_overlap','different_layer_overlap']
if RUN_POSITIVE_RANDOM_CONTROL:
    diag_cols += ['isolation_random_pct','gini_random','top10_share_random','difference_pp','valid_edges_H','valid_edges_random']
display(df[diag_cols].round(4))

# As 10% maiores entradas contra o restante, em unidades de porcentagem.
fig,axs=plt.subplots(1,2,figsize=(13,4.4),constrained_layout=True)
axs[0].bar(np.arange(len(df)),100*df['top10_share_H'],label='Maiores H',alpha=.7)
if RUN_POSITIVE_RANDOM_CONTROL:
    axs[0].plot(np.arange(len(df)),100*df['top10_share_random'],'ko',ms=3,label='Controle aleatório')
axs[0].set(ylabel='Entradas concentradas nos 10% principais (%)',xlabel='Índice do cérebro',title='Participação dos destinos dominantes')
axs[0].legend(fontsize=8)
axs[1].scatter(df['same_layer_overlap'],df['isolation_H_pct'])
axs[1].set(xlabel='Sobreposição de destinos (mesma camada)',ylabel='Isolamento (%)',title='Conexões semelhantes entre origens')
for ax in axs:ax.grid(alpha=.2)
fig.savefig(OUTPUT_DIR/'concentration_and_overlap.png',dpi=160)
plt.show()
```

## 9. Verificar se a restrição $H>0$ por si só explica o fenômeno

A coluna `H_positive_pct` informa a fração de pares geometricamente candidatos com $H>0$. Se ela for pequena, o espaço de candidatos já está restrito antes da seleção dos maiores valores. Se for grande e o controle aleatório tiver isolamento muito menor, o problema é principalmente a **concentração criada pelo ranking dos máximos de H**.

É importante observar simultaneamente `selection_removed_pct`: origens podem perder algumas das conexões escolhidas por apontarem para neurônios de entrada ou por conectarem entradas diretamente a saídas.

```python
display(df[['label','N','Nmax','H_positive_pct','selection_removed_pct',
            'isolation_H_pct']+(['isolation_random_pct','difference_pp'] if RUN_POSITIVE_RANDOM_CONTROL else [])].round(3))
fig,ax=plt.subplots(figsize=(9,4.5))
ax.scatter(df['H_positive_pct'],df['isolation_H_pct'],s=65)
for i,r in df.iterrows():
    ax.annotate(str(int(r['N'])),(r['H_positive_pct'],r['isolation_H_pct']),xytext=(4,3),textcoords='offset points',fontsize=8)
ax.set(xlabel='Candidatos com H > 0 (%)',ylabel='Isolados antes do resgate (%)',title='Positividade de H versus cobertura')
ax.grid(alpha=.2)
fig.savefig(OUTPUT_DIR/'positivity_vs_isolation.png',dpi=160)
plt.show()
```

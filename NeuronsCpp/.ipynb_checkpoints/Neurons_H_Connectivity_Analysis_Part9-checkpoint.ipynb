# Neurons — Parte IX · Anatomia de $H$: o somatório, o seno, o *shift* e os rankings

**Pergunta central:** o isolamento antes do resgate é causado pelo deslocamento $\pi/16$, pelo máximo do seno externo, pela distribuição do somatório $z$, pela exclusão de candidatos negativos ou pela concentração de preferências induzida pelo top-$N_{\max}$?

Nesta investigação **não introduzimos $r$, $\delta$ ou novos genes**: voltamos à expressão original. A fonte, os genes, o campo $A$, as camadas elegíveis, os papéis de input/output e $N_{\max}$ são mantidos entre condições. As medidas são **antes do resgate**.

$$z(p,q)=\frac{3\pi}{4C}\sum_{t=0}^{63}\sqrt{1+t}\,\cos(A_p U(x_t))\sin(A_qU(x_t)),\quad C=\sum_{t=0}^{63}\sqrt{1+t}.$$

$$H_s(p,q)=\sin(s+z(p,q)),\qquad s=\pi/16\text{ na versão original.}$$

### Roteiro — perguntas respondidas no mesmo notebook

1. **De onde vem o pico?** Histogramas comparáveis de $z$ e $H$, sua transformação exata e a saturação perto de $H=\pm1$.
2. **O *shift* é responsável?** Varredura de $s$ com os mesmos candidatos, sem mudar o somatório.
3. **É o limiar $H>0$ ou o ranking?** Controles com limiar fixo e com ranking linear por $z$.
4. **O máximo do seno gera empates ou rankings muito próximos?** Margens top-$N_{\max}$, empates exatos em `float32`, lados do pico e Jaccard pareado.
5. **Por que há destinos isolados?** Separar destinos sem oportunidades positivas de destinos positivos nunca selecionados.
6. **As origens disputam sistematicamente os mesmos destinos?** Sobreposição de rankings, popularidade dos destinos e relação com $A_q$.
7. **Quanto a ordem de filtragem importa?** Comparar top-$N_{\max}$ seguido de filtro (atual) versus filtrar antes de selecionar.
8. **Quanto da cobertura falta por causa da seleção determinística?** Controle aleatório *apenas entre candidatos com $H>0$*, sem modificar $H$.
9. **Qual próximo mecanismo merece investigação?** Diagnóstico e quadro final gerados automaticamente com evidência e ressalvas.

> **Atenção:** controles como selecionar por $z$, filtrar antes ou selecionar aleatoriamente são contrafactuais diagnósticos, **não sugestões para mudar o modelo**. Alterações em `float32` e desempates na fronteira top-$N_{\max}$ podem provocar pequenas diferenças em relação ao executável C++.

```python
import json, io, time, math, zipfile, warnings
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

PROFILE = 'quick'  # 'quick': 4 genomas (1.000 e 4.840); 'full': 11 genomas (até 17.640)
N_INPUTS, N_OUTPUTS = 10, 10
BASE_SEED = 20261025
SAMPLED_SOURCES = 64            # detalhes de ranking e amostras de z por genoma
RANK_PAIR_SOURCES = 24           # fontes da mesma camada para medir competição
STABILITY_MUTATIONS = 1        # mutações de 1 dígito por genoma (análise amostral)
STABILITY_ORIGINS = 16
OUTPUT_DIR = Path('neurons_h_part9_results')
OUTPUT_DIR.mkdir(exist_ok=True)
RESULT_ZIP = Path('Neurons_H_Part9_Results.zip')
INPUT_ZIP = Path('Neurons_H_Analysis_Results.zip')
assert INPUT_ZIP.is_file(), 'Coloque Neurons_H_Analysis_Results.zip na pasta do notebook.'
with zipfile.ZipFile(INPUT_ZIP) as archive:
    records = json.loads(archive.read('genomes.json'))
assert len(records)==11
if PROFILE=='quick':
    selected = [r for r in records if r['size'][0]==10][:2]+[r for r in records if r['size'][0]==22][:2]
elif PROFILE=='full':
    selected=records
else:
    raise ValueError('PROFILE deve ser quick ou full')
SHIFT0=float(np.pi/16)
SHIFTS=[-np.pi/8,-np.pi/16,0.,np.pi/16,np.pi/8,3*np.pi/16,np.pi/4]
SHIFT_MODES={f'shift_{i}':float(v) for i,v in enumerate(SHIFTS)}
BASELINE='shift_3'  # pi/16
EXTRAS=['sin_orig_zgt0','sin_zero_fixed_elig','rank_z_fixed_elig','rank_z_zgt0',
        'prefilter_original','random_positive_original']
MODES=list(SHIFT_MODES)+EXTRAS
MODE_LABELS={**{k:f's={v/np.pi:.4g}π' for k,v in SHIFT_MODES.items()},
    'sin_orig_zgt0':'H orig, z>0',
    'sin_zero_fixed_elig':'sin(z), eleg. original',
    'rank_z_fixed_elig':'ranking z, eleg. original',
    'rank_z_zgt0':'ranking z, z>0',
    'prefilter_original':'filtrar antes do top',
    'random_positive_original':'sorteio entre H>0'}
print('Perfil:',PROFILE,'| genomas:',len(selected),'| condições:',len(MODES))
print('Dimensões:',[r['size'] for r in selected])
```

## 1. Reproduzir exatamente $A$, os coeficientes genéticos e $N_{\max}$

Os mesmos genomas das Partes I–VIII e a mesma expansão $A$ do notebook anterior. A dependência genética em $H$ é agrupada nos quatro coeficientes $c_m$ **sem aproximação**:

$$z(p,q)=\sum_{m=1}^4 c_m\cos(mA_p)\sin(mA_q),\qquad
c_m=\frac{3\pi}{4C}\sum_{\{t:U(x_t)=m\}}\sqrt{1+t},\quad\sum_m c_m=\frac{3\pi}{4}. $$

```python
PI32 = np.float32(np.pi)
W8 = np.sqrt(np.arange(1, 9, dtype=np.float32))
W64 = np.sqrt(np.arange(1, 65, dtype=np.float32))

def random_genome(seed):
    rng = np.random.default_rng(seed)
    # O construtor C++ gera dígitos independentes e uniformes de 0 a 3.
    return {"connections": rng.integers(0,4,64,dtype=np.int8),
            "beta": rng.integers(0,4,8,dtype=np.int8),
            "max_connections": rng.integers(0,4,8,dtype=np.int8)}

def parse_genome(dictionary):
    lengths = {"connections": 64, "beta": 8, "max_connections": 8}
    ans = {}
    for name, count in lengths.items():
        value = dictionary[name]
        if isinstance(value, str):
            if len(value) != count or not set(value) <= set("0123"):
                raise ValueError(f"{name}: esperado {count} dígitos quaternários")
            ans[name] = np.array([int(ch) for ch in value], dtype=np.int8)
        else:
            ans[name] = np.asarray(value, dtype=np.int8)
            if ans[name].shape != (count,) or not np.isin(ans[name], [0,1,2,3]).all():
                raise ValueError(f"{name}: formato inválido")
    return ans

def genome_text(g):
    return {key: "".join(map(str, val)) for key, val in g.items()}

def beta_max(g, I, J):
    beta = np.float32(1.0 + np.sum(W8*g["beta"],dtype=np.float32)
                     / np.sum(3*W8,dtype=np.float32) * np.float32((I*J/16.)**(1/3)))
    maxconn = int(np.floor(np.sum(W8*np.power(beta, g["max_connections"].astype(np.float32)),dtype=np.float32)))
    return float(beta), maxconn

def h_coefficients(g):
    digits = g["connections"]
    return np.array([np.sum(W64[digits == d],dtype=np.float32) for d in range(4)],dtype=np.float32) * (np.float32(3)*PI32 / (np.float32(4)*np.sum(W64,dtype=np.float32)))

def make_A(g, I, J, K):
    # A[k,j,i], linearizado como i + I*j + I*J*k.
    coeff = (np.float32(2)/np.float32(3))*g["connections"].astype(np.float32)-np.float32(1)
    cx = np.cos(PI32*np.arange(4,dtype=np.float32)[:,None]*
                (np.arange(I,dtype=np.float32)[None,:]/np.float32(I-1))).astype(np.float32)
    cy = np.cos(PI32*np.arange(4,dtype=np.float32)[:,None]*
                (np.arange(J,dtype=np.float32)[None,:]/np.float32(J-1))).astype(np.float32)
    cz = np.cos(PI32*np.arange(4,dtype=np.float32)[:,None]*
                (np.arange(K,dtype=np.float32)[None,:]/np.float32(K-1))).astype(np.float32)
    field = np.zeros((K,J,I), dtype=np.float32)
    for r in range(4):
        for s in range(4):
            for q in range(4):
                field += coeff[q + 4*s + 16*r] * cx[r][None,None,:] * cy[s][None,:,None] * cz[q][:,None,None]
    return np.tanh(field / np.float32(8)).astype(np.float32).ravel()

def h_direct(a, b, c):
    z = np.float32(np.pi/16)
    for m in range(1, 5):
        z = z + c[m-1]*np.cos(np.float32(m)*a)*np.sin(np.float32(m)*b)
    return np.sin(z)

def make_fast_h(A, coeffs):
    harmonics = np.arange(1,5,dtype=np.float32)[:,None]
    ac = (coeffs[:,None]*np.cos(harmonics*A[None,:])).astype(np.float32)
    bs = np.sin(harmonics*A[None,:]).astype(np.float32)
    return ac, bs

def h_from_cached(ac, bs, src, target_indices):
    z = np.full(len(target_indices), np.float32(np.pi/16), dtype=np.float32)
    for m in range(4):
        z += ac[m, src] * bs[m, target_indices]
    return np.sin(z).astype(np.float32)




def z_from_cached(ac, bs, src, target_indices):
    z=np.zeros(len(target_indices),dtype=np.float32)
    for m in range(4):
        z += ac[m,src]*bs[m,target_indices]
    return z

def z_direct_64(a,b,genes):
    u=genes['connections'].astype(np.float32)+1.
    term=W64*np.cos(a*u)*np.sin(b*u)
    return (np.float32(3)*PI32/(np.float32(4)*np.sum(W64,dtype=np.float32))) * np.sum(term,dtype=np.float32)

_g=random_genome(214);_A=make_A(_g,5,5,4);_c=h_coefficients(_g)
_ac,_bs=make_fast_h(_A,_c)
_d=np.arange(1,len(_A),3,dtype=np.int32)
_z=z_from_cached(_ac,_bs,0,_d)
_z64=np.array([z_direct_64(_A[0],_A[q],_g) for q in _d],dtype=np.float32)
assert np.max(np.abs(_z-_z64)) < 1e-6
assert np.isclose(_c.sum(),3*np.pi/4,atol=1e-6)
assert np.all(np.abs(_z)<=3*np.pi/4+1e-6)
assert np.max(np.abs(np.sin(_z+SHIFT0)-h_from_cached(_ac,_bs,0,_d)))<2e-6
print('Soma direta de 64 genes, forma de 4 coeficientes e seno original: OK')
```

## 2. Geometria do seno e teste de identidade **antes dos dados**

Do limite $|\cos\sin|\le1$, segue $-3\pi/4\le z\le3\pi/4$. Com $s=\pi/16$:

$$H=\cos(z-z_*),\quad z_*=\frac{\pi}{2}-\frac{\pi}{16}=\frac{7\pi}{16}\approx1.37445.$$

Nesse intervalo, $H>0$ **se, e somente se,** $z>-\pi/16$. Entre os valores positivos, $H$ diminui estritamente com $|z-z_*|$, portanto **selecionar maiores $H$ equivale a escolher $z$ mais próximos de $z_*$**. Isto será verificado numericamente, inclusive quanto a empates em ponto flutuante.

O pico do histograma de $H$ pode ocorrer sem pico correspondente em $z$: para $H\to1$ e densidade $f_Z(z_*)>0$,

$$f_H(h)\sim\frac{\sqrt{2}\,f_Z(z_*)}{\sqrt{1-h}},\quad h\to1^-.$$

Trata-se de uma singularidade de mudança de variáveis, não prova, por si só, de preferências globais pelos mesmos destinos.

```python
zgrid=np.linspace(-3*np.pi/4,3*np.pi/4,10001)
hgrid=np.sin(zgrid+SHIFT0)
positive=hgrid>1e-10
assert np.array_equal(positive,(zgrid>-SHIFT0))
assert np.max(np.abs(hgrid-np.cos(zgrid-(np.pi/2-SHIFT0))))<1e-12
fig,axs=plt.subplots(1,2,figsize=(12,4),constrained_layout=True)
for s in [0,np.pi/16,np.pi/8,-np.pi/16]:
    axs[0].plot(zgrid,np.sin(zgrid+s),label=f's={s/np.pi:.3g}π')
axs[0].axvline(np.pi/2-SHIFT0,color='black',ls='--',alpha=.6,label='máximo original')
axs[0].axhline(0,color='gray',lw=.7);axs[0].legend(fontsize=8)
axs[0].set(xlabel='z',ylabel='H',title='Transformações do MESMO z')
axs[1].plot(zgrid,abs(zgrid-(np.pi/2-SHIFT0)),label='distância do pico')
axs[1].axvline(-SHIFT0,ls='--',color='black',alpha=.7)
axs[1].set(xlabel='z',ylabel='|z−z*|',title='Ranking original = menor distância (H>0)')
for ax in axs:ax.grid(alpha=.2)
fig.savefig(OUTPUT_DIR/'01_geometry_sine.png',dpi=150);plt.show()
```

## 3. Gerar os mesmos cérebros e os mesmos papéis input/output da Parte VIII

```python
def flags_by_rank(A, I, J, K, n_inputs, n_outputs):
    N = len(A)
    if n_inputs+n_outputs > N:
        raise ValueError("Número de inputs+outputs maior que número de neurônios")
    by_layer = []
    for k in range(K):
        ix = np.arange(k*I*J,(k+1)*I*J,dtype=np.int32)
        by_layer.append(ix[np.argsort(A[ix],kind="quicksort")])
    # rank 0 de cada camada; depois rank 1 de cada camada; ...
    permutation = np.stack(by_layer,axis=1).ravel()
    is_input = np.zeros(N,dtype=bool); is_output = np.zeros(N,dtype=bool)
    is_input[permutation[:n_inputs]] = True
    is_output[permutation[-n_outputs:]] = True
    return is_input, is_output

def gini_nonnegative(data):
    x = np.sort(np.asarray(data,dtype=np.float64)); n=len(x)
    if n == 0 or np.sum(x) == 0: return 0.
    return float((2*np.dot(np.arange(1,n+1),x)/(n*np.sum(x))) - (n+1)/n)

def metrics(incoming, outgoing, is_input, is_output):
    internal = ~(is_input | is_output)
    counts = {"internal":int(np.count_nonzero((incoming==0)&internal)),
              "output":int(np.count_nonzero((incoming==0)&is_output)),
              "input":int(np.count_nonzero((outgoing==0)&is_input))}
    N=len(incoming)
    eligible = ~is_input
    vals = incoming[eligible]
    tenpct = max(1,int(np.ceil(.1*len(vals))))
    total = np.sum(vals)
    top_share = float(np.partition(vals,-tenpct)[-tenpct:].sum()/total) if total else np.nan
    return {"isolated_total":sum(counts.values()),"isolation_pct":100.*sum(counts.values())/N,
            "isolated_internal":counts["internal"],"isolated_output":counts["output"],"isolated_input":counts["input"],
            "mean_incoming":float(np.mean(vals)),"median_incoming":float(np.median(vals)),
            "gini_incoming":gini_nonnegative(vals),"top_10pct_incoming_share":top_share,
            "zero_incoming_eligible_pct":100.*np.mean(vals==0),"total_valid_connections":int(np.sum(vals))}



cases=[]
for i,rec in enumerate(selected):
    size=tuple(rec['size']);genes=parse_genome(rec['genes']);A=make_A(genes,*size)
    roles=flags_by_rank(A,*size,N_INPUTS,N_OUTPUTS)
    beta,nmax=beta_max(genes,size[0],size[1]); coeff=h_coefficients(genes)
    ac,bs=make_fast_h(A,coeff)
    cases.append({'case':f"{size[0]}x{size[1]}x{size[2]}_g{i+1:02d}",
       'genome_seed':rec['seed'],'size':size,'genes':genes,'A':A,
       'roles':roles,'ac':ac,'bs':bs,'nmax':nmax,'beta':beta,'coeff':coeff})
assert all(np.isclose(c['coeff'].sum(),3*np.pi/4,atol=2e-6) for c in cases)
print([(c['case'],c['nmax']) for c in cases])
```

## 4. Histogramas de $z$ e $H$ nas **mesmas arestas candidatas**

Amostramos origens elegíveis de cada genoma, com candidatos de $|k_q-k_p|\le2$ e $p\ne q$. Armazenamos pares $(z,H)$ e separamos origens por sinal de $A_p$. Os histogramas são de **pares possíveis**, antes do top e antes da filtragem de destinos. A distribuição de arestas efetivamente escolhidas será mostrada posteriormente.

```python
SAMPLES_PER_ORIGIN=2500
sample_rows=[]; sampled_values={};rng=np.random.default_rng(BASE_SEED+101)
for case in cases:
    I,J,K=case['size'];L=I*J
    inputs,outputs=case['roles']
    origins=np.flatnonzero(~outputs)
    src_sample=np.sort(rng.choice(origins,size=min(SAMPLED_SOURCES,len(origins)),replace=False))
    allz=[]; alla=[]
    for src in src_sample:
        dest=np.arange(max(0,int(src//L)-2)*L,min(K,int(src//L)+3)*L,dtype=np.int32)
        dest=dest[dest!=src]
        if len(dest)>SAMPLES_PER_ORIGIN:
            dest=rng.choice(dest,size=SAMPLES_PER_ORIGIN,replace=False)
        z=z_from_cached(case['ac'],case['bs'],int(src),dest)
        allz.append(z);alla.append(np.full(len(z),case['A'][src],dtype=np.float32))
    z=np.concatenate(allz);srcA=np.concatenate(alla);h=np.sin(z+SHIFT0)
    zstar=np.pi/2-SHIFT0
    sampled_values[case['case']]={'z':z,'H':h,'srcA':srcA}
    sample_rows.append({'case':case['case'],'neurons':len(case['A']),'sampled_pairs':len(z),
       'z_mean':float(z.mean()),'z_sd':float(z.std()),'z_q01':float(np.quantile(z,.01)),
       'z_q50':float(np.median(z)),'z_q99':float(np.quantile(z,.99)),
       'positive_H_pct':100*float(np.mean(h>0)),
       'z_above_peak_pct':100*float(np.mean(z>zstar)),
       'H_gt_099_pct':100*float(np.mean(h>.99)),
       'H_gt_0999_pct':100*float(np.mean(h>.999)),
       'H_lt_m099_pct':100*float(np.mean(h<-.99)),
       'z_peak_pm01_pct':100*float(np.mean(abs(z-zstar)<.1))})
zsamples=pd.DataFrame(sample_rows)
zsamples.to_csv(OUTPUT_DIR/'sampled_z_H_distribution.csv',index=False)
display(zsamples.round(3))
example=sampled_values[cases[-1]['case']]
fig,axs=plt.subplots(2,2,figsize=(13,8),constrained_layout=True)
axs[0,0].hist(example['z'],bins=80,density=True)
axs[0,0].axvline(np.pi/2-SHIFT0,color='red',ls='--',label='z*')
axs[0,0].set(xlabel='z',ylabel='densidade',title='Somatório antes do seno');axs[0,0].legend()
axs[0,1].hist(example['H'],bins=80,range=(-1,1),density=True)
axs[0,1].set(xlabel='H original',ylabel='densidade',title='Mesmos pares transformados')
axs[1,0].scatter(example['z'][::max(1,len(example['z'])//3000)],example['H'][::max(1,len(example['H'])//3000)],s=2,alpha=.25)
axs[1,0].set(xlabel='z',ylabel='H',title='Curva exata H(z) e amostras reais')
for sign,label in [(example['srcA']<0,'A(origem)<0'),(example['srcA']>=0,'A(origem)≥0')]:
    axs[1,1].hist(example['z'][sign],bins=70,density=True,histtype='step',lw=1.5,label=label)
axs[1,1].set(xlabel='z',ylabel='densidade',title='Distribuição condicionada pela origem');axs[1,1].legend()
for ax in axs.flat:ax.grid(alpha=.15)
fig.savefig(OUTPUT_DIR/'02_z_and_H_distributions.png',dpi=150);plt.show()
```

## 5. Experimento fatorial: mudar pico, limiar e ordenação **separadamente**

Os sete *shifts* usam $H_s=\sin(z+s)$ com sua própria elegibilidade $H_s>0$.

Controles que isolam os mecanismos:

| Condição | Pontuação/ranking | Elegibilidade | Interpretação |
|---|---|---|---|
| `sin_orig_zgt0` | seno original | $z>0$ | muda só o limiar, mantém a ordem original |
| `sin_zero_fixed_elig` | $\sin z$ | $z>-\pi/16$ | muda o pico/ranking, mantém o limiar original |
| `rank_z_fixed_elig` | $z$ (decrescente) | $z>-\pi/16$ | ranking monotônico, limiar original |
| `rank_z_zgt0` | $z$ (decrescente) | $z>0$ | ranking monotônico e limiar zero |
| `prefilter_original` | seno original | $H>0$, **filtra antes** do top | efeito da ordem dos filtros |
| `random_positive_original` | escolha uniforme | $H>0$ | limite experimental para o top determinístico |

O controle `prefilter_original` não muda valores de $H$, somente a ordem de operações.

```python
def choose_destinations(z,dest,src,nmax,inputs,outputs,mode,rng):
    if mode in SHIFT_MODES:
        shift=SHIFT_MODES[mode]
        scores=np.sin(z+np.float32(shift)).astype(np.float32)
        eligible=scores>0
    elif mode in ['sin_orig_zgt0','sin_zero_fixed_elig','rank_z_fixed_elig',
                  'rank_z_zgt0','prefilter_original','random_positive_original']:
        if mode=='sin_zero_fixed_elig':scores=np.sin(z).astype(np.float32)
        elif mode in ('rank_z_fixed_elig','rank_z_zgt0'):scores=z
        else:scores=np.sin(z+np.float32(SHIFT0)).astype(np.float32)
        if mode in ('sin_orig_zgt0','rank_z_zgt0'):eligible=z>0
        else:eligible=z>-np.float32(SHIFT0)
    else:raise ValueError(mode)
    legal=(~inputs[dest]) & (~(inputs[src] & outputs[dest]))
    if mode=='prefilter_original':eligible=eligible&legal
    eligible_ids=np.flatnonzero(eligible)
    take=min(nmax,len(eligible_ids))
    if take==0:
        chosen=np.empty(0,dtype=np.int32)
    elif mode=='random_positive_original':
        chosen=rng.choice(eligible_ids,size=take,replace=False)
    elif take<len(eligible_ids):
        chosen=eligible_ids[np.argpartition(scores[eligible_ids],-take)[-take:]]
    else:chosen=eligible_ids
    pre=dest[chosen];keep=pre[legal[chosen]]
    return {'pre':pre,'keep':keep,'legal':legal,'eligible':eligible,
       'scores':scores,'chosen_indices':chosen,'eligible_indices':eligible_ids}

# Teste exato (exceto desempates em float32): ranking original = proximidade de z*
_testz=np.linspace(-3*np.pi/4,3*np.pi/4,120003,dtype=np.float64)
_p=_testz>-SHIFT0;h=np.sin(_testz[_p]+SHIFT0)
closeness=np.abs(_testz[_p]-(np.pi/2-SHIFT0))
assert np.array_equal(np.argsort(-h),np.argsort(closeness))
print('Identidade ranking no domínio positivo: OK')
```

## 6. Execução: seleção por origem e **causas do isolamento**

Para cada origem $p$, calculamos o vetor de $z(p,q)$ **uma vez**, e reusamos exatamente esses valores em todas as condições. Essa estratégia minimiza diferenças numéricas e acelera o experimento.

Registramos por destino as oportunidades de receber $H>0$ de alguma origem **legal** (`positive_opportunities`) e as entradas de fato selecionadas. Um destino com oportunidades positivas mas grau zero foi **preterido pelo ranking**; um destino com zero oportunidades positivas não poderia ser resgatado só modificando a seleção top entre positivos.

```python
def run_case(case):
    I,J,K=case['size'];N=len(case['A']);L=I*J
    inputs,outputs=case['roles'];A=case['A'];nmax=case['nmax']
    all_sources=np.flatnonzero(~outputs)
    windows=[np.arange(max(0,k-2)*L,min(K,k+3)*L,dtype=np.int32) for k in range(K)]
    rng_orig=np.random.default_rng(np.random.SeedSequence([BASE_SEED,case['genome_seed'],101]))
    sample=np.sort(rng_orig.choice(all_sources,size=min(SAMPLED_SOURCES,len(all_sources)),replace=False))
    sample_mask=np.zeros(N,dtype=bool);sample_mask[sample]=True
    rng_random=np.random.default_rng(np.random.SeedSequence([BASE_SEED,case['genome_seed'],777]))
    indeg={m:np.zeros(N,dtype=np.int32) for m in MODES}
    predeg={m:np.zeros(N,dtype=np.int32) for m in MODES}
    outdeg={m:np.zeros(N,dtype=np.int32) for m in MODES}
    opportunities=np.zeros(N,dtype=np.int32);possible=np.zeros(N,dtype=np.int32)
    counters={m:{'positive':0,'candidates':0,'pre':0,'removed':0,'sum_h':0.,'sum_score':0.,'score_count':0,
                 'ties_at_cutoff':0,'near_margin':0,'margin_count':0,
                 'selected_above_zpeak':0,'selected_total':0,'positive_above_zpeak':0}
              for m in MODES}
    origin_records=[];edge_samples={m:{} for m in MODES}
    peak=np.pi/2-SHIFT0
    for src in all_sources:
        dest=windows[int(src//L)];dest=dest[dest!=src]
        z=z_from_cached(case['ac'],case['bs'],int(src),dest)
        legal=(~inputs[dest]) & (~(inputs[src]&outputs[dest]))
        possible[dest[legal]]+=1
        orig_pos=z>-np.float32(SHIFT0)
        opportunities[dest[orig_pos&legal]]+=1
        for mode in MODES:
            result=choose_destinations(z,dest,int(src),nmax,inputs,outputs,mode,rng_random)
            pre=result['pre'];keep=result['keep'];chosen=result['chosen_indices'];eligible=result['eligible_indices'];scores=result['scores']
            indeg[mode][keep]+=1;predeg[mode][pre]+=1;outdeg[mode][src]+=len(keep)
            d=counters[mode]
            d['candidates']+=len(dest);d['positive']+=len(eligible)
            d['pre']+=len(pre);d['removed']+=len(pre)-len(keep)
            d['selected_total']+=len(keep)
            if len(chosen):
                d['sum_h']+=float(np.sin(z[chosen[result['legal'][chosen]]]+np.float32(SHIFT_MODES.get(mode,SHIFT0))).sum(dtype=np.float64)) if mode in SHIFT_MODES or mode not in ('rank_z_fixed_elig','rank_z_zgt0','sin_zero_fixed_elig') else float(np.sin(z[chosen[result['legal'][chosen]]]+np.float32(SHIFT0)).sum(dtype=np.float64))
                d['sum_score']+=float(scores[chosen[result['legal'][chosen]]].sum(dtype=np.float64));d['score_count']+=len(keep)
                d['selected_above_zpeak']+=int(np.sum(z[chosen[result['legal'][chosen]]]>peak))
            d['positive_above_zpeak']+=int(np.sum((z>peak)&result['eligible']))
            # Diagnose tight margin and ties in original-scale scoring; sampled origins only
            if sample_mask[src] and mode not in ('random_positive_original',):
                if len(chosen)>0 and len(eligible)>len(chosen):
                    boundary=np.min(scores[chosen])
                    margin=float(boundary-np.max(scores[eligible[scores[eligible]<boundary]])) if np.any(scores[eligible]<boundary) else 0.
                    ties=int(np.sum(scores[eligible]==boundary))
                    d['ties_at_cutoff']+=int(ties>1)
                    d['near_margin']+=int(margin<1e-6)
                    d['margin_count']+=1
                    origin_records.append({'case':case['case'],'neurons':N,'src':int(src),'src_A':float(A[src]),
                        'src_layer':int(src//L),'mode':mode,
                        'n_eligible':len(eligible),'picked_pre':len(pre),'picked_valid':len(keep),
                        'cutoff_score':float(boundary),'margin_next_distinct':margin,
                        'ties_at_cutoff':ties,'above_peak_candidates_pct':100*float(np.mean(z[eligible]>peak)),
                        'chosen_above_peak_pct':100*float(np.mean(z[chosen]>peak))})
            if sample_mask[src]:edge_samples[mode][int(src)]=set(keep.tolist())
    rows=[]
    for mode in MODES:
        m=metrics(indeg[mode],outdeg[mode],inputs,outputs);c=counters[mode]
        prem=metrics(predeg[mode],outdeg[mode],inputs,outputs)
        rows.append({'case':case['case'],'neurons':N,'size_x':I,'size_y':J,'size_z':K,
              'genome_seed':case['genome_seed'],'mode':mode,'shift':SHIFT_MODES.get(mode,np.nan),
              'Nmax':nmax,'beta':case['beta'],
              **m,'pre_filter_isolation_pct':prem['zero_incoming_eligible_pct'],
              'positive_pct':100*c['positive']/c['candidates'],
              'discarded_by_filter_pct':100*c['removed']/c['pre'] if c['pre'] else np.nan,
              'mean_selected_score':c['sum_score']/c['score_count'] if c['score_count'] else np.nan,
              'mean_selected_H':c['sum_h']/c['score_count'] if c['score_count'] else np.nan,
              'chosen_above_original_peak_pct':100*c['selected_above_zpeak']/c['selected_total'] if c['selected_total'] else np.nan,
              'sample_cutoff_ties_sources':c['ties_at_cutoff'],
              'sample_tiny_margin_pct':100*c['near_margin']/c['margin_count'] if c['margin_count'] else np.nan})
    original_indeg=indeg[BASELINE]
    eligible=~inputs
    breakdown={'case':case['case'],'neurons':N,
          'eligible_destinations':int(np.sum(eligible)),
          'destinations_no_legal_source':int(np.sum(eligible&(possible==0))),
          'destinations_zero_positive_opportunity':int(np.sum(eligible&(opportunities==0))),
          'destinations_positive_but_unselected':int(np.sum(eligible&(opportunities>0)&(original_indeg==0))),
          'destinations_selected':int(np.sum(eligible&(original_indeg>0))),
          'output_zero_positive_opportunity':int(np.sum(outputs&(opportunities==0))),
          'output_positive_but_unselected':int(np.sum(outputs&(opportunities>0)&(original_indeg==0))),
          'opportunities_p10':float(np.quantile(opportunities[eligible],.1)),
          'opportunities_median':float(np.median(opportunities[eligible])),
          'opportunities_p90':float(np.quantile(opportunities[eligible],.9))}
    assert breakdown['destinations_zero_positive_opportunity']+breakdown['destinations_positive_but_unselected']+breakdown['destinations_selected']==breakdown['eligible_destinations']
    return rows,origin_records,breakdown,indeg,opportunities,possible,edge_samples

# Small dry-run for selection and breakdown: no dependence on file-based results
mini=cases[0]
mini_sm={'case':'mini','size':(5,5,4),'A':make_A(mini['genes'],5,5,4),
         'roles':None,'nmax':mini['nmax'],'beta':mini['beta'],'genome_seed':mini['genome_seed']}
mini_sm['roles']=flags_by_rank(mini_sm['A'],5,5,4,N_INPUTS,N_OUTPUTS)
mini_sm['ac'],mini_sm['bs']=make_fast_h(mini_sm['A'],mini['coeff'])
_t=run_case(mini_sm)
assert len(_t[0])==len(MODES)
assert _t[2]['destinations_zero_positive_opportunity']>=0
print('Execução reduzida e partição das causas de isolamento: OK')
```

## 7. Executar os genomas e salvar as métricas comparáveis

```python
rows=[];origin_rows=[];breakdowns=[]; profiles={};opps={};possible_by_case={};edges_by_case={}
start=time.perf_counter()
for c in cases:
    t0=time.perf_counter()
    r,ors,b,indeg,opportunities,possible,es=run_case(c)
    rows+=r;origin_rows+=ors;breakdowns.append(b)
    profiles[c['case']]=indeg;opps[c['case']]=opportunities;possible_by_case[c['case']]=possible
    edges_by_case[c['case']]=es
    base=next(v for v in r if v['mode']==BASELINE)
    raw=next(v for v in r if v['mode']=='rank_z_fixed_elig')
    print(f"{c['case']}: original={base['isolation_pct']:.2f}% | ranking z={raw['isolation_pct']:.2f}% | t={time.perf_counter()-t0:.1f}s",flush=True)
print('Tempo total:',round(time.perf_counter()-start,1),'s')
df=pd.DataFrame(rows);src_df=pd.DataFrame(origin_rows);breakdown_df=pd.DataFrame(breakdowns)
df.to_csv(OUTPUT_DIR/'per_genome_shift_and_controls.csv',index=False)
src_df.to_csv(OUTPUT_DIR/'source_ranking_margins.csv',index=False)
breakdown_df.to_csv(OUTPUT_DIR/'why_isolated_destinations.csv',index=False)
summary=df.groupby('mode',as_index=False).agg(
    n_genomes=('case','nunique'),isolation_pct=('isolation_pct','mean'),
    gini=('gini_incoming','mean'),top10=('top_10pct_incoming_share','mean'),
    output_isolated=('isolated_output','sum'),
    edge_count=('total_valid_connections','sum'),
    pre_filter_zero_pct=('pre_filter_isolation_pct','mean'),
    filter_removed_pct=('discarded_by_filter_pct','mean'),
    positive_pct=('positive_pct','mean'),
    chosen_above_original_peak_pct=('chosen_above_original_peak_pct','mean'),
    sample_tiny_margin_pct=('sample_tiny_margin_pct','mean'))
summary['name']=summary['mode'].map(MODE_LABELS)
summary=summary.sort_values('isolation_pct')
summary.to_csv(OUTPUT_DIR/'summary_modes.csv',index=False)
print('Comparação agregada — peso igual por genoma (sem normalizar pelo número de neurônios):')
display(summary.round(4))
print('Causas do isolamento no H original:')
display(breakdown_df)
# Reprodução independente dos resultados de referência da Parte I, por semente.
with zipfile.ZipFile(INPUT_ZIP) as archive:
    reference=pd.read_csv(io.BytesIO(archive.read('per_brain_metrics.csv')))
observed=df[df['mode']==BASELINE].merge(reference,left_on='genome_seed',right_on='seed',validate='one_to_one')
max_iso_gap=float(abs(observed['isolation_pct']-observed['isolation_H_pct']).max())
edge_discrepancy=abs(observed['total_valid_connections']-observed['valid_edges_H'])
max_edge_gap=int(edge_discrepancy.max())
print(f'Reprodução da Parte I — desvio máximo de isolamento: {max_iso_gap:.5f} p.p.; diferenças máximas de conexões: {max_edge_gap}')
assert max_iso_gap < .3, 'Diferença excessiva em relação à Parte I; verifique genes, z e top-Nmax.'
assert (edge_discrepancy <= np.maximum(10,observed['valid_edges_H']*.005)).all(), 'Divergência considerável de arestas em relação à Parte I.'
```

## 11. Pergunta seguinte D — muitos empates perto de 1? Perfis top-$N_{\max}$

O seno comprime diferenças de $z$ perto de $z_*$. A seleção pode ser afetada de duas maneiras diferentes: (i) muitos candidatos com valores *próximos*, tornando o ranking sensível a pequenas mudanças, ou (ii) candidatos com valores **exatamente empatados após arredondamento `float32`**, introduzindo dependência da política de desempate de `argpartition`. Não confundimos os dois fenômenos.

```python
rank_diagnosis=src_df.groupby('mode',as_index=False).agg(
    sampled_origins=('src','size'),
    median_margin=('margin_next_distinct','median'),
    p10_margin=('margin_next_distinct',lambda s:float(s.quantile(.1))),
    ties_at_cutoff_source_pct=('ties_at_cutoff',lambda s:100*float((s>1).mean())),
    tiny_margin_source_pct=('margin_next_distinct',lambda s:100*float((s<1e-6).mean())),
    chosen_above_peak_pct=('chosen_above_peak_pct','mean'))
rank_diagnosis.to_csv(OUTPUT_DIR/'ranking_precision_diagnostics.csv',index=False)
display(rank_diagnosis.round(7))
fig,axs=plt.subplots(1,2,figsize=(12,4.5),constrained_layout=True)
for mode in [BASELINE,'shift_2','shift_4','rank_z_fixed_elig']:
    s=src_df[src_df['mode']==mode]['margin_next_distinct']
    s=s[s>0]
    if len(s):axs[0].hist(np.log10(s),bins=45,histtype='step',density=True,label=MODE_LABELS[mode])
axs[0].set(xlabel='log10(margem para próximo escore distinto)',ylabel='densidade',title='Precisão numérica do cutoff');axs[0].legend(fontsize=8)
for mode in [BASELINE,'shift_2','shift_4','rank_z_fixed_elig']:
    s=src_df[src_df['mode']==mode]
    axs[1].hist(s['chosen_above_peak_pct'],bins=np.linspace(0,100,21),histtype='step',density=True,label=MODE_LABELS[mode])
axs[1].set(xlabel='% dos selecionados com z>z* (pico original)',ylabel='densidade',title='Quais lados do pico foram selecionados?');axs[1].legend(fontsize=8)
fig.savefig(OUTPUT_DIR/'07_cutoff_margins_and_peak_sides.png',dpi=150);plt.show()
```

## 12. Pergunta seguinte E — origens diferentes escolhem destinos repetidos?

Usamos conjuntos de destinos **já selecionados e legalmente válidos** de origens amostradas. Comparamos origens na mesma camada para tornar suas janelas candidatas idênticas (exceto as autoconexões). A sobreposição Jaccard desses conjuntos mede o compartilhamento de destinos, **não** a correlação de valores de $H$ entre todos os candidatos.

Para evitar interpretar sobreposição inevitável como preferência compartilhada, calculamos também o valor esperado do Jaccard para conjuntos aleatórios do mesmo tamanho, com a aproximação $E[|S_1\cap S_2|]\approx |S_1||S_2|/M$ e $M$ sendo a quantidade de destinos legais comuns. Essa aproximação não impõe o limiar de positividade e serve apenas como referência grosseira.

```python
comp_rows=[];jaccard_between_modes=[]
for case in cases:
    I,J,K=case['size'];L=I*J;inputs,outputs=case['roles'];key=case['case'];ec=edges_by_case[key]
    rng=np.random.default_rng(np.random.SeedSequence([BASE_SEED,case['genome_seed'],121]))
    sources=sorted(ec[BASELINE]);
    for mode in [BASELINE,'shift_2','shift_4','rank_z_fixed_elig','random_positive_original']:
        values=[];references=[]
        for layer in range(K):
            ss=[x for x in sources if x//L==layer]
            if len(ss)<2:continue
            pairs=[tuple(rng.choice(ss,size=2,replace=False)) for _ in range(min(RANK_PAIR_SOURCES,2*len(ss)))]
            legal_window=np.arange(max(0,layer-2)*L,min(K,layer+3)*L)
            M=int(np.sum(~inputs[legal_window]))
            for a,b in pairs:
                s1=ec[mode][int(a)];s2=ec[mode][int(b)]
                if not s1 and not s2:continue
                j=len(s1&s2)/len(s1|s2)
                overlap_expected=(len(s1)*len(s2))/max(M,1)
                union_expected=len(s1)+len(s2)-overlap_expected
                baseline_exp=overlap_expected/union_expected if union_expected>0 else 1.
                values.append(j);references.append(baseline_exp)
        if values:
            comp_rows.append({'case':key,'mode':mode,'sampled_pairs':len(values),
                 'mean_jaccard_same_layer':float(np.mean(values)),
                 'random_sets_reference':float(np.mean(references)),
                 'jaccard_excess':float(np.mean(values)-np.mean(references))})
    for mode in [m for m in MODES if m!=BASELINE]:
        vals=[]
        for src in sources:
            s0=ec[BASELINE][src];s1=ec[mode][src]
            vals.append(len(s0&s1)/len(s0|s1) if s0|s1 else 1.)
        jaccard_between_modes.append({'case':key,'mode':mode,'mean_jaccard_with_original':float(np.mean(vals))})
comp=pd.DataFrame(comp_rows);pairj=pd.DataFrame(jaccard_between_modes)
comp.to_csv(OUTPUT_DIR/'shared_destination_preferences.csv',index=False)
pairj.to_csv(OUTPUT_DIR/'selected_edges_jaccard_vs_original.csv',index=False)
display(comp.groupby('mode')[['mean_jaccard_same_layer','random_sets_reference','jaccard_excess']].mean().round(4))
display(pairj.groupby('mode')['mean_jaccard_with_original'].mean().sort_values().round(4))
fig,axs=plt.subplots(1,2,figsize=(12,4.5),constrained_layout=True)
g=comp.groupby('mode')[['mean_jaccard_same_layer','random_sets_reference']].mean()
ind=np.arange(len(g));axs[0].bar(ind-.18,g['mean_jaccard_same_layer'],.36,label='medido');axs[0].bar(ind+.18,g['random_sets_reference'],.36,label='conjuntos aleatórios (referência)')
axs[0].set_xticks(ind,[MODE_LABELS[m] for m in g.index],rotation=50,ha='right');axs[0].legend(fontsize=8)
h=pairj.groupby('mode')['mean_jaccard_with_original'].mean().sort_values()
axs[1].bar(np.arange(len(h)),h);axs[1].set_xticks(np.arange(len(h)),[MODE_LABELS[m] for m in h.index],rotation=65,ha='right')
axs[1].set(ylabel='Jaccard médio',title='Quanto cada alternativa muda as escolhas?')
fig.savefig(OUTPUT_DIR/'08_shared_destinations_and_rank_changes.png',dpi=150);plt.show()
```

## 12A. Os destinos selecionados estão concentrados ao redor de $z_*$?

Um histograma de $H$ contendo muitos valores perto de 1 **não implica necessariamente** que os destinos escolhidos sejam todos próximos do máximo. Este controle sobrepõe a distribuição de $z$ de pares candidatos amostrados à distribuição de $z$ efetivamente selecionada pelo top-$N_{\max}$, nas **mesmas origens**. Para tornar a comparação justa, a amostra de candidatos é calculada nessas mesmas fontes e seus destinos legais.


```python
case=cases[-1];I,J,K=case['size'];L=I*J;inputs,outputs=case['roles']
origins=src_df[(src_df['case']==case['case']) & (src_df['mode']==BASELINE)]['src'].unique().astype(int)
zcandidates=[];zselected=[];chosen_scores=[]
for src in origins:
    ds=np.arange(max(0,(src//L)-2)*L,min(K,(src//L)+3)*L,dtype=np.int32)
    ds=ds[ds!=src]
    z=z_from_cached(case['ac'],case['bs'],src,ds)
    legal=(~inputs[ds]) & (~(inputs[src]&outputs[ds]))
    zcandidates.append(z[legal])
    picked=choose_destinations(z,ds,src,case['nmax'],inputs,outputs,BASELINE,np.random.default_rng(BASE_SEED))
    chosen=picked['chosen_indices']
    valid=chosen[picked['legal'][chosen]]
    zselected.append(z[valid])
    chosen_scores.append(picked['scores'][valid])
zc=np.concatenate(zcandidates);zs=np.concatenate(zselected);hs=np.concatenate(chosen_scores)
peak=np.pi/2-SHIFT0
print('Candidatos legais com z próximo ao pico ±0.1:',f'{np.mean(abs(zc-peak)<.1):.2%}')
print('Selecionados com z próximo ao pico ±0.1:',f'{np.mean(abs(zs-peak)<.1):.2%}')
fig,axs=plt.subplots(1,2,figsize=(12,4),constrained_layout=True)
axs[0].hist(zc,bins=80,density=True,alpha=.55,label='candidatos legais')
axs[0].hist(zs,bins=80,density=True,histtype='step',lw=1.5,label='selecionados')
axs[0].axvline(peak,ls='--',color='black',alpha=.7,label='z*')
axs[0].legend();axs[0].set(xlabel='z',ylabel='densidade',title='Mudança de distribuição depois do top')
axs[1].hist(hs,bins=60,range=(-1,1),density=True)
axs[1].set(xlabel='H escolhido',ylabel='densidade',title='Afinidade das conexões selecionadas')
fig.savefig(OUTPUT_DIR/'09_candidate_vs_selected_z.png',dpi=150);plt.show()
```

## 12B. Outra causa estrutural — a simetria em $A_p$

**Identidade exata:** $z(a,b)=z(-a,b)$, pois a origem entra somente por $\cos(ma)$, uma função par. Logo, origens com o mesmo $|A_p|$ têm perfis de afinidade idênticos para os mesmos destinos candidatos, **independentemente do shift**. Isso continua verdadeiro para todas as variantes da Parte IX, exceto as que fazem sorteio no ranking.

Vamos testar a identidade numericamente e relacionar a diferença $||A_{p_1}|-|A_{p_2}||$ à sobreposição dos destinos selecionados de origens na mesma camada. Essa análise verifica uma possível fonte de rankings compartilhados que simplesmente variar $s$ não pode remover.


```python
# Verificação independente de paridade da função original
rng=np.random.default_rng(BASE_SEED+404)
a=rng.uniform(-1,1,3000);b=rng.uniform(-1,1,3000)
cc=cases[0]['coeff'].astype(np.float64)
def z_scalar(a,b,c):
    out=np.zeros_like(a,dtype=np.float64)
    for m in range(1,5):out+=c[m-1]*np.cos(m*a)*np.sin(m*b)
    return out
assert np.allclose(z_scalar(a,b,cc),z_scalar(-a,b,cc),rtol=1e-12,atol=1e-12)
rows_sign=[]
for case in cases:
    key=case['case'];A=case['A'];I,J,K=case['size'];L=I*J
    es=edges_by_case[key][BASELINE]
    srcs=sorted(es)
    for layer in range(K):
        S=[src for src in srcs if src//L==layer]
        if len(S)<3:continue
        for src in S:
            others=np.array([q for q in S if q!=src],dtype=np.int32)
            da=np.abs(np.abs(A[src])-np.abs(A[others]))
            for group,q in [('mais próximo em |A|',int(others[np.argmin(da)])),
                            ('mais distante em |A|',int(others[np.argmax(da)]))]:
                x=es[src];y=es[q]
                Jacc=len(x&y)/len(x|y) if x|y else 1.
                rows_sign.append({'case':key,'group':group,'src':src,'other_src':q,
                    'delta_absA':float(abs(abs(A[src])-abs(A[q]))),
                    'opposite_sign':bool(np.sign(A[src])!=np.sign(A[q])),
                    'jaccard':Jacc})
sign_df=pd.DataFrame(rows_sign)
sign_df.to_csv(OUTPUT_DIR/'absA_similarity_rank_overlap.csv',index=False)
print('Identidade z(a,b)=z(-a,b): validada para 3000 pares numéricos.')
display(sign_df.groupby('group')[['delta_absA','jaccard']].agg(['mean','median']).round(4))
print('Correlação descritiva delta|A| vs Jaccard:',sign_df[['delta_absA','jaccard']].corr().iloc[0,1])
fig,ax=plt.subplots(figsize=(8,4.5),constrained_layout=True)
for group,g in sign_df.groupby('group'):
    ax.scatter(g['delta_absA'],g['jaccard'],s=9,alpha=.2,label=group)
ax.set(xlabel='diferença entre |A| das duas origens',ylabel='Jaccard dos destinos escolhidos',
       title='Ranking compartilhado e simetria da origem')
ax.legend();fig.savefig(OUTPUT_DIR/'10_source_absA_and_overlap.png',dpi=150);plt.show()
```

## 13. Pergunta seguinte F — o filtro após o top está produzindo parte da perda?

Uma origem pode escolher destinos proibidos para depois descartá-los. O controle `prefilter_original` seleciona a partir dos destinos legais desde o início. Não se trata de uma alteração de $H$; o diagnóstico verifica se uma restrição *administrativa* amplifica a falta de cobertura.

```python
base=df[df['mode']==BASELINE].set_index('case')
pre=df[df['mode']=='prefilter_original'].set_index('case')
filter_test=pd.DataFrame({'neurons':base['neurons'],
      'current_removed_pct':base['discarded_by_filter_pct'],
      'current_isolation':base['isolation_pct'],
      'prefilter_isolation':pre['isolation_pct'],
      'delta_isolation_pp':pre['isolation_pct']-base['isolation_pct'],
      'extra_valid_edges':pre['total_valid_connections']-base['total_valid_connections'],
      'current_prefilter_isolation':base['pre_filter_isolation_pct']})
filter_test.to_csv(OUTPUT_DIR/'filter_order_diagnostic.csv')
display(filter_test.round(4))
```

## 14. Pergunta seguinte G — pequenas mutações mudam rankings por causa do pico?

Controle amostral: mutar um dígito do gene `connections` por $\pm1$, recalcular $A$ e $z$, manter papéis de entrada/saída fixos e comparar os conjuntos top-$N_{\max}$ de algumas origens. Comparamos o seno original, seno sem shift, shift maior e ranking monotônico com elegibilidade original. Trata-se de estabilidade de ranking amostrada, não de cognição ou de estabilidade fenotípica completa.

```python
mut_modes=[BASELINE,'shift_2','shift_4','rank_z_fixed_elig']
mutation_rows=[];mut_rng=np.random.default_rng(BASE_SEED+555)
for case in cases:
    I,J,K=case['size'];L=I*J;inputs,outputs=case['roles'];orig=np.flatnonzero(~outputs)
    chosen_origins=mut_rng.choice(orig,size=min(len(orig),STABILITY_ORIGINS),replace=False)
    windows=[np.arange(max(0,k-2)*L,min(K,k+3)*L,dtype=np.int32) for k in range(K)]
    for mut in range(STABILITY_MUTATIONS):
        mutant={k:v.copy() for k,v in case['genes'].items()}
        idx=int(mut_rng.integers(0,len(mutant['connections'])))
        old=int(mutant['connections'][idx]);step=1 if old==0 else (-1 if old==3 else int(mut_rng.choice([-1,1])))
        mutant['connections'][idx]=old+step
        Anew=make_A(mutant,I,J,K);cnew=h_coefficients(mutant)
        ac1,bs1=make_fast_h(Anew,cnew)
        for mode in mut_modes:
            js=[];delta_h=[]
            for src in chosen_origins:
                ds=windows[int(src//L)];ds=ds[ds!=src]
                z0=z_from_cached(case['ac'],case['bs'],int(src),ds)
                z1=z_from_cached(ac1,bs1,int(src),ds)
                r0=choose_destinations(z0,ds,int(src),case['nmax'],inputs,outputs,mode,mut_rng)
                r1=choose_destinations(z1,ds,int(src),case['nmax'],inputs,outputs,mode,mut_rng)
                s0=set(r0['keep'].tolist());s1=set(r1['keep'].tolist())
                js.append(len(s0&s1)/len(s0|s1) if s0|s1 else 1.)
                delta_h.append(float(np.mean(abs(np.sin(z0+SHIFT0)-np.sin(z1+SHIFT0)))))
            mutation_rows.append({'case':case['case'],'mode':mode,'mutation_no':mut,
                'mutated_gene_index':idx,'old_digit':old,'new_digit':old+step,
                'sampled_origins':len(chosen_origins),'jaccard':float(np.mean(js)),
                'mean_abs_H_change_original':float(np.mean(delta_h))})
mutation_df=pd.DataFrame(mutation_rows)
mutation_df.to_csv(OUTPUT_DIR/'mutation_rank_stability.csv',index=False)
display(mutation_df.groupby('mode')[['jaccard','mean_abs_H_change_original']].agg(['mean','std']).round(4))
```

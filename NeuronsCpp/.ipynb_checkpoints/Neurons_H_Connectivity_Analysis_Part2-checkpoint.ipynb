# Neurons — Parte II: por que $H$ concentra conexões?

**Continuação independente do primeiro notebook.** Esta parte investiga o mecanismo por trás do acúmulo de conexões nos mesmos destinos, **sem modificar a função $H$ nem as regras de construção do cérebro**.

**Quatro hipóteses a distinguir:**

1. Muitos neurônios de origem possuem perfis $H(a,b)$ parecidos e favorecem os mesmos valores de $b=A_{\rm destino}$.
2. Existem classes diferentes de preferência, mas algumas recebem muito mais conexões do que outras.
3. A distribuição dos valores de $A$ nos neurônios de destino amplifica ou atenua essas preferências.
4. O aumento de escala torna a seleção *top-$N_{\max}$* mais competitiva, mesmo para um **mesmo genoma**.

Amostras de neurônios de origem são usadas para acelerar os novos diagnósticos; **isso não é uma nova estimativa exata da porcentagem total de isolamento**. Para esse percentual, permanecem válidos os experimentos exaustivos da Parte I.

### Como executar

Coloque **`Neurons_H_Analysis_Results.zip`**, produzido na Parte I, na **mesma pasta** deste notebook; depois execute **Run All**. São necessários `numpy`, `pandas` e `matplotlib`. Não é necessário Godot, C++ ou instalar a primeira parte do notebook. O ZIP contém os genomas reais daquela execução, que serão reutilizados aqui.

Os resultados novos serão exportados como **`Neurons_H_Part2_Results.zip`**. Envie esse arquivo para análise.

## 1. Reproduzir a mesma matemática da Parte I

Usamos a fórmula implementada no C++ do Neurons, com NumPy `float32`:

$$H(a,b)=\sin\!\left[\frac{\pi}{16}+\sum_{m=1}^{4}c_m\cos(ma)\sin(mb)\right],\qquad a=A_{\rm origem},\ b=A_{\rm destino}.$$

O campo $A$ e $N_{\max}$ são reconstruídos a partir dos genes registrados na primeira análise. O processo inicial **seleciona os maiores $H>0$ entre destinos com $|\Delta k|\leq 2$**, e só *depois* elimina conexões proibidas. O resgate não é executado aqui.

**Atenção:** essa é uma reprodução numérica em NumPy da matemática nativa, não uma execução bit a bit idêntica ao C++. Valores exatamente empatados podem selecionar destinos diferentes.

```python
PI32 = np.float32(np.pi)
W8 = np.sqrt(np.arange(1,9,dtype=np.float32))
W64 = np.sqrt(np.arange(1,65,dtype=np.float32))

def parse_genome(record):
    assert set(('connections','beta','max_connections')).issubset(record)
    widths = {'connections':64,'beta':8,'max_connections':8}
    return {k:np.array([int(c) for c in record[k]],dtype=np.int8) for k in widths}

def h_coefficients(g):
    digits = g['connections']
    return np.array([np.sum(W64[digits == d],dtype=np.float32) for d in range(4)],dtype=np.float32) * (np.float32(3)*PI32/(np.float32(4)*np.sum(W64,dtype=np.float32)))

def beta_max(g,I,J):
    beta = np.float32(1.0 + np.sum(W8*g['beta'],dtype=np.float32)/np.sum(3*W8,dtype=np.float32)*np.float32((I*J/16.)**(1/3)))
    maxconn = int(np.floor(np.sum(W8*np.power(beta,g['max_connections'].astype(np.float32)),dtype=np.float32)))
    return float(beta),maxconn

def make_A(g,I,J,K):
    coeff=(np.float32(2)/np.float32(3))*g['connections'].astype(np.float32)-np.float32(1)
    cx=np.cos(PI32*np.arange(4,dtype=np.float32)[:,None]*(np.arange(I,dtype=np.float32)[None,:]/np.float32(I-1))).astype(np.float32)
    cy=np.cos(PI32*np.arange(4,dtype=np.float32)[:,None]*(np.arange(J,dtype=np.float32)[None,:]/np.float32(J-1))).astype(np.float32)
    cz=np.cos(PI32*np.arange(4,dtype=np.float32)[:,None]*(np.arange(K,dtype=np.float32)[None,:]/np.float32(K-1))).astype(np.float32)
    field=np.zeros((K,J,I),dtype=np.float32)
    for r in range(4):
        for s in range(4):
            for q in range(4):
                field += coeff[q+4*s+16*r]*cx[r][None,None,:]*cy[s][None,:,None]*cz[q][:,None,None]
    return np.tanh(field/np.float32(8)).astype(np.float32).ravel()

def h_direct(a,b,c):
    z=np.float32(np.pi/16)
    for m in range(1,5):
        z=z+c[m-1]*np.cos(np.float32(m)*a)*np.sin(np.float32(m)*b)
    return np.sin(z).astype(np.float32)

def make_fast_h(A,c):
    m=np.arange(1,5,dtype=np.float32)[:,None]
    return (c[:,None]*np.cos(m*A[None,:])).astype(np.float32),np.sin(m*A[None,:]).astype(np.float32)

def h_fast(ac,bs,src,targets):
    z=np.full(len(targets),np.float32(np.pi/16),dtype=np.float32)
    for m in range(4):
        z+=ac[m,src]*bs[m,targets]
    return np.sin(z).astype(np.float32)

def flags_by_rank(A,I,J,K,n_inputs,n_outputs):
    by_layer=[]
    for k in range(K):
        ids=np.arange(k*I*J,(k+1)*I*J,dtype=np.int32)
        by_layer.append(ids[np.argsort(A[ids],kind='quicksort')])
    order=np.stack(by_layer,axis=1).ravel()
    input_flags=np.zeros(len(A),dtype=bool)
    output_flags=np.zeros(len(A),dtype=bool)
    input_flags[order[:n_inputs]]=True
    output_flags[order[-n_outputs:]]=True
    return input_flags,output_flags

def prepare_case(rec,size=None):
    size=tuple(map(int,size or rec['size']))
    I,J,K=size
    g=parse_genome(rec['genes'])
    A=make_A(g,I,J,K)
    c=h_coefficients(g)
    beta,nmax=beta_max(g,I,J)
    is_input,is_output=flags_by_rank(A,I,J,K,previous_config['n_inputs'],previous_config['n_outputs'])
    ac,bs=make_fast_h(A,c)
    per_layer=[np.arange(k*I*J,(k+1)*I*J,dtype=np.int32) for k in range(K)]
    windows=[np.concatenate(per_layer[max(0,k-2):min(K,k+3)]) for k in range(K)]
    return dict(label=rec['label'],size=size,N=I*J*K,genome=g,A=A,coeff=c,beta=beta,nmax=nmax,
                is_input=is_input,is_output=is_output,ac=ac,bs=bs,windows=windows)

cases=[prepare_case(rec) for rec in genome_records]
for rec,case in zip(genome_records,cases):
    assert np.allclose(case['coeff'],rec['h_coefficients'],atol=2e-5)
    assert case['nmax']==rec['Nmax'],(rec['label'],case['nmax'],rec['Nmax'])
assert np.allclose(h_fast(cases[0]['ac'],cases[0]['bs'],0,np.array([1,2,3])),
                   h_direct(cases[0]['A'][0],cases[0]['A'][[1,2,3]],cases[0]['coeff']),atol=2e-6)
print('Reprodução de A, H, coeficientes e Nmax: verificações aprovadas.')
```

## 2. Existe uma preferência comum por determinados valores de $A$?

Aqui avaliamos $H(a,b)$ **sem restrições espaciais**, para cinco valores de $a$ extraídos dos quantis reais dos neurônios de origem. Depois avaliamos, para muitas origens, qual valor de $b$ maximiza $H$ no intervalo $[-1,1]$.

Se as curvas forem semelhantes e os máximos se agruparem numa faixa estreita de $b$, isso favorece a hipótese de preferência comum. Se divergirem, pode haver diferentes classes de origem. Esses gráficos descrevem a **afinidade**, não ainda as conexões efetivamente escolhidas.

```python
representative=max(cases,key=lambda x:x['N'])
real_sources=representative['A'][~representative['is_output']]
a_quantiles=np.quantile(real_sources,[.05,.25,.5,.75,.95]).astype(np.float32)
b_grid=np.linspace(-1,1,H_GRID_POINTS,dtype=np.float32)
a_grid=np.linspace(-1,1,181,dtype=np.float32)
Hgrid=h_direct(a_grid[:,None],b_grid[None,:],representative['coeff'])
fig,axs=plt.subplots(1,2,figsize=(14,5),constrained_layout=True)
for q,a in zip([5,25,50,75,95],a_quantiles):
    axs[0].plot(b_grid,h_direct(a,b_grid,representative['coeff']),label=f'Origem: p{q}, A={a:.3f}')
axs[0].axhline(0,color='black',lw=.8,alpha=.5)
axs[0].set(xlabel='A do destino (b)',ylabel='H(a,b)',title='Perfis H para cinco origens representativas')
axs[0].legend(fontsize=8)
im=axs[1].imshow(Hgrid,origin='lower',aspect='auto',extent=[-1,1,-1,1],vmin=-1,vmax=1,cmap='coolwarm')
axs[1].set(xlabel='A do destino (b)',ylabel='A da origem (a)',title='Mapa completo de H(a,b)')
fig.colorbar(im,ax=axs[1],label='H')
fig.savefig(OUTPUT_DIR/'01_h_profiles_and_surface.png',dpi=160)
plt.show()
```

### 2.1. Simetria exata: $H(a,b)=H(-a,b)$

Como os quatro termos de origem são cossenos, a afinidade é **par em $a$**:

$$\cos(m(-a))=\cos(ma)\quad\Longrightarrow\quad H(-a,b)=H(a,b).$$

Duas origens com valores de $A$ opostos possuem **a mesma ordenação de destinos**, quando comparamos os mesmos candidatos elegíveis. Assim, $a$ e $-a$ não correspondem a duas classes de afinidade distintas. Isso não prova que a simetria seja a causa do isolamento, mas é uma propriedade importante da modelagem.


```python
symmetric_a=np.linspace(0,1,41,dtype=np.float32)
symmetric_b=np.linspace(-1,1,101,dtype=np.float32)
left=h_direct(symmetric_a[:,None],symmetric_b[None,:],representative['coeff'])
right=h_direct(-symmetric_a[:,None],symmetric_b[None,:],representative['coeff'])
assert np.allclose(left,right,atol=2e-6)
fig,ax=plt.subplots(figsize=(9,4),constrained_layout=True)
for a in [.15,.45,.75]:
    ax.plot(symmetric_b,h_direct(np.float32(a),symmetric_b,representative['coeff']),label=f'a=±{a:g}')
ax.set(xlabel='A do destino (b)',ylabel='H(a,b)',title='Origens com A opostos possuem perfis H idênticos')
ax.legend();ax.grid(alpha=.2)
fig.savefig(OUTPUT_DIR/'01b_H_even_in_source_A.png',dpi=160)
plt.show()
print('Simetria H(a,b)=H(-a,b) verificada numericamente.')
```

## 3. Máximos de $H$ e classes de preferência

Para cada origem real, identificamos o melhor $b$ numa malha uniforme. Depois medimos a **similaridade da preferência por faixas de $b$**: duas origens são semelhantes quando suas faixas de maior afinidade se sobrepõem.

A sobreposição aqui ocorre no **espaço do escalar $A$**, não entre IDs de neurônios; isso permite comparações mesmo quando as origens não compartilham as mesmas camadas acessíveis.

```python
profile_rng=np.random.default_rng(SEED_PART2+55)
sources_A=real_sources[profile_rng.choice(len(real_sources),min(400,len(real_sources)),replace=False)]
score=h_direct(sources_A[:,None],b_grid[None,:],representative['coeff'])
argmax=np.argmax(score,axis=1)
best_b=b_grid[argmax]
qthr=np.quantile(score,.9,axis=1)
top_mask=score>=qthr[:,None]
# Jaccard de faixas top-10% em pares de origens, no mesmo domínio b.
pair_ix=profile_rng.integers(0,len(sources_A),size=(1600,2))
validpairs=pair_ix[:,0]!=pair_ix[:,1]
pair_ix=pair_ix[validpairs]
intersection=np.sum(top_mask[pair_ix[:,0]] & top_mask[pair_ix[:,1]],axis=1)
union=np.sum(top_mask[pair_ix[:,0]] | top_mask[pair_ix[:,1]],axis=1)
similarity=intersection/np.maximum(1,union)
delta_a=np.abs(sources_A[pair_ix[:,0]]-sources_A[pair_ix[:,1]])
profile_pairs=pd.DataFrame({'delta_A_source':delta_a,'top10_Jaccard':similarity})
profile_pairs.to_csv(OUTPUT_DIR/'h_profile_similarity_pairs.csv',index=False)
fig,axs=plt.subplots(1,3,figsize=(16,4.3),constrained_layout=True)
axs[0].hist(best_b,bins=25,edgecolor='white')
axs[0].set(xlabel='b de máximo H',ylabel='Origens amostradas',title='Destinos preferidos no espaço A')
axs[1].scatter(sources_A,best_b,s=13,alpha=.55)
axs[1].set(xlabel='A da origem',ylabel='b que maximiza H',title='Classes de preferência')
axs[2].scatter(delta_a,similarity,alpha=.2,s=10)
axs[2].set(xlabel='|ΔA| entre origens',ylabel='Jaccard das faixas top-10% de b',title='Similaridade entre perfis de H')
for ax in axs: ax.grid(alpha=.2)
fig.savefig(OUTPUT_DIR/'02_h_favorite_b_and_similarity.png',dpi=160)
plt.show()
print('Jaccard mediano entre os perfis top-10%:',round(float(np.median(similarity)),3))
print('Faixa central (p10/p90) dos máximos de H:',np.round(np.quantile(best_b,[.1,.9]),4))
```

## 4. Seleção real: faixas de $A$ disponíveis, positivas e escolhidas

O teste anterior mede somente $H$. Agora simulamos a seleção inicial **com as mesmas restrições do construtor** para uma amostra uniforme de origens de cada cérebro:

- **Destinos permitidos:** $|\Delta k|\leq2$, sem autoconexão e sem destinos estruturalmente proibidos.
- **Destinos positivos:** candidatos permitidos com $H>0$.
- **Destinos escolhidos:** melhores até $N_{\max}$ dentre todos os candidatos $H>0$; em seguida removem-se destinos proibidos, como no C++.

Calculamos **enriquecimento por faixa de $A$**:

$$E(B)=\frac{P(A_{\rm destino}\in B\mid\text{conexão escolhida})}{P(A_{\rm destino}\in B\mid\text{candidato permitido e }H>0)}.$$

Se $E(B)\gg1$, aquela faixa recebe mais conexões do que seria esperado apenas por sua presença entre os candidatos positivos. Se $E(B)\ll1$, é negligenciada. Faixas com amostras desprezíveis são assinaladas como ausentes, evitando divisões enganosas.

```python
def pick_one(case,src):
    I,J,K=case['size']
    if case['is_output'][src]: return None
    targets=case['windows'][int(src//(I*J))]
    targets=targets[targets!=src]
    h=h_fast(case['ac'],case['bs'],int(src),targets)
    positive=np.flatnonzero(h>0)
    k_take=min(case['nmax'],len(positive))
    if k_take==0:
        selected=np.empty(0,dtype=np.int32)
        chosen_h=np.empty(0,dtype=np.float32)
    else:
        chosen=(positive[np.argpartition(h[positive],-k_take)[-k_take:]]
                if k_take<len(positive) else positive)
        chosen_targets=targets[chosen]
        clean=(~case['is_input'][chosen_targets]) & (~(case['is_input'][src] & case['is_output'][chosen_targets]))
        selected=chosen_targets[clean]
        chosen_h=h[chosen[clean]]
    allowed=(~case['is_input'][targets]) & (~(case['is_input'][src] & case['is_output'][targets]))
    return dict(selected=selected,selected_h=chosen_h,
                allowed=targets[allowed],positive_allowed=targets[allowed & (h>0)],
                n_positive_before_clean=len(positive),n_top_before_clean=k_take)

def sample_sources(case,n,seed):
    src=np.flatnonzero(~case['is_output'])
    rng=np.random.default_rng(seed)
    return np.sort(rng.choice(src,size=min(n,len(src)),replace=False))

def safe_probs(counts):
    counts=np.asarray(counts,dtype=np.float64)
    s=counts.sum()
    return counts/s if s>0 else np.zeros_like(counts)

def sampled_case(case,n_sources,seed):
    srcs=sample_sources(case,n_sources,seed)
    nb=len(A_BINS)-1
    hist_allowed=np.zeros(nb,dtype=np.int64)
    hist_positive=np.zeros(nb,dtype=np.int64)
    hist_selected=np.zeros(nb,dtype=np.int64)
    per_source=[]; pair_profiles=[]
    topk_fracs=[]; h_sel=[]; positive_vals=[]
    source_a=[];layers=[]
    for src in srcs:
        p=pick_one(case,int(src))
        if p is None: continue
        ha=np.histogram(case['A'][p['allowed']],bins=A_BINS)[0]
        hp=np.histogram(case['A'][p['positive_allowed']],bins=A_BINS)[0]
        hs=np.histogram(case['A'][p['selected']],bins=A_BINS)[0]
        hist_allowed+=ha;hist_positive+=hp;hist_selected+=hs
        per_source.append((ha,hp,hs))
        pair_profiles.append(safe_probs(hs))
        topk_fracs.append(p['n_top_before_clean']/max(1,p['n_positive_before_clean']))
        if len(p['selected_h']): h_sel.append(p['selected_h'])
        source_a.append(float(case['A'][src]));layers.append(int(src//(case['size'][0]*case['size'][1])))
    p_sel=safe_probs(hist_selected)
    p_pos=safe_probs(hist_positive)
    sufficient=hist_positive>=max(20,int(.001*hist_positive.sum()))
    enrich=np.full(nb,np.nan)
    enrich[sufficient]=p_sel[sufficient]/np.maximum(p_pos[sufficient],1e-12)
    tv=.5*np.abs(p_sel-p_pos).sum()
    return dict(case=case,srcs=srcs,source_a=np.asarray(source_a),layers=np.asarray(layers),
                profiles=np.asarray(pair_profiles),per_source=per_source,
                allowed=hist_allowed,positive=hist_positive,selected=hist_selected,
                enrich=enrich,topk_fracs=np.asarray(topk_fracs),h_selected=np.concatenate(h_sel) if h_sel else np.array([]),
                tv=float(tv),n_sources=len(srcs),n_selected=int(hist_selected.sum()))

one_test=sampled_case(cases[0],10,SEED_PART2)
assert one_test['selected'].sum()==one_test['n_selected']
assert one_test['n_selected']<=10*cases[0]['nmax']
assert np.isfinite(one_test['tv']) and 0<=one_test['tv']<=1+1e-10
print('Simulação amostrada e contagens: OK')
```

```python
all_samples=[]
begin=time.perf_counter()
for i,case in enumerate(cases):
    samp=sampled_case(case,SOURCES_PER_BRAIN,SEED_PART2+137*i)
    all_samples.append(samp)
    print(f"{i+1:2d}/{len(cases)} {case['label']:24} | {len(samp['srcs']):3d} origens"
          f" | pressão top-K={np.median(samp['topk_fracs'])*100:5.2f}%"
          f" | TV={samp['tv']:.3f}",flush=True)
print(f'Diagnósticos por cérebro concluídos em {time.perf_counter()-begin:.1f} segundos.')
```

## 5. Resultado por faixa de $A$: o que os candidatos oferecem versus o que $H$ escolhe

A comparação usa exatamente **os mesmos candidatos permitidos e positivos** em cada cérebro. Uma diferença grande entre as curvas aponta para *preferência seletiva*, não apenas escassez de neurônios em certas faixas de $A$.

```python
rep_samp=all_samples[int(np.argmax([s['case']['N'] for s in all_samples]))]
centers=(A_BINS[:-1]+A_BINS[1:])/2
fig,axs=plt.subplots(1,3,figsize=(17,4.5),constrained_layout=True)
axs[0].step(centers,safe_probs(rep_samp['allowed']),where='mid',label='Disponíveis/permitidos')
axs[0].step(centers,safe_probs(rep_samp['positive']),where='mid',label='Permitidos e H > 0')
axs[0].step(centers,safe_probs(rep_samp['selected']),where='mid',label='Efetivamente escolhidos')
axs[0].set(xlabel='A do destino',ylabel='Proporção',title='Oferta × seleção (mesmo cérebro)')
axs[0].legend(fontsize=8)
axs[1].bar(centers,np.nan_to_num(rep_samp['enrich']),width=np.diff(A_BINS)*.8,alpha=.85)
axs[1].axhline(1,color='black',linestyle='--',lw=1)
axs[1].set(xlabel='A do destino',ylabel='Enriquecimento E(B)',title='Preferência acima da disponibilidade positiva')
axs[2].hist(rep_samp['case']['A'],bins=A_BINS,alpha=.7,label='Todos os neurônios')
axs[2].hist(rep_samp['case']['A'][~rep_samp['case']['is_input']],bins=A_BINS,histtype='step',lw=1.8,label='Destinos potenciais')
axs[2].set(xlabel='A dos neurônios',ylabel='Contagem',title='Distribuição real do campo A')
axs[2].legend(fontsize=8)
for ax in axs: ax.grid(alpha=.2)
fig.savefig(OUTPUT_DIR/'03_availability_positivity_and_enrichment.png',dpi=160)
plt.show()
```

## 6. Há diferentes classes de origem?

Agrupamos as origens amostradas em quantis de seu $A$. Cada linha do mapa mostra para quais faixas de $A_{\rm destino}$ elas enviam conexões. Além disso, a similaridade de dois histogramas de preferências é medida por

$$S(p,q)=\sum_j\min(p_j,q_j),\quad 0\leq S\leq1.$$

Esse número vale $1$ quando as distribuições por faixa são idênticas. Ele não significa que as origens tenham escolhido os **mesmos neurônios individuais**.

```python
prof=rep_samp['profiles']; a_src=rep_samp['source_a']
quantiles=np.unique(np.quantile(a_src,[0,.2,.4,.6,.8,1]))
groups=np.digitize(a_src,quantiles[1:-1],right=True)
heat=np.vstack([safe_probs(rep_samp['selected']*0+np.sum([rep_samp['per_source'][i][2] for i in np.flatnonzero(groups==g)],axis=0))
               if np.any(groups==g) else np.zeros(len(centers)) for g in range(len(quantiles)-1)])
rng_pairs=np.random.default_rng(SEED_PART2+991)
pair_idx=rng_pairs.integers(0,len(a_src),size=(SOURCE_PAIRS_PER_BRAIN*3,2))
pair_idx=pair_idx[pair_idx[:,0]!=pair_idx[:,1]][:SOURCE_PAIRS_PER_BRAIN]
pair_similarity=np.minimum(prof[pair_idx[:,0]],prof[pair_idx[:,1]]).sum(axis=1)
pair_delta=np.abs(a_src[pair_idx[:,0]]-a_src[pair_idx[:,1]])
pair_table=pd.DataFrame({'delta_A_source':pair_delta,'same_layer':rep_samp['layers'][pair_idx[:,0]]==rep_samp['layers'][pair_idx[:,1]],
                         'A_bin_overlap':pair_similarity})
pair_table.to_csv(OUTPUT_DIR/'selected_A_profile_pairs.csv',index=False)
fig,axs=plt.subplots(1,2,figsize=(14,5),constrained_layout=True)
im=axs[0].imshow(heat,origin='lower',aspect='auto',extent=(A_BINS[0],A_BINS[-1],0,len(heat)),cmap='viridis')
axs[0].set(xlabel='A do destino',ylabel='Grupo (quantil de A da origem)',title='Distribuição dos destinos escolhidos por classe de origem')
fig.colorbar(im,ax=axs[0],label='Proporção das conexões da classe')
axs[1].scatter(pair_delta,pair_similarity,alpha=.35,s=12)
axs[1].set(xlabel='|ΔA| entre origens',ylabel='Sobreposição das faixas escolhidas',title='Convergência das preferências reais')
for ax in axs: ax.grid(alpha=.15)
fig.savefig(OUTPUT_DIR/'04_source_classes_and_overlap.png',dpi=160)
plt.show()
print('Sobreposição mediana entre origens:',round(float(np.median(pair_similarity)),3))
```

## 7. Padrão se repete entre genomas e dimensões?

O índice **TV** (*total variation distance*) entre a distribuição de $A$ escolhida e a distribuição dos candidatos positivos varia entre 0 e 1. Valores altos indicam preferência concentrada em faixas particulares do espaço $A$. A **pressão top-$K$** é a fração de candidatos com $H>0$ que entra no limite inicial $N_{\max}$ **antes da limpeza**. Quanto menor essa fração, mais seletiva é a competição.

Esses números são calculados a partir de origens amostradas; o percentual exato de neurônios isolados continua sendo o da Parte I.

```python
rows=[]
for s in all_samples:
    case=s['case']; pp=safe_probs(s['selected'])
    row={'label':case['label'],'N':case['N'],'I':case['size'][0], 'Nmax':case['nmax'],
         'sampled_sources':s['n_sources'], 'sampled_connections':s['n_selected'],
         'median_topK_pct':100*np.median(s['topk_fracs']),
         'TV_selected_vs_positive':s['tv'],
         'top2_A_bins_share':float(np.sort(pp)[-2:].sum()),
         'median_selected_H':float(np.median(s['h_selected'])) if len(s['h_selected']) else np.nan}
    rows.append(row)
selection_metrics=pd.DataFrame(rows)
selection_metrics=selection_metrics.merge(old_metrics[['label','isolation_H_pct']],on='label',how='left')
display(selection_metrics.round(4))
selection_metrics.to_csv(OUTPUT_DIR/'selection_bias_by_brain.csv',index=False)
fig,axs=plt.subplots(1,3,figsize=(16,4.5),constrained_layout=True)
for key,y,title in [('median_topK_pct','Fraçao top-K dos H positivos (%)','Competição efetiva'),
                    ('TV_selected_vs_positive','TV: escolhidos vs positivos','Viés de preferência por A'),
                    ('top2_A_bins_share','Fraçao em duas faixas dominantes','Concentração de destinos em A')]:
    axs[{'median_topK_pct':0,'TV_selected_vs_positive':1,'top2_A_bins_share':2}[key]].scatter(selection_metrics['N'],selection_metrics[key])
    ax=axs[{'median_topK_pct':0,'TV_selected_vs_positive':1,'top2_A_bins_share':2}[key]]
    ax.set(xlabel='Total de neurônios',ylabel=y,title=title)
    ax.grid(alpha=.2)
fig.savefig(OUTPUT_DIR/'05_metrics_by_size.png',dpi=160)
plt.show()
```

### 7.1. Preferências em **todos** os genomas

Para não depender exclusivamente do maior cérebro de referência, repetimos a comparação dos histogramas de destinos escolhidos em pares de origens de cada cérebro. Também exportamos, para **cada genoma e cada faixa de $A$**, as candidaturas permitidas e positivas e as conexões efetivamente selecionadas. Os pares são amostrais, mas permitem avaliar a variabilidade entre genomas.


```python
pair_rows=[]; abins_rows=[]; per_brain_pair=[]
centers=(A_BINS[:-1]+A_BINS[1:])/2
for i,samp in enumerate(all_samples):
    label=samp['case']['label']
    rng=np.random.default_rng(SEED_PART2+600+i)
    n=len(samp['srcs'])
    pairs=rng.integers(0,n,size=(SOURCE_PAIRS_PER_BRAIN*3,2))
    pairs=pairs[pairs[:,0]!=pairs[:,1]][:SOURCE_PAIRS_PER_BRAIN]
    ov=np.minimum(samp['profiles'][pairs[:,0]],samp['profiles'][pairs[:,1]]).sum(axis=1)
    da=np.abs(samp['source_a'][pairs[:,0]]-samp['source_a'][pairs[:,1]])
    sl=samp['layers'][pairs[:,0]]==samp['layers'][pairs[:,1]]
    pair_rows.extend({'label':label,'N':samp['case']['N'],'delta_A_source':float(d),'same_layer':bool(j),'A_bin_overlap':float(o)}
                     for d,j,o in zip(da,sl,ov))
    per_brain_pair.append({'label':label,'N':samp['case']['N'],'median_pair_A_bin_overlap':float(np.median(ov)),
                           'mean_pair_A_bin_overlap':float(np.mean(ov)),
                           'median_pair_overlap_within_layer':float(np.median(ov[sl])) if np.any(sl) else np.nan})
    for j in range(len(centers)):
        abins_rows.append({'label':label,'N':samp['case']['N'],'A_bin_left':A_BINS[j],
                           'A_bin_right':A_BINS[j+1],'allowed':int(samp['allowed'][j]),
                           'H_positive_allowed':int(samp['positive'][j]),
                           'selected':int(samp['selected'][j]),'enrichment':float(samp['enrich'][j])})
perbrain_pref=pd.DataFrame(per_brain_pair)
perbrain_pref=perbrain_pref.merge(selection_metrics[['label','isolation_H_pct','median_topK_pct','TV_selected_vs_positive']],on='label')
display(perbrain_pref.round(3))
pd.DataFrame(pair_rows).to_csv(OUTPUT_DIR/'all_genomes_preference_pair_samples.csv',index=False)
pd.DataFrame(abins_rows).to_csv(OUTPUT_DIR/'all_genomes_A_bin_counts.csv',index=False)
perbrain_pref.to_csv(OUTPUT_DIR/'all_genomes_preference_summary.csv',index=False)
fig,axs=plt.subplots(1,2,figsize=(12,4.5),constrained_layout=True)
axs[0].scatter(perbrain_pref['N'],perbrain_pref['median_pair_A_bin_overlap'])
axs[0].set(xlabel='Neurônios',ylabel='Sobreposição mediana entre pares',title='Preferências por A: similaridade')
axs[1].scatter(perbrain_pref['median_pair_A_bin_overlap'],perbrain_pref['isolation_H_pct'])
axs[1].set(xlabel='Sobreposição mediana entre pares',ylabel='Isolamento real (%)',title='Semelhança e isolamento (exploratório)')
for ax in axs: ax.grid(alpha=.2)
fig.savefig(OUTPUT_DIR/'05b_similarity_all_genomes.png',dpi=160)
plt.show()
```

## 8. Experimento pareado: **mesmo genoma** em tamanhos diferentes

Na Parte I, foram gerados genomas aleatórios diferentes em cada tamanho. Aqui repetimos **os mesmos genes** nas dimensões 10, 22, 32 e 42 (mantendo $K=10$), calculando $A$, $\beta$ e $N_{\max}$ em cada dimensão conforme a modelagem.

Isso permite separar melhor o efeito da escala da variação genética. **Não medimos a taxa total de isolamento aqui**: usamos um número fixo de origens amostradas, por custo computacional.

```python
large_recs=sorted(genome_records,key=lambda r:(np.prod(r['size']),r['seed']),reverse=True)[:MATCHED_GENOMES]
matched=[];matched_samples=[]
for gi,rec in enumerate(large_recs):
    for I in MATCHED_SIZES:
        case=prepare_case(rec,(I,I,int(rec['size'][2])))
        samp=sampled_case(case,MATCHED_SOURCES,SEED_PART2+5000*gi+I)
        matched_samples.append((gi,samp))
        matched.append(dict(genome=gi,source_genome_label=rec['label'],I=I,N=case['N'],Nmax=case['nmax'],
                            median_topK_pct=100*np.median(samp['topk_fracs']),
                            TV_selected_vs_positive=samp['tv'],
                            top2_A_bins_share=float(np.sort(safe_probs(samp['selected']))[-2:].sum()),
                            A_std=float(np.std(case['A'])),A_min=float(case['A'].min()),A_max=float(case['A'].max())))
        print('Genoma',gi+1,'/',I,'×',I,'| K/Npositivos=',f"{matched[-1]['median_topK_pct']:.2f}%",'| TV=',f"{samp['tv']:.3f}",flush=True)
matched_df=pd.DataFrame(matched)
display(matched_df.round(4))
matched_df.to_csv(OUTPUT_DIR/'matched_genomes_across_sizes.csv',index=False)
fig,axs=plt.subplots(1,3,figsize=(16,4.5),constrained_layout=True)
for gi,g in matched_df.groupby('genome'):
    for ax,col,y in [(axs[0],'median_topK_pct','Top-K / positivos (%)'),
                     (axs[1],'TV_selected_vs_positive','TV (viés sobre faixas de A)'),
                     (axs[2],'A_std','Desvio-padrão do campo A')]:
        ax.plot(g['N'],g[col],'-o',label=f'Genoma {gi+1}')
        ax.set(xlabel='Neurônios',ylabel=y)
        ax.grid(alpha=.2)
axs[0].set_title('Seleção torna-se mais competitiva?')
axs[1].set_title('Seleção concentra mais em A?')
axs[2].set_title('Distribuição de A mudou?')
axs[0].legend(fontsize=8)
fig.savefig(OUTPUT_DIR/'06_same_genomes_multiple_sizes.png',dpi=160)
plt.show()
```

## 9. A localização horizontal importa por si só?

Teste de simetria: **embaralhar os valores de $A$ dentro de cada camada** (e transportar as identidades de input/output junto com eles) mantém o multiconjunto de $A$ de cada camada. Como $H$ depende apenas de $A_{\rm origem}$ e $A_{\rm destino}$, e a restrição espacial depende apenas da diferença entre camadas $k$, a seleção deveria ser **invariante à permutação horizontal**, salvo desempates numéricos.

Este teste não modifica o modelo: apenas confirma uma simetria já implicada por suas equações. Isso é diferente de embaralhar neurônios **entre camadas**, o que mudaria os conjuntos geometricamente elegíveis.

```python
shuffle_case=next((c for c in cases if c['size'][0]==22),cases[0])
I,J,K=shuffle_case['size']
rng_shuffle=np.random.default_rng(SEED_PART2+907)
# new_idx -> old_idx; embaralhamento estritamente dentro de cada camada.
perm=np.concatenate([rng_shuffle.permutation(np.arange(k*I*J,(k+1)*I*J,dtype=np.int32)) for k in range(K)])
inverse=np.empty_like(perm);inverse[perm]=np.arange(len(perm))
sh=dict(shuffle_case)
sh['A']=shuffle_case['A'][perm]
sh['is_input']=shuffle_case['is_input'][perm]
sh['is_output']=shuffle_case['is_output'][perm]
sh['ac'],sh['bs']=make_fast_h(sh['A'],sh['coeff'])
old_sources=sample_sources(shuffle_case,SHUFFLE_TEST_SOURCES,SEED_PART2+444)
matches=0;total=0;examples=[]
for src in old_sources:
    base=pick_one(shuffle_case,int(src))
    permuted=pick_one(sh,int(inverse[src]))
    old_targets=set(base['selected'].tolist())
    new_targets=set(perm[permuted['selected']].tolist())
    ok=old_targets==new_targets
    matches+=int(ok);total+=1
    if not ok and len(examples)<5:
        examples.append(dict(old_source=int(src),old_edges=len(old_targets),new_edges=len(new_targets),
                             symmetric_difference=len(old_targets^new_targets)))
shuffle_report={'tested_sources':total,'same_destinations_after_relabeling':matches,
                'match_percentage':100*matches/max(1,total),'tie_or_numerical_examples':examples,
                'genome':shuffle_case['label'],'size':shuffle_case['size']}
print(json.dumps(shuffle_report,indent=2,ensure_ascii=False))
with (OUTPUT_DIR/'within_layer_permutation_test.json').open('w',encoding='utf-8') as f:
    json.dump(shuffle_report,f,indent=2,ensure_ascii=False)
```

### Como interpretar os próximos resultados

- Perfis $H(a,b)$ semelhantes + máximos em $b$ semelhantes: favorece a hipótese de uma **preferência comum**.
- Curvas distintas por quantil de origem: favorece **classes de preferência**; verificar se alguma domina na amostra real.
- Enriquecimento $E(B)$ muito acima de 1 para poucos bins: sugere que o ranking de $H$ concentra destinos para além de sua disponibilidade.
- Com o **mesmo genoma**, pressão top-$K$ decrescente com a dimensão: sugere competição crescente pela seleção dos destinos favoritos.
- Permutação dentro das camadas preservando seleções: sugere que a **posição horizontal não desempata** destinos; é o valor $A$ que determina a preferência entre eles.

**Limitações:** fontes amostradas; resultados de $H$ representados por `float32`/NumPy; seleções podem variar apenas em empates. Não houve modificação nem simulação do resgate.

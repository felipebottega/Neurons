# Neurons — Parte VIII: controle de campo aleatório independente

**Pergunta falsificável:** se atribuirmos uma fase espacial **independente e uniforme** a cada neurônio, o isolamento antes do resgate deixa de ser elevado?

$$r(p)\overset{\mathrm{iid}}{\sim}\mathcal U(0,1),\qquad\delta(p)=8r(p)-4.$$

Cada $r(p)$ é sorteado **uma só vez por genoma e réplica**, mantido fixo durante a construção de todas as conexões e reutilizado antes/depois das mutações. O sorteio ocorre por **neurônio**, nunca por aresta. Os campos aleatórios não são candidatos ao modelo final porque não preservam a suavidade espacial: são um *controle extremo* para separar a diversidade de fases das limitações de $H$ e da seleção top-$N_{\max}$.

**Condições pareadas:** `original`, `radial` (Parte V) e três realizações independentes `random_1`, `random_2`, `random_3`. Todas usam os mesmos genomas, funções $A$ e $H$, posições válidas, papéis entrada/saída, $N_{\max}$, filtragem e política de resgate (medidas **antes** do resgate).

**Como interpretar:** comparar cada réplica ao radial em cada genoma, além de verificar a variância entre réplicas. Se a cobertura não melhorar muito com fases independentes e uniformes, a hipótese de que a irregularidade espacial não é a limitação dominante ganhará força. O resultado não prova que *nenhum* campo espacial seja capaz de resolver o problema.


## 0. Configuração, dados e perfis

```python
import json, time, io, zipfile
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

PROFILE = 'quick'           # 'quick': 4 genomas (1.000 e 4.840 neurônios); 'full': os 11 genomas (até 17.640)
N_INPUTS, N_OUTPUTS = 10, 10
SHUFFLE_SEED = 20261025
MODES = ['original', 'radial', 'random_1', 'random_2', 'random_3']
MUTATION_SEED = 20261026
MUTATIONS_PER_GENOME = 2 if PROFILE == 'quick' else 2
STABILITY_SOURCES_PER_GENOME = 32
OUTPUT_DIR = Path('neurons_h_part8_results')
OUTPUT_DIR.mkdir(exist_ok=True)
RESULTS_ZIP = Path('Neurons_H_Part8_Results.zip')
SOURCE_ZIP = Path('Neurons_H_Analysis_Results.zip')
PART4_ZIP = Path('Neurons_H_Part4_Results.zip')  # opcional, usado apenas para referências comparativas
assert SOURCE_ZIP.exists(), 'Coloque Neurons_H_Analysis_Results.zip na mesma pasta deste notebook.'
with zipfile.ZipFile(SOURCE_ZIP) as archive:
    records = json.loads(archive.read('genomes.json'))
assert len(records) == 11, 'Esperados os 11 genomas da Parte I.'
if PROFILE == 'quick':
    selected = [r for r in records if r['size'][0] == 10][:2] + [r for r in records if r['size'][0] == 22][:2]
elif PROFILE == 'full':
    selected = records
else:
    raise ValueError('PROFILE precisa ser quick ou full')
print('Perfil:', PROFILE, '| Genomas:', len(selected), '| Condições:', MODES)
print('Dimensões:', [r['size'] for r in selected])
```

## 1. Reproduzindo $A$, os genes e os coeficientes de $H$

Estas funções são **as mesmas reproduções numéricas em NumPy usadas nos notebooks anteriores**; seguimos trabalhando com `float32`. Não modificamos a definição de $A$, $\beta$ ou $N_{\max}$. Pequenas diferenças em empates/arredondamento frente ao C++ são possíveis.

```python
PI32 = np.float32(np.pi)
W8 = np.sqrt(np.arange(1, 9, dtype=np.float32))
W64 = np.sqrt(np.arange(1, 65, dtype=np.float32))

def random_genome(seed):
    rng = np.random.default_rng(seed)
    # O construtor C++ gera dígitos independentes e uniformes de 0 a 3.
    return {"connections": rng.integers(0,4,64,dtype=np.int8),
            "beta": rng.integers(0,4,8,dtype=np.int8),
            "max_connections": rng.integers(0,4,8,dtype=np.int8)}

def parse_genome(dictionary):
    lengths = {"connections": 64, "beta": 8, "max_connections": 8}
    ans = {}
    for name, count in lengths.items():
        value = dictionary[name]
        if isinstance(value, str):
            if len(value) != count or not set(value) <= set("0123"):
                raise ValueError(f"{name}: esperado {count} dígitos quaternários")
            ans[name] = np.array([int(ch) for ch in value], dtype=np.int8)
        else:
            ans[name] = np.asarray(value, dtype=np.int8)
            if ans[name].shape != (count,) or not np.isin(ans[name], [0,1,2,3]).all():
                raise ValueError(f"{name}: formato inválido")
    return ans

def genome_text(g):
    return {key: "".join(map(str, val)) for key, val in g.items()}

def beta_max(g, I, J):
    beta = np.float32(1.0 + np.sum(W8*g["beta"],dtype=np.float32)
                     / np.sum(3*W8,dtype=np.float32) * np.float32((I*J/16.)**(1/3)))
    maxconn = int(np.floor(np.sum(W8*np.power(beta, g["max_connections"].astype(np.float32)),dtype=np.float32)))
    return float(beta), maxconn

def h_coefficients(g):
    digits = g["connections"]
    return np.array([np.sum(W64[digits == d],dtype=np.float32) for d in range(4)],dtype=np.float32) * (np.float32(3)*PI32 / (np.float32(4)*np.sum(W64,dtype=np.float32)))

def make_A(g, I, J, K):
    # A[k,j,i], linearizado como i + I*j + I*J*k.
    coeff = (np.float32(2)/np.float32(3))*g["connections"].astype(np.float32)-np.float32(1)
    cx = np.cos(PI32*np.arange(4,dtype=np.float32)[:,None]*
                (np.arange(I,dtype=np.float32)[None,:]/np.float32(I-1))).astype(np.float32)
    cy = np.cos(PI32*np.arange(4,dtype=np.float32)[:,None]*
                (np.arange(J,dtype=np.float32)[None,:]/np.float32(J-1))).astype(np.float32)
    cz = np.cos(PI32*np.arange(4,dtype=np.float32)[:,None]*
                (np.arange(K,dtype=np.float32)[None,:]/np.float32(K-1))).astype(np.float32)
    field = np.zeros((K,J,I), dtype=np.float32)
    for r in range(4):
        for s in range(4):
            for q in range(4):
                field += coeff[q + 4*s + 16*r] * cx[r][None,None,:] * cy[s][None,:,None] * cz[q][:,None,None]
    return np.tanh(field / np.float32(8)).astype(np.float32).ravel()

def h_direct(a, b, c):
    z = np.float32(np.pi/16)
    for m in range(1, 5):
        z = z + c[m-1]*np.cos(np.float32(m)*a)*np.sin(np.float32(m)*b)
    return np.sin(z)

def make_fast_h(A, coeffs):
    harmonics = np.arange(1,5,dtype=np.float32)[:,None]
    ac = (coeffs[:,None]*np.cos(harmonics*A[None,:])).astype(np.float32)
    bs = np.sin(harmonics*A[None,:]).astype(np.float32)
    return ac, bs

def h_from_cached(ac, bs, src, target_indices):
    z = np.full(len(target_indices), np.float32(np.pi/16), dtype=np.float32)
    for m in range(4):
        z += ac[m, src] * bs[m, target_indices]
    return np.sin(z).astype(np.float32)


# Testes matemáticos diretos da referência original
_test_g = random_genome(2026)
_test_A = make_A(_test_g, 5,5,4)
_test_c = h_coefficients(_test_g)
_ac, _bs = make_fast_h(_test_A, _test_c)
_ind = np.arange(0, len(_test_A), 3, dtype=np.int32)
assert np.allclose(h_from_cached(_ac,_bs,0,_ind),h_direct(_test_A[0],_test_A[_ind],_test_c),atol=1e-6)
print('A / H original: OK')
```

## 2. Campos de controle: original, radial e uniforme independente por neurônio

```python
def position_delta(size, kind='radial', seed=SHUFFLE_SEED):
    I,J,K = map(int,size)
    assert min(I,J,K)>1
    u=np.arange(I,dtype=np.float32)/np.float32(I-1)
    v=np.arange(J,dtype=np.float32)/np.float32(J-1)
    w=np.arange(K,dtype=np.float32)/np.float32(K-1)
    U=u[None,None,:];V=v[None,:,None];W=w[:,None,None]
    if kind=='original': return np.zeros(I*J*K,dtype=np.float32)
    if kind=='radial':
        r=np.sqrt((U*U+V*V+W*W)/np.float32(3)).astype(np.float32)
        return (np.float32(8)*r-np.float32(4)).ravel()
    if kind.startswith('random_'):
        replicate=int(kind.split('_')[1])
        if replicate not in (1,2,3): raise ValueError(kind)
        # Gerador separado em cada réplica; seed da simulação/grupo de genes
        # define campo fixo e reproduzível, inclusive durante mutações.
        rng=np.random.default_rng(np.random.SeedSequence([int(seed),replicate,I,J,K]))
        r=rng.random(I*J*K,dtype=np.float64)
        return (8*r-4).astype(np.float32)
    raise ValueError(kind)

def cache_perturbed_H(A, coeffs, delta):
    # Padrão original, apenas substituindo m*A por m*A + delta.
    arg = np.arange(1,5,dtype=np.float32)[:,None]*A[None,:]+delta[None,:]
    cos_src = (coeffs[:,None]*np.cos(arg)).astype(np.float32)
    sin_dst = np.sin(arg).astype(np.float32)
    return cos_src, sin_dst

# Sanidade: extremos, identidade sem perturbacao e continuidade da formula
size_check=(7,9,5)
for kind in MODES:
    _d=position_delta(size_check,kind)
    assert len(_d)==np.prod(size_check)
    assert np.isfinite(_d).all() and np.max(np.abs(_d))<=4.00001
assert np.isclose(position_delta(size_check,'radial')[0],-4,atol=2e-6)
assert np.isclose(position_delta(size_check,'radial')[-1],4,atol=2e-6)
# Aleatoriedade espacial: reprodutibilidade, troca de semente e independência aproximada.
_r1=position_delta(size_check,'random_1',seed=1729)
assert np.array_equal(_r1,position_delta(size_check,'random_1',seed=1729))
assert not np.array_equal(_r1,position_delta(size_check,'random_2',seed=1729))
assert not np.array_equal(_r1,position_delta(size_check,'random_1',seed=1730))
assert np.std(_r1)>1.5 and np.min(_r1)>=-4 and np.max(_r1)<4
_zero=np.zeros(len(_test_A),dtype=np.float32)
_ac0,_bs0=cache_perturbed_H(_test_A,_test_c,_zero)
assert np.array_equal(_ac0,_ac) and np.array_equal(_bs0,_bs)
_dtest=position_delta((5,5,4),'random_1',seed=1729)
_a,_b=cache_perturbed_H(_test_A,_test_c,_dtest)
_direct=np.full(len(_ind),np.float32(np.pi/16),dtype=np.float32)
for m in range(1,5):
    _direct+=_test_c[m-1]*np.cos(np.float32(m)*_test_A[0]+_dtest[0])*np.sin(np.float32(m)*_test_A[_ind]+_dtest[_ind])
assert np.allclose(h_from_cached(_a,_b,0,_ind),np.sin(_direct),atol=1e-6)
print('H original, limites e campo aleatorio fixo: OK')
```

## 3. Mesmo algoritmo de conexões da Parte I

Para cada origem elegível, percorremos todos os candidatos situados a até duas camadas $k$, excluindo a autoconexão. Selecionamos até $N_{\max}$ maiores valores positivos de $H$ **antes de remover arestas proibidas**. Como no construtor, saídas não são origens; entradas não podem ser destinos; entradas não ligam diretamente a saídas.

Registramos separadamente: isolamento dos internos (sem entradas), isolamento das saídas (sem entradas), isolamento das entradas (sem saídas), grau de entrada, concentração, fração de candidatos $H>0$, número de arestas e afinidade média. **Não simulamos resgate**.

```python
def flags_by_rank(A, I, J, K, n_inputs, n_outputs):
    N = len(A)
    if n_inputs+n_outputs > N:
        raise ValueError("Número de inputs+outputs maior que número de neurônios")
    by_layer = []
    for k in range(K):
        ix = np.arange(k*I*J,(k+1)*I*J,dtype=np.int32)
        by_layer.append(ix[np.argsort(A[ix],kind="quicksort")])
    # rank 0 de cada camada; depois rank 1 de cada camada; ...
    permutation = np.stack(by_layer,axis=1).ravel()
    is_input = np.zeros(N,dtype=bool); is_output = np.zeros(N,dtype=bool)
    is_input[permutation[:n_inputs]] = True
    is_output[permutation[-n_outputs:]] = True
    return is_input, is_output

def gini_nonnegative(data):
    x = np.sort(np.asarray(data,dtype=np.float64)); n=len(x)
    if n == 0 or np.sum(x) == 0: return 0.
    return float((2*np.dot(np.arange(1,n+1),x)/(n*np.sum(x))) - (n+1)/n)

def metrics(incoming, outgoing, is_input, is_output):
    internal = ~(is_input | is_output)
    counts = {"internal":int(np.count_nonzero((incoming==0)&internal)),
              "output":int(np.count_nonzero((incoming==0)&is_output)),
              "input":int(np.count_nonzero((outgoing==0)&is_input))}
    N=len(incoming)
    eligible = ~is_input
    vals = incoming[eligible]
    tenpct = max(1,int(np.ceil(.1*len(vals))))
    total = np.sum(vals)
    top_share = float(np.partition(vals,-tenpct)[-tenpct:].sum()/total) if total else np.nan
    return {"isolated_total":sum(counts.values()),"isolation_pct":100.*sum(counts.values())/N,
            "isolated_internal":counts["internal"],"isolated_output":counts["output"],"isolated_input":counts["input"],
            "mean_incoming":float(np.mean(vals)),"median_incoming":float(np.median(vals)),
            "gini_incoming":gini_nonnegative(vals),"top_10pct_incoming_share":top_share,
            "zero_incoming_eligible_pct":100.*np.mean(vals==0),"total_valid_connections":int(np.sum(vals))}


def simulate_perturbed(A, genome, size, delta, roles=None, keep_indegree=False):
    I,J,K=map(int,size)
    N=I*J*K
    inputs,outputs=flags_by_rank(A,I,J,K,N_INPUTS,N_OUTPUTS) if roles is None else roles
    beta,nmax=beta_max(genome,I,J)
    c=h_coefficients(genome)
    ac,bs=cache_perturbed_H(A,c,delta)
    incoming=np.zeros(N,dtype=np.int32)
    outgoing=np.zeros(N,dtype=np.int32)
    layer_size=I*J
    windows=[np.arange(max(0,k-2)*layer_size,min(K,k+3)*layer_size,dtype=np.int32) for k in range(K)]
    num_candidates=num_positive=num_before=num_removed=0
    sum_H=0.0; num_selected=0
    for src in np.flatnonzero(~outputs):
        candidates=windows[int(src//layer_size)]
        dest=candidates[candidates!=src]
        h=h_from_cached(ac,bs,int(src),dest)
        positive=np.flatnonzero(h>0)
        num_positive+=len(positive);num_candidates+=len(dest)
        take=min(int(nmax),len(positive))
        if take==0:
            continue
        picked = (positive[np.argpartition(h[positive],-take)[-take:]]
                  if take<len(positive) else positive)
        candidate_selected=dest[picked]
        valid=(~inputs[candidate_selected]) & (~(inputs[src] & outputs[candidate_selected]))
        keep=candidate_selected[valid]
        incoming[keep]+=1
        outgoing[src]+=len(keep)
        num_before+=len(picked);num_removed+=len(picked)-len(keep)
        sum_H+=float(h[picked][valid].sum(dtype=np.float64));num_selected+=len(keep)
    obs=metrics(incoming,outgoing,inputs,outputs)
    obs.update(beta=float(beta),Nmax=int(nmax),
      positive_pct=100*num_positive/num_candidates if num_candidates else np.nan,
      removed_pct=100*num_removed/num_before if num_before else np.nan,
      mean_H=sum_H/num_selected if num_selected else np.nan,
      outputs_isolated_pct=100*obs['isolated_output']/N_OUTPUTS,
      inputs_isolated_pct=100*obs['isolated_input']/N_INPUTS,
      max_incoming=int(incoming.max()))
    if keep_indegree:
        obs['incoming']=incoming
    return obs

# Controle reduzido: nenhuma mudança deve ocorrer com δ=0
_sm_g=random_genome(101)
_sm_A=make_A(_sm_g,5,5,4)
_sm0=simulate_perturbed(_sm_A,_sm_g,(5,5,4),position_delta((5,5,4),'original'))
_smr=simulate_perturbed(_sm_A,_sm_g,(5,5,4),position_delta((5,5,4),'radial'))
assert 0<=_sm0['isolation_pct']<=100 and 0<=_smr['isolation_pct']<=100
assert _sm0['total_valid_connections']==_sm0['mean_incoming']*(100-10) or np.isclose(_sm0['total_valid_connections'],_sm0['mean_incoming']*90)
print('Seleção de conexões: OK | referência:',round(_sm0['isolation_pct'],2),'% | radial:',round(_smr['isolation_pct'],2),'%')
```

## 4. Simulação pareada: top-$N_{\max}$ e arestas válidas antes do resgate

```python
cases=[]
for i,r in enumerate(selected):
    genes=parse_genome(r['genes']);size=tuple(r['size']);A=make_A(genes,*size)
    roles=flags_by_rank(A,*size,N_INPUTS,N_OUTPUTS)
    cases.append(dict(case=f"{size[0]}x{size[1]}x{size[2]}_g{i+1:02d}",
       genome_seed=r['seed'],genome=genes,size=size,A=A,roles=roles))
rows=[];degree_profiles={}; started=time.perf_counter()
for case in cases:
    for mode in MODES:
        seed=SHUFFLE_SEED+int(case['genome_seed'])
        delta=position_delta(case['size'],mode,seed)
        tic=time.perf_counter()
        m=simulate_perturbed(case['A'],case['genome'],case['size'],delta,case['roles'],
                             keep_indegree=(mode in ('original','radial','random_1')))
        row=dict(case=case['case'],genome_seed=case['genome_seed'],neurons=len(case['A']),
                 size_x=case['size'][0],size_y=case['size'][1],size_z=case['size'][2],
                 mode=mode,delta_seed=int(seed),
                 runtime_s=time.perf_counter()-tic,
                 delta_min=float(delta.min()),delta_max=float(delta.max()),
                 delta_mean=float(delta.mean()),delta_sd=float(delta.std()))
        row.update({k:v for k,v in m.items() if k!='incoming'})
        rows.append(row)
        if 'incoming' in m:
            degree_profiles[(case['case'],mode)]=m['incoming']
        print(f"{case['case']:17} | {mode:17} | isolated={m['isolation_pct']:6.2f}%"
              f" | out={m['isolated_output']:2d}/{N_OUTPUTS}"
              f" | Gini={m['gini_incoming']:.3f} | H={m['mean_H']:.4f}"
              f" | {row['runtime_s']:.1f}s",flush=True)
print('Tempo total:',round(time.perf_counter()-started,1),'s')
df=pd.DataFrame(rows)
df.to_csv(OUTPUT_DIR/'per_genome_delta.csv',index=False)
display(df.head(12).round(4))
```

## 5. Diferenças pareadas e médias por tamanho

```python
baseline=df[df['mode']=='original'].drop(columns=['mode','runtime_s','delta_seed','delta_min','delta_max','delta_mean','delta_sd'])
paired=df.merge(baseline,on=['case','genome_seed','neurons','size_x','size_y','size_z'],suffixes=('','_original'),validate='many_to_one')
comparison_fields=['isolation_pct','isolated_internal','isolated_output','isolated_input',
                   'outputs_isolated_pct','gini_incoming','top_10pct_incoming_share',
                   'total_valid_connections','mean_H','positive_pct','removed_pct','max_incoming']
for col in comparison_fields:
    paired['delta_'+col]=paired[col]-paired[col+'_original']
paired.to_csv(OUTPUT_DIR/'paired_vs_original.csv',index=False)
summary=df.groupby(['neurons','mode'],as_index=False).agg(
    genomes=('case','nunique'),isolation=('isolation_pct','mean'),
    internal_isolated=('isolated_internal','mean'),
    output_isolated=('isolated_output','mean'),input_isolated=('isolated_input','mean'),
    out_isolated_pct=('outputs_isolated_pct','mean'),
    gini=('gini_incoming','mean'),top10=('top_10pct_incoming_share','mean'),
    mean_H=('mean_H','mean'),edges=('total_valid_connections','mean'),
    positive_pct=('positive_pct','mean'),removed_pct=('removed_pct','mean'))
summary.to_csv(OUTPUT_DIR/'summary_by_size_mode.csv',index=False)
ranked=paired.groupby('mode').agg(
    n_genomes=('case','nunique'),mean_delta_isolation_pp=('delta_isolation_pct','mean'),
    improved_genomes=('delta_isolation_pct',lambda s:int((s<0).sum())),
    worst_delta_isolation_pp=('delta_isolation_pct','max'),
    mean_delta_output_isolated=('delta_isolated_output','mean'),
    mean_delta_gini=('delta_gini_incoming','mean'),
    mean_delta_H=('delta_mean_H','mean')).sort_values('mean_delta_isolation_pp')
ranked.to_csv(OUTPUT_DIR/'delta_ranking.csv')
print('RESUMO POR TAMANHO:')
display(summary.round(4))
print('MUDANÇAS PAREADAS — CADA GENOMA COM O MESMO PESO:')
display(ranked.round(4))
```

## 12. Robustez perante mutações de um dígito

O gene `connections` recebe uma mutação de **um dígito**, alterado por $+1$ ou $-1$ quando permitido. Recalculamos $A$ e os coeficientes de $H$, mas **congelamos os papéis input/output do cérebro original** para isolar o efeito das afinidades, não mudanças na classificação. Para cada genoma e mutação, amostramos origens elegíveis determinísticas e comparamos (i) mudança absoluta de $H$ em pares origem-destino idênticos e (ii) Jaccard dos destinos válidos escolhidos entre os maiores $N_{\max}$. Valores de Jaccard altos indicam maior estabilidade.

**Importante:** não é um novo construtor completo após a mutação; é uma comparação de topologia em uma amostra fixa de origens, antes do resgate. Mutações em dígitos de impacto muito diferente podem ter efeitos desiguais. A amostragem e a semente ficam registradas.

```python
def sampled_edges_and_H(A, genes, size, delta, origins, roles):
    I,J,K=map(int,size); layer=I*J
    inputs,outputs=roles
    _,nmax=beta_max(genes,I,J)
    ac,bs=cache_perturbed_H(A,h_coefficients(genes),delta)
    windows=[np.arange(max(0,k-2)*layer,min(K,k+3)*layer,dtype=np.int32) for k in range(K)]
    result={}
    for src in origins:
        ds=windows[int(src//layer)]
        ds=ds[ds!=src]
        h=h_from_cached(ac,bs,int(src),ds)
        pos=np.flatnonzero(h>0)
        take=min(int(nmax),len(pos))
        selected=pos[np.argpartition(h[pos],-take)[-take:]] if take and take<len(pos) else pos
        picked=ds[selected]
        valid=(~inputs[picked]) & (~(inputs[src]&outputs[picked]))
        result[int(src)]=(set(picked[valid].tolist()),h,ds)
    return result

mutation_rows=[];rng=np.random.default_rng(MUTATION_SEED)
for case in cases:
    genes={k:v.copy() for k,v in case['genome'].items()}
    # Escolha reproduzível das origens e das mutações; papéis mantidos constantes.
    origins=np.flatnonzero(~case['roles'][1])
    origins=np.sort(rng.choice(origins,size=min(STABILITY_SOURCES_PER_GENOME,len(origins)),replace=False))
    for rep in range(MUTATIONS_PER_GENOME):
        mutant={k:v.copy() for k,v in genes.items()}
        gene_index=int(rng.integers(len(mutant['connections'])))
        # Os genes originais são representados como uma sequência de dígitos 0..3.
        old=int(mutant['connections'][gene_index]); step=1 if old==0 else (-1 if old==3 else int(rng.choice([-1,1])))
        mutant['connections'][gene_index]=old+step
        newA=make_A(mutant,*case['size'])
        for mode in MODES:
            delta=position_delta(case['size'],mode,SHUFFLE_SEED+int(case['genome_seed']))
            before=sampled_edges_and_H(case['A'],genes,case['size'],delta,origins,case['roles'])
            after=sampled_edges_and_H(newA,mutant,case['size'],delta,origins,case['roles'])
            js=[];changes=[]
            for src in origins:
                e0,h0,d0=before[int(src)];e1,h1,d1=after[int(src)]
                assert np.array_equal(d0,d1)
                js.append(len(e0&e1)/len(e0|e1) if e0|e1 else 1.)
                changes.append(np.mean(np.abs(h0.astype(np.float64)-h1.astype(np.float64))))
            mutation_rows.append({'case':case['case'],'neurons':len(case['A']),
              'mutation_number':rep,'gene_index':gene_index,'old_digit':old,'new_digit':old+step,
              'mode':mode,'sampled_sources':len(origins),
              'mean_jaccard':float(np.mean(js)),'mean_abs_delta_H':float(np.mean(changes))})
mutation_df=pd.DataFrame(mutation_rows)
mutation_df.to_csv(OUTPUT_DIR/'mutation_stability.csv',index=False)
display(mutation_df.groupby('mode')[['mean_jaccard','mean_abs_delta_H']].agg(['mean','std']).round(5))
fig,axs=plt.subplots(1,2,figsize=(12,4),constrained_layout=True)
means=mutation_df.groupby('mode')[['mean_jaccard','mean_abs_delta_H']].mean().reindex(MODES)
for ax,col,title in zip(axs,['mean_jaccard','mean_abs_delta_H'],['Jaccard dos destinos selecionados (maior é mais estável)','Mudança absoluta média de H (menor é mais estável)']):
    ax.bar(np.arange(len(MODES)),means[col])
    ax.set_xticks(np.arange(len(MODES)),MODES,rotation=25,ha='right')
    ax.set_title(title);ax.grid(axis='y',alpha=.2)
fig.savefig(OUTPUT_DIR/'06_mutation_stability.png',dpi=150);plt.show()
```

## 13. Interpretação e limites

- `random_1`, `random_2`, `random_3`: realizações **independentes por neurônio** de $r(p)\sim\mathcal U(0,1)$; em cada genoma, as sementes são derivadas de `SHUFFLE_SEED`, da semente do genoma e do número da réplica. O mesmo campo fica **congelado** durante a seleção de todas as arestas e as mutações.
- O campo aleatório é **descontínuo na grade** por construção; seus resultados testam o limite de diversidade de fases, não são candidatos a uma topologia espacialmente suave.
- A média e dispersão entre três realizações dão uma primeira noção de variabilidade, **não** garantem precisão estatística arbitrária.
- `radial` reproduz o campo de controle da Parte V; os papéis neuronais continuam fixos.
- A seleção top-$N_{\max}$ pode continuar concentrando muitos neurônios de origem nos mesmos destinos, mesmo com fases independentes.
- O teste de mutação usa campos fixos e origens amostradas; ele não mede estabilidade fenotípica completa.
- As métricas são **antes do resgate**; cobertura não é sinônimo de desempenho cognitivo.


## 14. Teste central: diferença entre radial e campo aleatório, pareada por genoma

Uma réplica individual pode ter desempenho atípico. A tabela abaixo usa a média de três campos uniformes por genoma, comparando ao mesmo controle radial. Também informa variação entre réplicas, cobertura das saídas e a fração de genomas em que o aleatório melhora o isolamento.

```python
rnd=df[df['mode'].str.startswith('random_')].copy()
rad=df[df['mode']=='radial'].set_index('case')
agg=rnd.groupby('case',as_index=True).agg(
    neurons=('neurons','first'),random_mean_isolation=('isolation_pct','mean'),
    random_sd_isolation=('isolation_pct','std'),random_min_isolation=('isolation_pct','min'),
    random_max_isolation=('isolation_pct','max'),random_mean_output_isolated=('isolated_output','mean'),
    random_mean_gini=('gini_incoming','mean'),random_mean_H=('mean_H','mean'))
agg['radial_isolation']=rad['isolation_pct']
agg['random_minus_radial_pp']=agg['random_mean_isolation']-agg['radial_isolation']
agg['radial_output_isolated']=rad['isolated_output']
agg.to_csv(OUTPUT_DIR/'random_vs_radial_paired.csv')
display(agg.round(4))
print('Isolamento médio radial:',round(float(rad['isolation_pct'].mean()),4))
print('Isolamento médio aleatório (3 réplicas, peso igual por genoma):',round(float(agg['random_mean_isolation'].mean()),4))
print('Diferença aleatório − radial (p.p.):',round(float(agg['random_minus_radial_pp'].mean()),4))
print('Genomas onde aleatório teve menor isolamento:',int((agg['random_minus_radial_pp']<0).sum()),'de',len(agg))
```

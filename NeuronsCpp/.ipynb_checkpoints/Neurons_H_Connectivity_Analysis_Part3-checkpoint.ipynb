# Neurons — Parte III: o campo $A$ está causando o isolamento?

**Experimento controlado, sem alterar o projeto Godot.** Substituímos somente os valores espaciais de $A$ fornecidos à função $H$, preservando o genoma que determina $H$, $N_{\max}$, o universo de candidatos, a seleção dos maiores $H>0$ e a filtragem de conexões proibidas.

**Questão central:** o isolamento elevado decorre da distribuição/correlação espacial de $A$, ou persiste mesmo com um campo artificialmente bem distribuído?

Distinguimos quatro condições:

1. **Original**: $A(i,j,k,x)$ do Neurons.
2. **Permutado**: exatamente os mesmos valores de $A$, mas embaralhados globalmente — preserva a distribuição, destrói sua organização espacial.
3. **Uniforme independente**: novos valores pseudoaleatórios $U[-1,1]$ — distribuição aproximadamente uniforme, sem correlações espaciais planejadas.
4. **Uniforme estratificado**: valores igualmente espaçados em $[-1,1]$, permutados — distribuição marginal uniforme com cobertura garantida.

Cada condição usa o **mesmo genoma e mesmas dimensões**, com duas avaliações: **papéis fixos** (inputs/outputs determinados pelo $A$ original) e **papéis reatribuídos** (a seleção dos tipos depende do novo $A$, como faria o construtor). As duas avaliações evitam conclusões confundidas pela reclassificação neuronal.

**Importante:** isolamento é medido **após a remoção das conexões proibidas e antes do resgate**; estas simulações não modelam o resgate, a plasticidade ou a dinâmica. A escolha parcial por `argpartition` tem a mesma regra de top-$K$, com possíveis diferenças apenas em empates.


## 0. Parâmetros e os 11 genomas reais da Parte I

```python
import json, time, zipfile, io
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

SEED = 20261025
PROFILE = 'quick'  # 'quick' = dois casos 10³ + dois 22²×10; 'full' = 11 casos das partes anteriores
INPUT_OUTPUT_COUNT = (10, 10)
REPEATS = 2 if PROFILE == 'quick' else 3   # réplicas para controles aleatórios; A original é executado uma única vez
INCLUDE_LARGEST = True # no full, processa também 32²×10 e 42²×10 (leva mais tempo)
OUTPUT_DIR = Path('neurons_h_part3_results')
OUTPUT_DIR.mkdir(exist_ok=True)
PREVIOUS_ZIP = Path('Neurons_H_Analysis_Results.zip')
assert PREVIOUS_ZIP.exists(), 'Coloque Neurons_H_Analysis_Results.zip ao lado do notebook.'
with zipfile.ZipFile(PREVIOUS_ZIP) as archive:
    records=json.loads(archive.read('genomes.json'))
if PROFILE=='quick':
    selected=[r for r in records if r['size'][0] in (10,22)][:4]
    selected=[r for r in records if r['size'][0]==10][:2]+[r for r in records if r['size'][0]==22][:2]
else:
    selected=records if INCLUDE_LARGEST else [r for r in records if r['size'][0]<=22]
print('Casos:',len(selected),'| réplicas por controle:',REPEATS)
print('Dimensões:',[tuple(r['size']) for r in selected])
```

## 1. Reprodução das funções matemáticas de referência

```python
PI32 = np.float32(np.pi)
W8 = np.sqrt(np.arange(1, 9, dtype=np.float32))
W64 = np.sqrt(np.arange(1, 65, dtype=np.float32))

def random_genome(seed):
    rng = np.random.default_rng(seed)
    # O construtor C++ gera dígitos independentes e uniformes de 0 a 3.
    return {"connections": rng.integers(0,4,64,dtype=np.int8),
            "beta": rng.integers(0,4,8,dtype=np.int8),
            "max_connections": rng.integers(0,4,8,dtype=np.int8)}

def parse_genome(dictionary):
    lengths = {"connections": 64, "beta": 8, "max_connections": 8}
    ans = {}
    for name, count in lengths.items():
        value = dictionary[name]
        if isinstance(value, str):
            if len(value) != count or not set(value) <= set("0123"):
                raise ValueError(f"{name}: esperado {count} dígitos quaternários")
            ans[name] = np.array([int(ch) for ch in value], dtype=np.int8)
        else:
            ans[name] = np.asarray(value, dtype=np.int8)
            if ans[name].shape != (count,) or not np.isin(ans[name], [0,1,2,3]).all():
                raise ValueError(f"{name}: formato inválido")
    return ans

def genome_text(g):
    return {key: "".join(map(str, val)) for key, val in g.items()}

def beta_max(g, I, J):
    beta = np.float32(1.0 + np.sum(W8*g["beta"],dtype=np.float32)
                     / np.sum(3*W8,dtype=np.float32) * np.float32((I*J/16.)**(1/3)))
    maxconn = int(np.floor(np.sum(W8*np.power(beta, g["max_connections"].astype(np.float32)),dtype=np.float32)))
    return float(beta), maxconn

def h_coefficients(g):
    digits = g["connections"]
    return np.array([np.sum(W64[digits == d],dtype=np.float32) for d in range(4)],dtype=np.float32) * (np.float32(3)*PI32 / (np.float32(4)*np.sum(W64,dtype=np.float32)))

def make_A(g, I, J, K):
    # A[k,j,i], linearizado como i + I*j + I*J*k.
    coeff = (np.float32(2)/np.float32(3))*g["connections"].astype(np.float32)-np.float32(1)
    cx = np.cos(PI32*np.arange(4,dtype=np.float32)[:,None]*
                (np.arange(I,dtype=np.float32)[None,:]/np.float32(I-1))).astype(np.float32)
    cy = np.cos(PI32*np.arange(4,dtype=np.float32)[:,None]*
                (np.arange(J,dtype=np.float32)[None,:]/np.float32(J-1))).astype(np.float32)
    cz = np.cos(PI32*np.arange(4,dtype=np.float32)[:,None]*
                (np.arange(K,dtype=np.float32)[None,:]/np.float32(K-1))).astype(np.float32)
    field = np.zeros((K,J,I), dtype=np.float32)
    for r in range(4):
        for s in range(4):
            for q in range(4):
                field += coeff[q + 4*s + 16*r] * cx[r][None,None,:] * cy[s][None,:,None] * cz[q][:,None,None]
    return np.tanh(field / np.float32(8)).astype(np.float32).ravel()

def h_direct(a, b, c):
    z = np.float32(np.pi/16)
    for m in range(1, 5):
        z = z + c[m-1]*np.cos(np.float32(m)*a)*np.sin(np.float32(m)*b)
    return np.sin(z)

def make_fast_h(A, coeffs):
    harmonics = np.arange(1,5,dtype=np.float32)[:,None]
    ac = (coeffs[:,None]*np.cos(harmonics*A[None,:])).astype(np.float32)
    bs = np.sin(harmonics*A[None,:]).astype(np.float32)
    return ac, bs

def h_from_cached(ac, bs, src, target_indices):
    z = np.full(len(target_indices), np.float32(np.pi/16), dtype=np.float32)
    for m in range(4):
        z += ac[m, src] * bs[m, target_indices]
    return np.sin(z).astype(np.float32)


# Validação da identidade trigonométrica agrupada
_test = random_genome(71)
_A = make_A(_test,5,5,4)
_coeff = h_coefficients(_test)
_ac,_bs = make_fast_h(_A,_coeff)
_ix=np.arange(0,len(_A),3)
assert np.allclose(h_from_cached(_ac,_bs,0,_ix),h_direct(_A[0],_A[_ix],_coeff),atol=1e-6)
print('H e A: verificação numérica OK')
```

## 2. Seleção idêntica em todas as condições

A função abaixo percorre **todas as origens** e **todos os destinos das camadas $|\Delta k|\le2$** (exceto o próprio neurônio), seleciona até $N_{\max}$ maiores valores positivos e então remove as conexões proibidas. O uso de índices lineares é apenas uma representação do mesmo conjunto de posições, não uma perturbação introduzida pelo experimento.

Medimos também Gini das entradas, participação dos 10% de destinos mais conectados e taxa de pares com $H>0$.

```python
def flags_by_rank(A, I, J, K, n_inputs, n_outputs):
    N = len(A)
    if n_inputs+n_outputs > N:
        raise ValueError("Número de inputs+outputs maior que número de neurônios")
    by_layer = []
    for k in range(K):
        ix = np.arange(k*I*J,(k+1)*I*J,dtype=np.int32)
        by_layer.append(ix[np.argsort(A[ix],kind="quicksort")])
    # rank 0 de cada camada; depois rank 1 de cada camada; ...
    permutation = np.stack(by_layer,axis=1).ravel()
    is_input = np.zeros(N,dtype=bool); is_output = np.zeros(N,dtype=bool)
    is_input[permutation[:n_inputs]] = True
    is_output[permutation[-n_outputs:]] = True
    return is_input, is_output

def gini_nonnegative(data):
    x = np.sort(np.asarray(data,dtype=np.float64)); n=len(x)
    if n == 0 or np.sum(x) == 0: return 0.
    return float((2*np.dot(np.arange(1,n+1),x)/(n*np.sum(x))) - (n+1)/n)

def metrics(incoming, outgoing, is_input, is_output):
    internal = ~(is_input | is_output)
    counts = {"internal":int(np.count_nonzero((incoming==0)&internal)),
              "output":int(np.count_nonzero((incoming==0)&is_output)),
              "input":int(np.count_nonzero((outgoing==0)&is_input))}
    N=len(incoming)
    eligible = ~is_input
    vals = incoming[eligible]
    tenpct = max(1,int(np.ceil(.1*len(vals))))
    total = np.sum(vals)
    top_share = float(np.partition(vals,-tenpct)[-tenpct:].sum()/total) if total else np.nan
    return {"isolated_total":sum(counts.values()),"isolation_pct":100.*sum(counts.values())/N,
            "isolated_internal":counts["internal"],"isolated_output":counts["output"],"isolated_input":counts["input"],
            "mean_incoming":float(np.mean(vals)),"median_incoming":float(np.median(vals)),
            "gini_incoming":gini_nonnegative(vals),"top_10pct_incoming_share":top_share,
            "zero_incoming_eligible_pct":100.*np.mean(vals==0),"total_valid_connections":int(np.sum(vals))}


def simulate_field(A, g, size, fixed_roles=None):
    I,J,K=map(int,size); N=I*J*K
    if fixed_roles is None:
        is_input,is_output=flags_by_rank(A,I,J,K,*INPUT_OUTPUT_COUNT)
    else:
        is_input,is_output=fixed_roles
    _, nmax=beta_max(g,I,J)
    coeff=h_coefficients(g)
    ac,bs=make_fast_h(A,coeff)
    incoming=np.zeros(N,dtype=np.int32)
    outgoing=np.zeros(N,dtype=np.int32)
    layers=[np.arange(k*I*J,(k+1)*I*J,dtype=np.int32) for k in range(K)]
    windows=[np.concatenate(layers[max(0,k-2):min(K,k+3)]) for k in range(K)]
    sources=np.flatnonzero(~is_output)
    npositive=0; ncandidates=0; selected_h_sum=0.; selected_h_n=0
    for src in sources:
        cand=windows[int(src//(I*J))]
        targets=cand[cand!=src]
        h=h_from_cached(ac,bs,int(src),targets)
        good=np.flatnonzero(h>0)
        npositive+=len(good);ncandidates+=len(targets)
        take=min(nmax,len(good))
        if not take: continue
        if take<len(good):
            picked=good[np.argpartition(h[good], -take)[-take:]]
        else: picked=good
        dest=targets[picked]
        valid=(~is_input[dest]) & (~(is_input[src] & is_output[dest]))
        keep=dest[valid]
        incoming[keep]+=1
        outgoing[src]+=len(keep)
        selected_h_sum+=float(h[picked][valid].sum(dtype=np.float64))
        selected_h_n+=int(valid.sum())
    m=metrics(incoming,outgoing,is_input,is_output)
    return {**m,'Nmax':int(nmax), 'H_positive_pct':100*npositive/ncandidates,
            'mean_selected_H':selected_h_sum/selected_h_n if selected_h_n else np.nan,
            'incoming':incoming,'outgoing':outgoing}

def produce_field(A_original, kind, seed):
    rng=np.random.default_rng(seed)
    n=len(A_original)
    if kind=='original':return A_original.copy()
    if kind=='permuted':return A_original[rng.permutation(n)].copy()
    if kind=='iid_uniform':return rng.uniform(-1,1,n).astype(np.float32)
    if kind=='stratified':
        vals=np.linspace(-1,1,n,dtype=np.float32)
        return vals[rng.permutation(n)].copy()
    raise ValueError(kind)

# Ensaios pequenos para os dois critérios de escolha de papéis
G0=random_genome(502)
A0=make_A(G0,5,5,4)
P0=produce_field(A0,'permuted',905)
assert np.array_equal(np.sort(A0),np.sort(P0))
assert np.array_equal(produce_field(A0,'original',1),A0)
for name in ('original','permuted','iid_uniform','stratified'):
    x=produce_field(A0,name,606)
    t=simulate_field(x,G0,(5,5,4))
    assert 0 <= t['isolation_pct'] <= 100
print('Testes de sanidade da seleção/filtragem: OK')
```

## 3. Experimento pareado nos mesmos genomas

Para cada genoma, construímos o $A$ original e registramos duas matrizes de resultados:

- **Papéis fixos**: controla o efeito de alterar quem é input ou output. É a comparação mais limpa da transformação $A\mapsto H$.
- **Papéis reatribuídos**: aproxima o que aconteceria se trocássemos a definição de $A$ no construtor.

Os campos artificiais são gerados com sementes registradas. A réplica original não é repetida porque é determinística. O experimento pode levar bastante tempo para os maiores cérebros; o console mostra cada condição concluída.

```python
results=[]; a_descriptors=[]; degree_profiles={}; started=time.perf_counter()
MODES=['original','permuted','iid_uniform','stratified']
for idx,r in enumerate(selected):
    g=parse_genome(r['genes']); size=tuple(r['size']); I,J,K=size
    A0=make_A(g,*size)
    roles0=flags_by_rank(A0,*size,*INPUT_OUTPUT_COUNT)
    original_id=f"{I}x{J}x{K}_g{idx+1:02d}"
    for mode in MODES:
        for rep in range(1 if mode=='original' else REPEATS):
            seed=int(SEED+idx*10000+MODES.index(mode)*100+rep)
            A=produce_field(A0,mode,seed)
            a_descriptors.append(dict(case=original_id,mode=mode,rep=rep,
                        A_mean=float(A.mean()), A_sd=float(A.std()),
                        A_min=float(A.min()),A_max=float(A.max()),
                        A_abs_mean=float(np.abs(A).mean()),
                        A_neighbor_corr=float(np.corrcoef(A.reshape(K,J,I)[:,:,:-1].ravel(),
                                         A.reshape(K,J,I)[:,:,1:].ravel())[0,1])))
            for role_style in ['fixed','reassigned']:
                tic=time.perf_counter()
                m=simulate_field(A,g,size,roles0 if role_style=='fixed' else None)
                row=dict(case=original_id,genome_seed=r['seed'],size=I,N=len(A),
                         mode=mode,rep=rep,field_seed=seed,roles=role_style,
                         runtime_seconds=time.perf_counter()-tic)
                row.update({k:v for k,v in m.items() if k not in ('incoming','outgoing')})
                results.append(row)
                if mode=='original' or (rep==0 and role_style=='fixed'):
                    degree_profiles[(original_id,mode,role_style)] = m['incoming']
                print(f"{original_id} | {mode:13} | {role_style:10} | "
                      f"isolados {m['isolation_pct']:6.2f}% | "
                      f"Gini {m['gini_incoming']:.3f} | {row['runtime_seconds']:.1f}s",flush=True)
print('Tempo total:',round(time.perf_counter()-started,1),'s')
df=pd.DataFrame(results)
a_df=pd.DataFrame(a_descriptors)
df.to_csv(OUTPUT_DIR/'per_case_field_results.csv',index=False)
a_df.to_csv(OUTPUT_DIR/'field_distribution_diagnostics.csv',index=False)
display(df.head(10).round(3))
```

## 4. Comparações por dimensão, com controle pareado

Calculamos a diferença **controle menos original** para cada genoma. Uma diferença negativa de isolamento indica redução da necessidade de resgate. As réplicas do mesmo genoma **não contam como genomas independentes**: fazemos a média das réplicas primeiro.

O foco não é apenas diminuir o isolamento: é necessário verificar também o número de conexões finais, a concentração e a afinidade média selecionada.

```python
aggregated=df.groupby(['case','N','size','mode','roles'],as_index=False).agg(
    isolation_pct=('isolation_pct','mean'),gini=('gini_incoming','mean'),
    top10=('top_10pct_incoming_share','mean'),
    edges=('total_valid_connections','mean'),
    positive_pct=('H_positive_pct','mean'),
    mean_selected_H=('mean_selected_H','mean'))
orig=aggregated[aggregated['mode']=='original'].drop(columns='mode')
paired=aggregated.merge(orig,on=['case','N','size','roles'],suffixes=('','_original'))
for metric in ['isolation_pct','gini','top10','edges','positive_pct','mean_selected_H']:
    paired[metric+'_change']=paired[metric]-paired[metric+'_original']
paired.to_csv(OUTPUT_DIR/'paired_changes.csv',index=False)
summary=paired.groupby(['size','mode','roles'],as_index=False).agg(
    brains=('case','nunique'),isolation=('isolation_pct','mean'),
    isolation_delta_pp=('isolation_pct_change','mean'),
    gini=('gini','mean'),top10=('top10','mean'),
    mean_H=('mean_selected_H','mean'),edges=('edges','mean'))
summary.to_csv(OUTPUT_DIR/'summary_by_size.csv',index=False)
for role_style in ['fixed','reassigned']:
    print('\nPAPÉIS',role_style.upper())
    display(summary[summary.roles==role_style].round(3))
```

## 5. Gráficos: efeito sobre cobertura e desigualdade das entradas

```python
fig,axs=plt.subplots(1,2,figsize=(15,5),constrained_layout=True)
for mode in MODES:
    for ax,metric,unit in [(axs[0],'isolation','% isolados'),(axs[1],'gini','Gini de entradas')]:
        s=summary[(summary['mode']==mode)&(summary['roles']=='fixed')].sort_values('size')
        ax.plot(s['size']**2*10,s[metric],marker='o',label=mode)
        ax.set_xlabel('Neurônios (I × J × 10)');ax.set_ylabel(unit);ax.grid(alpha=.2)
axs[0].set_title('Isolamento por campo A (papéis fixos)')
axs[1].set_title('Concentração por campo A (papéis fixos)')
axs[0].legend()
fig.savefig(OUTPUT_DIR/'01_isolation_and_gini.png',dpi=160)
plt.show()
fig,axs=plt.subplots(1,2,figsize=(15,5),constrained_layout=True)
for mode in MODES:
    for ax,metric,label in [(axs[0],'isolation_delta_pp','Diferença no isolamento (pp)'),
                            (axs[1],'mean_H','Afinidade H média das conexões')]:
        s=summary[(summary['mode']==mode)&(summary['roles']=='fixed')].sort_values('size')
        ax.plot(s['size']**2*10,s[metric],marker='o',label=mode)
        ax.set_xlabel('Neurônios'); ax.set_ylabel(label);ax.grid(alpha=.2)
axs[0].axhline(0,color='gray',linestyle='--',linewidth=1)
axs[0].legend();fig.savefig(OUTPUT_DIR/'02_paired_change_and_affinity.png',dpi=160)
plt.show()
```

## 6. Qual propriedade de $A$ importa: distribuição marginal ou correlação espacial?

O campo **permutado** preserva exatamente o histograma de $A$ (até os empates), mas muda suas posições. Assim:

- Se **permutar** reduzir muito o isolamento, a disposição/correlação espacial de $A$ contribui.
- Se **uniformizar** reduzir, mas permutar não, a distribuição marginal de $A$ provavelmente contribui.
- Se ambos persistirem com isolamento alto, a estrutura de $H$ e a seleção extrema continuam suspeitas — ainda não são a única causa possível.

A comparação entre **papéis fixos** e **reatribuídos** informa quanto da diferença passa pela escolha dos neurônios de entrada/saída.

```python
fig,axs=plt.subplots(1,2,figsize=(14,5),constrained_layout=True)
for mode in MODES:
    s=a_df[a_df['mode']==mode]
    axs[0].hist(s['A_sd'],bins=12,alpha=.5,label=mode)
    axs[1].scatter(s['A_neighbor_corr'],s['A_sd'],alpha=.7,label=mode)
axs[0].set(xlabel='Desvio-padrão de A',ylabel='Campos',title='Amplitude da distribuição de A')
axs[1].set(xlabel='Correlação horizontal entre vizinhos de A',ylabel='Desvio-padrão de A',title='Amplitude versus estrutura espacial')
for ax in axs:ax.grid(alpha=.2)
axs[0].legend();fig.savefig(OUTPUT_DIR/'03_field_diagnostics.png',dpi=160);plt.show()
fig,ax=plt.subplots(figsize=(12,5),constrained_layout=True)
for role in ['fixed','reassigned']:
    for mode in MODES[1:]:
        ss=summary[(summary.roles==role)&(summary['mode']==mode)].sort_values('size')
        ax.plot(ss['size']**2*10,ss['isolation_delta_pp'],marker='o',label=f'{mode} / {role}')
ax.axhline(0,color='gray',linestyle='--');ax.set(xlabel='Neurônios',ylabel='Mudança no isolamento (pp)',title='Efeito da reclassificação de input/output')
ax.legend(fontsize=8);ax.grid(alpha=.2);fig.savefig(OUTPUT_DIR/'04_role_sensitivity.png',dpi=160);plt.show()
```

## 8. Como interpretar as conclusões

Antes de reformular $A$ ou $H$, considere:

- **Isolamento muito menor só após uniformizar $A$**: evidência de que a concentração dos valores de $A$ importa.
- **Isolamento muito menor só após permutar $A$**: evidência de que a organização espacial de $A$ importa; o histograma não basta.
- **Redução ao mudar a classificação dos neurônios, mas não com papéis fixos**: efeitos de input/output estão confundindo o diagnóstico.
- **Isolamento ainda elevado com $A$ uniforme e papéis fixos**: $H$ (e sua seleção top-$K$) pode ser concentradora mesmo com campo $A$ bem distribuído.

**Advertência:** o `A` original é um campo espacial *suave*, enquanto os controles uniformes são espacialmente descontínuos. Melhor cobertura não significa necessariamente conectividade mais útil para cognição. A amostra de 11 genomas é exploratória; use as mudanças pareadas, não apenas médias misturadas entre dimensões.

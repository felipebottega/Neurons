# Neurons — Parte XI: busca sistemática da formulação de $H$

**Problema central.** Encontrar, entre muitas expressões matemáticas interpretáveis, funções

$$\boxed{H_{a,\eta,F}(p,q)=\sin\!\left(a+\frac{\eta}{C}\sum_{t=0}^{63}\sqrt{1+t}\;F\left(U(x_t)A_p,U(x_t)A_q\right)\right)}$$

que reduzam o isolamento **antes do resgate**, sem sacrificar o número de arestas, a intensidade dos pesos, a cobertura de *outputs* ou a estabilidade diante de mutações. Aqui $C=\sum_t\sqrt{1+t}$, $U(x_t)=1+x_t$ e $b=\eta/C$. A busca **inclui exatamente $a=0$ e $b=1/C$** e a expressão original $a=\pi/16$, $\eta=3\pi/4$, $F(X,Y)=\cos X\sin Y$.

Não modificamos $A$, os genes, $\beta$, $N_{\max}$, os papéis input/output ou a seleção determinística **dos maiores $H>0$**. Não alteramos o algoritmo de resgate: medimos os isolados **antes** dele.

### Estratégia experimental — evitando escolher um vencedor ilusório

1. **Varredura ampla e pareada** de operações $F$, deslocamentos $a$ e escalas $\eta$; vários genomas, fontes espalhadas pelas camadas e exatamente os mesmos candidatos em todos os modelos.
2. **Seleção sem vazamento:** escolher fórmulas usando apenas genomas de descoberta. Quatro genomas, um por tamanho, ficam fora da triagem no perfil `full`.
3. **Avaliação exata** de um conjunto de finalistas nos 11 genomas, com todos os neurônios de origem, sem extrapolar isolamento a partir da amostragem.
4. **Diagnósticos incluídos:** pesos positivos fracos, afinidades quase constantes, similaridade dos rankings, causas do isolamento, mutação genética, robustez por tamanho e generalização no conjunto reservado.

A triagem estima *cobertura sob um número fixo de origens amostradas*, **não** o percentual de isolamento completo. O ranking de triagem é apenas um filtro para escolher quais fórmulas merecem a simulação completa.

### Observação sobre expressões equivalentes

Por exemplo, $\cos X\sin Y+\sin X\cos Y=\sin(X+Y)$. Essa identidade **não** torna a operação equivalente ao modelo original: muda sua dependência das coordenadas de origem e destino. Operações aditivas ($\cos X+\sin Y$ etc.) podem ter amplitude até 2, enquanto produtos chegam até 1; por isso a escala $\eta$ será explorada explicitamente, sem normalização oculta.

```python
import io, json, zipfile, math, time, warnings
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

PROFILE = 'quick'                 # 'quick' valida o fluxo; 'full' executa 11 genomas
BASE_SEED = 20261025
N_INPUTS = N_OUTPUTS = 10
SCREEN_SOURCES_PER_GENOME = 24 if PROFILE=='quick' else 56
FINALISTS_MAX = 8 if PROFILE=='quick' else 18
RANK_SOURCES = 26
MUTATIONS_PER_GENOME = 1 if PROFILE=='quick' else 2
MUTATION_SOURCES = 12 if PROFILE=='quick' else 24
RESULT_DIR = Path('neurons_h_part11_results');RESULT_DIR.mkdir(exist_ok=True)
RESULT_ZIP = Path('Neurons_H_Part11_Results.zip')
SOURCE_ZIP = Path('Neurons_H_Analysis_Results.zip')
assert SOURCE_ZIP.exists(), 'Coloque Neurons_H_Analysis_Results.zip na pasta do notebook'
with zipfile.ZipFile(SOURCE_ZIP) as zf:
    RECORDS = json.loads(zf.read('genomes.json'))
    REFERENCE = pd.read_csv(io.BytesIO(zf.read('per_brain_metrics.csv')))
assert len(RECORDS)==11

# Divisão pré-fixada, independente dos valores de H: último genoma de cada tamanho reservado.
by_size={}
for r in RECORDS:by_size.setdefault(tuple(r['size']),[]).append(r)
HOLDOUT_SEEDS={items[-1]['seed'] for items in by_size.values()}
DISCOVERY=[r for r in RECORDS if r['seed'] not in HOLDOUT_SEEDS]
HOLDOUT=[r for r in RECORDS if r['seed'] in HOLDOUT_SEEDS]
if PROFILE=='quick':
    # Inclui 1k, 4.840 e 10.240 no screening; 3 genomas na validação completa.
    SCREEN_RECORDS=[DISCOVERY[0], DISCOVERY[4], DISCOVERY[-1]]
    FINAL_RECORDS=[DISCOVERY[0], DISCOVERY[4], HOLDOUT[0]]
elif PROFILE=='full':
    SCREEN_RECORDS=DISCOVERY
    FINAL_RECORDS=RECORDS
else:raise ValueError('PROFILE deve ser quick ou full')

OPS={
 'cos*sin':'cos(X) sin(Y)', 'sin*cos':'sin(X) cos(Y)',
 'sin*sin':'sin(X) sin(Y)', 'cos*cos':'cos(X) cos(Y)',
 'cos+sin':'cos(X) + sin(Y)', 'sin+cos':'sin(X) + cos(Y)',
 'sin+sin':'sin(X) + sin(Y)', 'cos+cos':'cos(X) + cos(Y)',
 'cos-sin':'cos(X) - sin(Y)', 'sin-cos':'sin(X) - cos(Y)',
 'sin-sin':'sin(X) - sin(Y)', 'cos-cos':'cos(X) - cos(Y)',
 'cs+sc':'sin(X+Y) [=CS+SC]', 'cs-sc':'sin(Y-X) [=CS-SC]',
 'cc+ss':'cos(X-Y) [=CC+SS]', 'cc-ss':'cos(X+Y) [=CC-SS]',
}
A_VALUES = [-np.pi/2,-np.pi/4,-np.pi/8,0.,np.pi/16,np.pi/8,np.pi/4,3*np.pi/8,np.pi/2,3*np.pi/4]
ETA_VALUES = [-3.,-3*np.pi/4,-1.,.5,1.,1.5,2.,3*np.pi/4,3.,4.]
PARAMS=pd.DataFrame([{'op':op,'a':a,'eta':eta,'b_times_C':eta,
                      'formula':OPS[op], 'id':f'p{idx:04d}'}
    for idx,(op,a,eta) in enumerate((op,a,e) for op in OPS for a in A_VALUES for e in ETA_VALUES)])
assert any((PARAMS.op=='cos*sin') & np.isclose(PARAMS.a,np.pi/16) & np.isclose(PARAMS.eta,3*np.pi/4))
assert np.isclose(PARAMS.a,0).any() and np.isclose(PARAMS.eta,1).any()
BASE=PARAMS.loc[(PARAMS.op=='cos*sin') & np.isclose(PARAMS.a,np.pi/16) & np.isclose(PARAMS.eta,3*np.pi/4),'id'].iloc[0]
print(f'{len(OPS)} operações × {len(A_VALUES)} shifts × {len(ETA_VALUES)} escalas = {len(PARAMS)} fórmulas')
print('Baseline original:',BASE,'| Screening:',len(SCREEN_RECORDS),'genomas | final:',len(FINAL_RECORDS))
display(PARAMS.head())
```

## 1. Reprodução do cérebro original

Usamos a implementação das Partes IX–X, preservando a ordem dos índices $n=i+Ij+IJk$ e os grupos harmônicos $m=1,2,3,4$. O agrupamento

$$\frac1C\sum_t w_t F(U_t A_p,U_t A_q)=\sum_{m=1}^4 g_mF(mA_p,mA_q),\qquad g_m=\frac{\sum_{t:U_t=m}w_t}{C}$$

é **exato em aritmética real**, e evita avaliar 64 senos e cossenos para cada par de neurônios. As discrepâncias de arredondamento float32 na fronteira top-$N_{\max}$ serão verificadas contra o baseline da Parte I.

```python
PI32 = np.float32(np.pi)
W8 = np.sqrt(np.arange(1, 9, dtype=np.float32))
W64 = np.sqrt(np.arange(1, 65, dtype=np.float32))

def random_genome(seed):
    rng = np.random.default_rng(seed)
    # O construtor C++ gera dígitos independentes e uniformes de 0 a 3.
    return {"connections": rng.integers(0,4,64,dtype=np.int8),
            "beta": rng.integers(0,4,8,dtype=np.int8),
            "max_connections": rng.integers(0,4,8,dtype=np.int8)}

def parse_genome(dictionary):
    lengths = {"connections": 64, "beta": 8, "max_connections": 8}
    ans = {}
    for name, count in lengths.items():
        value = dictionary[name]
        if isinstance(value, str):
            if len(value) != count or not set(value) <= set("0123"):
                raise ValueError(f"{name}: esperado {count} dígitos quaternários")
            ans[name] = np.array([int(ch) for ch in value], dtype=np.int8)
        else:
            ans[name] = np.asarray(value, dtype=np.int8)
            if ans[name].shape != (count,) or not np.isin(ans[name], [0,1,2,3]).all():
                raise ValueError(f"{name}: formato inválido")
    return ans

def genome_text(g):
    return {key: "".join(map(str, val)) for key, val in g.items()}

def beta_max(g, I, J):
    beta = np.float32(1.0 + np.sum(W8*g["beta"],dtype=np.float32)
                     / np.sum(3*W8,dtype=np.float32) * np.float32((I*J/16.)**(1/3)))
    maxconn = int(np.floor(np.sum(W8*np.power(beta, g["max_connections"].astype(np.float32)),dtype=np.float32)))
    return float(beta), maxconn

def h_coefficients(g):
    digits = g["connections"]
    return np.array([np.sum(W64[digits == d],dtype=np.float32) for d in range(4)],dtype=np.float32) * (np.float32(3)*PI32 / (np.float32(4)*np.sum(W64,dtype=np.float32)))

def make_A(g, I, J, K):
    # A[k,j,i], linearizado como i + I*j + I*J*k.
    coeff = (np.float32(2)/np.float32(3))*g["connections"].astype(np.float32)-np.float32(1)
    cx = np.cos(PI32*np.arange(4,dtype=np.float32)[:,None]*
                (np.arange(I,dtype=np.float32)[None,:]/np.float32(I-1))).astype(np.float32)
    cy = np.cos(PI32*np.arange(4,dtype=np.float32)[:,None]*
                (np.arange(J,dtype=np.float32)[None,:]/np.float32(J-1))).astype(np.float32)
    cz = np.cos(PI32*np.arange(4,dtype=np.float32)[:,None]*
                (np.arange(K,dtype=np.float32)[None,:]/np.float32(K-1))).astype(np.float32)
    field = np.zeros((K,J,I), dtype=np.float32)
    for r in range(4):
        for s in range(4):
            for q in range(4):
                field += coeff[q + 4*s + 16*r] * cx[r][None,None,:] * cy[s][None,:,None] * cz[q][:,None,None]
    return np.tanh(field / np.float32(8)).astype(np.float32).ravel()


def flags_by_rank(A, I, J, K, n_inputs, n_outputs):
    N = len(A)
    if n_inputs+n_outputs > N:
        raise ValueError("Número de inputs+outputs maior que número de neurônios")
    by_layer = []
    for k in range(K):
        ix = np.arange(k*I*J,(k+1)*I*J,dtype=np.int32)
        by_layer.append(ix[np.argsort(A[ix],kind="quicksort")])
    # rank 0 de cada camada; depois rank 1 de cada camada; ...
    permutation = np.stack(by_layer,axis=1).ravel()
    is_input = np.zeros(N,dtype=bool); is_output = np.zeros(N,dtype=bool)
    is_input[permutation[:n_inputs]] = True
    is_output[permutation[-n_outputs:]] = True
    return is_input, is_output

def gini_nonnegative(data):
    x = np.sort(np.asarray(data,dtype=np.float64)); n=len(x)
    if n == 0 or np.sum(x) == 0: return 0.
    return float((2*np.dot(np.arange(1,n+1),x)/(n*np.sum(x))) - (n+1)/n)

def metrics(incoming, outgoing, is_input, is_output):
    internal = ~(is_input | is_output)
    counts = {"internal":int(np.count_nonzero((incoming==0)&internal)),
              "output":int(np.count_nonzero((incoming==0)&is_output)),
              "input":int(np.count_nonzero((outgoing==0)&is_input))}
    N=len(incoming)
    eligible = ~is_input
    vals = incoming[eligible]
    tenpct = max(1,int(np.ceil(.1*len(vals))))
    total = np.sum(vals)
    top_share = float(np.partition(vals,-tenpct)[-tenpct:].sum()/total) if total else np.nan
    return {"isolated_total":sum(counts.values()),"isolation_pct":100.*sum(counts.values())/N,
            "isolated_internal":counts["internal"],"isolated_output":counts["output"],"isolated_input":counts["input"],
            "mean_incoming":float(np.mean(vals)),"median_incoming":float(np.median(vals)),
            "gini_incoming":gini_nonnegative(vals),"top_10pct_incoming_share":top_share,
            "zero_incoming_eligible_pct":100.*np.mean(vals==0),"total_valid_connections":int(np.sum(vals))}
```

## 2. Operações $F(X,Y)$: avaliação vetorizada e testes algébricos

```python
M=np.arange(1,5,dtype=np.float32)
C=float(np.sum(W64,dtype=np.float32))

def get_case(record):
    I,J,K=tuple(record['size']);g=parse_genome(record['genes'])
    A=make_A(g,I,J,K);inp,out=flags_by_rank(A,I,J,K,N_INPUTS,N_OUTPUTS)
    beta,nmax=beta_max(g,I,J)
    gw=np.array([np.sum(W64[g['connections']==d],dtype=np.float32) for d in range(4)],dtype=np.float32)
    weights=gw/np.sum(W64,dtype=np.float32)
    cos=np.cos(M[:,None]*A[None,:]).astype(np.float32)
    sin=np.sin(M[:,None]*A[None,:]).astype(np.float32)
    return {'seed':record['seed'],'case':f'{I}x{J}x{K}_{record["seed"]}',
            'size':(I,J,K),'genes':g,'A':A,'input':inp,'output':out,
            'beta':beta,'nmax':nmax,'w':weights,'cos':cos,'sin':sin}

def features(case,src,dst):
    w=case['w'];cx=case['cos'][:,src];sx=case['sin'][:,src]
    cy=case['cos'][:,dst];sy=case['sin'][:,dst]
    cs=(w*cx)@sy;sc=(w*sx)@cy;ss=(w*sx)@sy;cc=(w*cx)@cy
    cp=float(w@cx);sp=float(w@sx)
    cq=w@cy;sq=w@sy
    return {
        'cos*sin':cs, 'sin*cos':sc, 'sin*sin':ss, 'cos*cos':cc,
        'cos+sin':cp+sq, 'sin+cos':sp+cq,
        'sin+sin':sp+sq, 'cos+cos':cp+cq,
        'cos-sin':cp-sq, 'sin-cos':sp-cq,
        'sin-sin':sp-sq, 'cos-cos':cp-cq,
        'cs+sc':cs+sc,'cs-sc':cs-sc,'cc+ss':cc+ss,'cc-ss':cc-ss}

def window_destinations(case,src):
    I,J,K=case['size'];L=I*J;k=int(src//L)
    dest=np.arange(max(0,k-2)*L,min(K,k+3)*L,dtype=np.int32)
    return dest[dest!=src]

def choose_top_positive(H,nmax,legal_mask=None):
    # H: [config, destinations]. Seleção ANTES da filtragem de input/output.
    H=np.asarray(H,dtype=np.float32)
    if H.ndim==1:H=H[None,:]
    take=min(nmax,H.shape[1]);ids=np.arange(H.shape[0])[:,None]
    if take==0:return np.empty((H.shape[0],0),dtype=np.int32),np.empty((H.shape[0],0),dtype=bool)
    selected=np.argpartition(H,-take,axis=1)[:,-take:]
    keep=H[ids,selected]>0
    if legal_mask is not None:keep &= legal_mask[selected]
    return selected,keep

def outgoing_legal_mask(case,src,dst):
    return (~case['input'][dst]) & (~(case['input'][src] & case['output'][dst]))

TEST_CASE=get_case(RECORDS[0]);rng=np.random.default_rng(120)
for _ in range(60):
    src=int(rng.integers(len(TEST_CASE['A'])));dst=np.array(rng.integers(len(TEST_CASE['A']),size=8))
    v=features(TEST_CASE,src,dst);assert set(v)==set(OPS)
    X=M[:,None]*TEST_CASE['A'][src];Y=M[:,None]*TEST_CASE['A'][dst]
    w=TEST_CASE['w'][:,None]
    assert np.max(np.abs(v['cs+sc']-np.sum(w*np.sin(X+Y),axis=0)))<2e-6
    assert np.max(np.abs(v['cc+ss']-np.sum(w*np.cos(X-Y),axis=0)))<2e-6
    assert np.max(np.abs(v['cs-sc']-np.sum(w*np.sin(Y-X),axis=0)))<2e-6
    assert np.max(np.abs(v['cc-ss']-np.sum(w*np.cos(X+Y),axis=0)))<2e-6
# Exatamente a forma original quando a=π/16 e η=3π/4
src=21;dst=window_destinations(TEST_CASE,src)[:100]
q=features(TEST_CASE,src,dst)['cos*sin']
new=np.sin(np.float32(np.pi/16)+np.float32(3*np.pi/4)*q)
old=np.sin(np.float32(np.pi/16)+np.sum(h_coefficients(TEST_CASE['genes'])[:,None]*TEST_CASE['cos'][:,src,None]*TEST_CASE['sin'][:,dst],axis=0))
assert np.max(abs(new-old))<3e-6
print('16 operações, identidades e baseline matemático: OK')
```

## 3. Triagem em larga escala — mesmas fontes por genoma

A triagem escolhe fontes espalhadas entre as camadas, calcula os top-$N_{\max}$ sobre **todos os destinos candidatos** e mede o número de destinos distintos atingidos. Também verifica: pesos selecionados médios, conexões válidas e outputs tocados. **Não chame essa métrica de isolamento**, pois apenas uma fração das fontes foi executada nesta etapa.

Para não favorecer só uma família, a lista final mistura vencedores globais e representantes de operações diferentes, mantendo a referência original. Escolhemos parâmetros apenas a partir dos genomas de descoberta.

```python
def sample_sources(case,limit,tag):
    I,J,K=case['size'];L=I*J;rng=np.random.default_rng(np.random.SeedSequence([BASE_SEED,case['seed'],tag]))
    amounts=np.full(K,limit//K,dtype=int);amounts[:limit%K]+=1
    result=[]
    for k,cnt in enumerate(amounts):
        valid=np.arange(k*L,(k+1)*L,dtype=np.int32)
        valid=valid[~case['output'][valid]]
        if len(valid):result.extend(rng.choice(valid,size=min(cnt,len(valid)),replace=False).tolist())
    return np.asarray(result,dtype=np.int32)

OP_GROUPS={op:PARAMS.index[PARAMS.op==op].to_numpy() for op in OPS}

# Batches 56 x num_dest, memory < ~5MB for biggest brains.
def screen_case(case,limit):
    N=len(case['A']);nmax=case['nmax'];sources=sample_sources(case,limit,314)
    indeg=np.zeros((len(PARAMS),N),dtype=np.uint16)
    sumw=np.zeros(len(PARAMS),dtype=np.float64)
    edges=np.zeros(len(PARAMS),dtype=np.int64)
    pos_counts=np.zeros(len(PARAMS),dtype=np.int64)
    dest_counts=0
    for src in sources:
        dst=window_destinations(case,int(src));legal=outgoing_legal_mask(case,int(src),dst)
        values=features(case,int(src),dst)
        dest_counts+=len(dst)
        for op,ii in OP_GROUPS.items():
            p=PARAMS.iloc[ii]
            H=np.sin(p.a.to_numpy(dtype=np.float32)[:,None]+p.eta.to_numpy(dtype=np.float32)[:,None]*values[op][None,:]).astype(np.float32)
            pos_counts[ii]+=np.count_nonzero(H>0,axis=1)
            sel,keep=choose_top_positive(H,nmax,legal)
            kk=np.repeat(ii[:,None],sel.shape[1],axis=1)[keep]
            picked=dst[sel[keep]]
            np.add.at(indeg,(kk,picked),1)
            edges[ii]+=np.bincount(kk,minlength=len(PARAMS))[ii] if len(kk) else 0
            if len(kk):sumw[ii]+=np.bincount(kk,weights=H[np.arange(len(ii))[:,None],sel][keep],minlength=len(PARAMS))[ii]
    touched=np.count_nonzero(indeg[:,~case['input']],axis=1)
    outputs=np.count_nonzero(indeg[:,case['output']],axis=1)
    return pd.DataFrame({'id':PARAMS.id, 'seed':case['seed'],'case':case['case'],'neurons':N,
                         'source_count':len(sources),'touched':touched,
                         'touched_pct':100*touched/(N-N_INPUTS),
                         'outputs_touched':outputs,'selected_edges':edges,
                         'mean_selected_H':np.divide(sumw,edges,out=np.full(len(edges),np.nan),where=edges>0),
                         'positive_candidate_pct':100*pos_counts/(len(PARAMS)*0+dest_counts)})

start=time.perf_counter();screens=[]
for rec in SCREEN_RECORDS:
    c=get_case(rec);t=time.perf_counter();d=screen_case(c,SCREEN_SOURCES_PER_GENOME)
    screens.append(d);print(c['case'],'screen feito,',round(time.perf_counter()-t,1),'s',flush=True)
screen_df=pd.concat(screens,ignore_index=True)
screen_df.to_csv(RESULT_DIR/'screen_per_genome.csv',index=False)
# Comparação relativa ao baseline de cada genoma (amostragem emparelhada)
base_screen=screen_df[screen_df.id==BASE][['seed','touched_pct','selected_edges']].rename(columns={
    'touched_pct':'baseline_touched_pct','selected_edges':'baseline_edges'})
screen_df=screen_df.merge(base_screen,on='seed',validate='many_to_one')
screen_df['coverage_gain_pp']=screen_df.touched_pct-screen_df.baseline_touched_pct
screen_df['edge_retention']=screen_df.selected_edges/screen_df.baseline_edges.replace(0,np.nan)
score=screen_df.groupby('id',as_index=False).agg(
    screen_gain_pp=('coverage_gain_pp','mean'),
    min_gain_pp=('coverage_gain_pp','min'),
    edge_retention=('edge_retention','mean'),
    mean_H=('mean_selected_H','mean'),
    touched_outputs=('outputs_touched','sum'))
score=score.merge(PARAMS,on='id',validate='one_to_one')
# Penalidades contínuas, em vez de remover arbitrariamente candidatos.
score['screen_score']=(score.screen_gain_pp
   - 12*np.maximum(0,.80-score.edge_retention)
   - 8*np.maximum(0,.30-score.mean_H))
score=score.sort_values('screen_score',ascending=False)
score.to_csv(RESULT_DIR/'screen_ranking.csv',index=False)
print('Tempo screening:',round(time.perf_counter()-start,1),'s')
display(score.head(20).round(4))
```

## 4. Seleção de finalistas e diversidade de operações

O conjunto final inclui: baseline; vencedores por score de triagem; melhores alternativas de operações distintas; e (no perfil `full`) pelo menos uma solução encontrada por uma das famílias de diferença/soma de harmônicos. Desempates usam o identificador fixo da fórmula. Não avaliamos o *holdout* até depois de congelar essa lista.

```python
score=score.sort_values(['screen_score','id'],ascending=[False,True]).reset_index(drop=True)
final=[BASE]
# Primeiro top global, depois diversidade por família, depois completar.
for id in score.id.head(8 if PROFILE=='full' else 3):
    if id not in final and len(final)<FINALISTS_MAX:final.append(id)
for op in OPS:
    id=score.loc[score.op==op,'id'].iloc[0]
    if id not in final and len(final)<FINALISTS_MAX:final.append(id)
for id in score.id:
    if len(final)>=FINALISTS_MAX:break
    if id not in final:final.append(id)
FINAL_PARAMS=PARAMS.set_index('id').loc[final].reset_index()
# Controles não pertencentes à varredura H=sin(a+b*sum(F)), logo fora da seleção por score.
CONTROL_NAMES=['radial_baseline','outer_shift_pi4']
ALL_MODES=final+CONTROL_NAMES
FINAL_PARAMS.to_csv(RESULT_DIR/'selected_finalists.csv',index=False)
print('Finalistas da varredura:',len(final),'| controles externos:',CONTROL_NAMES)
display(FINAL_PARAMS[['id','op','a','eta','formula']].round(5))
```

## 5. Avaliação exata em todos os neurônios (sem amostragem da conectividade)

Para cada candidato, executamos todas as origens válidas, a mesma seleção por top-$N_{\max}$ e a mesma filtragem posterior. A radial é incluída apenas como **referência externa**, com $\delta(p)=\frac8{\sqrt3}\sqrt{(i/(I-1))^2+(j/(J-1))^2+(k/(K-1))^2}-4$ aplicada aos argumentos cos/sin como na Parte V. Também testamos o deslocamento externo $a=\pi/4$ com o somatório original. Ambos são rotulados como controles.

```python
def radial_delta(I,J,K):
    k,j,i=np.indices((K,J,I),dtype=np.float32)
    rr=np.sqrt((i/max(I-1,1))**2+(j/max(J-1,1))**2+(k/max(K-1,1))**2)
    return ((8/np.sqrt(3))*rr-4).astype(np.float32).ravel()

def mode_spec(mid):
    if mid=='radial_baseline':return ('radial_baseline',np.pi/16,3*np.pi/4)
    if mid=='outer_shift_pi4':return ('cos*sin',np.pi/4,3*np.pi/4)
    v=FINAL_PARAMS.set_index('id').loc[mid]
    return (v.op,float(v.a),float(v.eta))

# Não recalcular a célula anterior para cada fonte.
FAMILY_BY_OP={op:[] for op in OPS}
for mi,mid in enumerate(ALL_MODES):
    op,a,eta=mode_spec(mid)
    FAMILY_BY_OP.setdefault(op,[]).append((mi,mid,np.float32(a),np.float32(eta)))

def run_full_case(case, modes=ALL_MODES, collect_sample=True):
    N=len(case['A']);I,J,K=case['size'];L=I*J;inp=case['input'];out=case['output']
    nmax=case['nmax'];indeg=np.zeros((len(modes),N),dtype=np.int32)
    outdeg=np.zeros((len(modes),N),dtype=np.int32)
    edges=np.zeros(len(modes),dtype=np.int64);sum_H=np.zeros(len(modes),dtype=np.float64)
    positive=np.zeros(len(modes),dtype=np.int64);evaluated=0
    abs_A_diff=np.zeros(len(modes),dtype=np.float64)
    near_A_edges=np.zeros(len(modes),dtype=np.int64)
    cross_layer_edges=np.zeros(len(modes),dtype=np.int64)
    sample=set(sample_sources(case,RANK_SOURCES,667).tolist()) if collect_sample else set()
    sampled_selected={mid:{} for mid in modes};sampled_weights={mid:[] for mid in modes}
    radial=None
    if 'radial_baseline' in modes:
        delta=radial_delta(I,J,K)
        mc=np.arange(1,5,dtype=np.float32)[:,None]
        rc=np.cos(mc*case['A'][None,:]+delta[None,:]).astype(np.float32)
        rs=np.sin(mc*case['A'][None,:]+delta[None,:]).astype(np.float32)
        radial=(rc,rs)
    params={mid:mode_spec(mid) for mid in modes}
    grouped={}
    for idx,mid in enumerate(modes):
        op,a,eta=params[mid];grouped.setdefault(op,[]).append((idx,mid,a,eta))
    for src in np.flatnonzero(~out):
        dst=window_destinations(case,int(src));legal=outgoing_legal_mask(case,int(src),dst)
        q=features(case,int(src),dst)
        evaluated+=len(dst)
        if radial is not None:
            rc,rs=radial
            q['radial_baseline']=(case['w']*rc[:,src])@rs[:,dst]
        for op,items in grouped.items():
            a=np.asarray([v[2] for v in items],dtype=np.float32)
            eta=np.asarray([v[3] for v in items],dtype=np.float32)
            H=np.sin(a[:,None]+eta[:,None]*q[op][None,:]).astype(np.float32)
            chosen,keep=choose_top_positive(H,nmax,legal)
            positive[[v[0] for v in items]]+=np.count_nonzero(H>0,axis=1)
            for row,(idx,mid,_,_) in enumerate(items):
                picked=dst[chosen[row,keep[row]]]
                indeg[idx,picked]+=1
                outdeg[idx,src]+=len(picked)
                edges[idx]+=len(picked)
                da=np.abs(case['A'][picked]-case['A'][src])
                abs_A_diff[idx]+=da.sum(dtype=np.float64)
                near_A_edges[idx]+=np.count_nonzero(da<.05)
                cross_layer_edges[idx]+=np.count_nonzero(picked//L!=src//L)
                sw=H[row,chosen[row,keep[row]]]
                sum_H[idx]+=sw.sum(dtype=np.float64)
                if src in sample:
                    sampled_selected[mid][int(src)]=set(picked.tolist())
                    sampled_weights[mid].extend(sw.tolist())
    rows=[]
    for idx,mid in enumerate(modes):
        met=metrics(indeg[idx],outdeg[idx],inp,out)
        vals=np.asarray(sampled_weights[mid],dtype=np.float32)
        rows.append({'id':mid,'case':case['case'],'seed':case['seed'],
                     'neurons':N,'held_out':case['seed'] in HOLDOUT_SEEDS,
                     'nmax':nmax,'beta':case['beta'], **met,
                     'mean_selected_H':float(sum_H[idx]/edges[idx]) if edges[idx] else np.nan,
                     'mean_abs_A_difference_on_edges':float(abs_A_diff[idx]/edges[idx]) if edges[idx] else np.nan,
                     'edge_A_difference_below_0_05_pct':float(100*near_A_edges[idx]/edges[idx]) if edges[idx] else np.nan,
                     'cross_layer_edges_pct':float(100*cross_layer_edges[idx]/edges[idx]) if edges[idx] else np.nan,
                     'sampled_weight_p05':float(np.quantile(vals,.05)) if len(vals) else np.nan,
                     'sampled_weight_below_0_1_pct':float(np.mean(vals<.1)*100) if len(vals) else np.nan,
                     'positive_candidate_pct':float(positive[idx]/evaluated*100) if evaluated else np.nan})
    return rows,sampled_selected,indeg

start=time.perf_counter();all_rows=[];edges_by_case={};indeg_by_case={}
for rec in FINAL_RECORDS:
    case=get_case(rec);t=time.perf_counter();rows,edge_sets,indeg=run_full_case(case)
    all_rows+=rows;edges_by_case[case['seed']]=edge_sets;indeg_by_case[case['seed']]=indeg
    base=next(r for r in rows if r['id']==BASE)
    print(f"{case['case']}: referência={base['isolation_pct']:.2f}%, "
          f"finalistas={len(final)}, tempo={time.perf_counter()-t:.1f}s",flush=True)
full_df=pd.DataFrame(all_rows)
base_metrics=full_df[full_df.id==BASE].merge(REFERENCE,on='seed',validate='one_to_one')
print('Erro máximo de isolamento vs Parte I:',float(np.max(abs(base_metrics.isolation_pct-base_metrics.isolation_H_pct))),'p.p.')
assert np.max(abs(base_metrics.isolation_pct-base_metrics.isolation_H_pct))<.30
assert np.all(abs(base_metrics.total_valid_connections-base_metrics.valid_edges_H)<=np.maximum(10,.005*base_metrics.valid_edges_H))
full_df.to_csv(RESULT_DIR/'full_per_genome.csv',index=False)
print('Tempo completo:',round(time.perf_counter()-start,1),'s')
```

## 6. Resultados principais, Pareto e generalização para genomas não vistos

As médias dão peso igual a cada genoma, não a cada neurônio. Apresentamos separadamente descoberta e *holdout*. Um bom candidato precisa simultaneamente reduzir isolamento, preservar pesos, manter a contagem de conexões e não sacrificar saídas. Os finalistas foram escolhidos **sem olhar os genomas reservados**.

```python
agg=full_df.groupby('id',as_index=False).agg(n_genomes=('case','nunique'),
    isolation_pct=('isolation_pct','mean'),gini=('gini_incoming','mean'),
    top10=('top_10pct_incoming_share','mean'),
    mean_H=('mean_selected_H','mean'),weight_p05=('sampled_weight_p05','mean'),
    mean_abs_A_diff=('mean_abs_A_difference_on_edges','mean'),
    near_A_pct=('edge_A_difference_below_0_05_pct','mean'),
    cross_layer_pct=('cross_layer_edges_pct','mean'),
    weak_weight_pct=('sampled_weight_below_0_1_pct','mean'),
    total_edges=('total_valid_connections','sum'),isolated_outputs=('isolated_output','sum'),
    isolated_inputs=('isolated_input','sum'),positive_pct=('positive_candidate_pct','mean'))
agg['edge_retention']=agg.total_edges/agg.loc[agg.id==BASE,'total_edges'].iloc[0]
agg=agg.merge(PARAMS[['id','op','a','eta','formula']],on='id',how='left')
agg.loc[agg.id=='radial_baseline','formula']='Controle radial Parte V'
agg.loc[agg.id=='outer_shift_pi4','formula']='Controle a=π/4, cos*sin'
agg=agg.sort_values(['isolation_pct','id'])
agg.to_csv(RESULT_DIR/'final_summary.csv',index=False)
by_split=full_df.groupby(['id','held_out'],as_index=False).agg(
    isolation_pct=('isolation_pct','mean'),mean_H=('mean_selected_H','mean'),
    outputs=('isolated_output','sum'),edges=('total_valid_connections','sum'))
by_split['split']=np.where(by_split.held_out,'holdout','discovery')
by_split.to_csv(RESULT_DIR/'generalization_discovery_holdout.csv',index=False)
per_size=full_df.groupby(['id','neurons'],as_index=False).agg(
    isolation_pct=('isolation_pct','mean'),outputs=('isolated_output','sum'),mean_H=('mean_selected_H','mean'))
per_size.to_csv(RESULT_DIR/'per_size_results.csv',index=False)
display(agg[['id','op','a','eta','isolation_pct','mean_H','weight_p05',
             'weak_weight_pct','edge_retention','isolated_outputs','gini',
             'mean_abs_A_diff','near_A_pct','cross_layer_pct']].round(4))

fig,axes=plt.subplots(1,2,figsize=(14,5),constrained_layout=True)
axes[0].scatter(agg.isolation_pct,agg.mean_H)
for _,r in agg.head(10).iterrows():axes[0].annotate(str(r['id']), (r.isolation_pct,r.mean_H),fontsize=8)
axes[0].set(xlabel='isolamento médio (%) — menor é melhor',ylabel='H selecionado médio',
            title='Trade-off cobertura × intensidade sináptica')
axes[1].scatter(agg.isolation_pct,agg.gini)
axes[1].set(xlabel='isolamento médio (%)',ylabel='Gini das entradas',
            title='Cobertura × concentração')
fig.savefig(RESULT_DIR/'01_search_tradeoffs.png',dpi=160);plt.show()
```

## 8. Perguntas posteriores já investigadas: pesos quase zero, saturação e rankings compartilhados

* Os vencedores têm muitos pesos quase nulos ($H<0.1$) ou simplesmente alteram o ranking?
* A melhoria de cobertura acontece à custa de saídas isoladas ou perda de arestas?
* A rede fica restrita a pares com $|A_p-A_q|$ muito pequeno? Qual a fração de conexões entre camadas distintas?
* O grau de compartilhamento dos destinos entre diferentes origens diminui?
* A melhora está associada a uma função que depende efetivamente da origem **e** do destino? Operações puramente aditivas podem gerar preferências sistemáticas mesmo com grande dispersão aparente de $H$.

```python
# Comparar rankings de fontes *da mesma camada*, para não confundir com restrições de candidatos.
rank_rows=[]
for rec in FINAL_RECORDS:
    case=get_case(rec);edges=edges_by_case[case['seed']]
    I,J,K=case['size'];L=I*J
    ref_sources=sorted(edges[BASE])
    pairs=[]
    for k in range(K):
        ss=[v for v in ref_sources if v//L==k]
        for idx in range(0,len(ss)-1,2):pairs.append((ss[idx],ss[idx+1]))
    for mid in ALL_MODES:
        jac=[]
        for u,v in pairs:
            a=edges[mid][u];b=edges[mid][v]
            if a|b:jac.append(len(a&b)/len(a|b))
        rank_rows.append({'seed':case['seed'],'id':mid,'pairs':len(jac),
                          'same_layer_jaccard':np.mean(jac) if jac else np.nan})
rank_df=pd.DataFrame(rank_rows)
rank_df.to_csv(RESULT_DIR/'ranking_overlap_by_genome.csv',index=False)
rank_summary=rank_df.groupby('id',as_index=False).same_layer_jaccard.mean().sort_values('same_layer_jaccard')
print('Sobreposição entre rankings de diferentes origens:')
display(rank_summary.round(4))
fig,ax=plt.subplots(figsize=(12,5),constrained_layout=True)
top=agg.head(min(12,len(agg)))
ax.barh(range(len(top)),top.weak_weight_pct)
ax.set(yticks=range(len(top)),yticklabels=top.id,xlabel='% de H selecionados < 0,1 (amostra)',
       title='Risco de conexões quase nulas');ax.invert_yaxis()
fig.savefig(RESULT_DIR/'04_weak_weights.png',dpi=160);plt.show()
```

## 9. De onde vem o isolamento? Candidatos positivos rejeitados ou ausência de oportunidades?

Analisamos os três melhores finalistas, o baseline e a radial. Para economizar memória, contamos somente oportunidades legais **dos destinos que ficaram sem entrada** na execução completa. A classificação é exata para esses destinos nos genomas avaliados e distingue seleção top-$N_{\max}$ de impossibilidade de obter $H>0$. O resgate permanece desativado.

```python
CAUSAL_MODES=list(dict.fromkeys([BASE,'radial_baseline']+[i for i in agg.id if i not in (BASE,'radial_baseline','outer_shift_pi4')][:3]))
causal=[]
for rec in FINAL_RECORDS:
    case=get_case(rec);N=len(case['A']);inp=case['input'];out=case['output']
    indeg=indeg_by_case[case['seed']]
    for mid in CAUSAL_MODES:
        mode_idx=ALL_MODES.index(mid)
        isolated=np.flatnonzero((indeg[mode_idx]==0)&(~inp))
        if not len(isolated):
            causal.append({'seed':case['seed'],'id':mid,'isolated_targets':0,
                           'no_positive_legal':0,'positive_but_never_selected':0});continue
        isolated_flag=np.zeros(N,dtype=bool);isolated_flag[isolated]=True
        no_positive=np.ones(N,dtype=bool)
        op,a,eta=mode_spec(mid)
        c=case
        if mid=='radial_baseline':
            delta=radial_delta(*c['size']);mc=M[:,None]
            ccos=np.cos(mc*c['A'][None,:]+delta[None,:]).astype(np.float32)
            ssin=np.sin(mc*c['A'][None,:]+delta[None,:]).astype(np.float32)
        for src in np.flatnonzero(~out):
            dst=window_destinations(c,int(src));dst=dst[isolated_flag[dst]]
            if not len(dst):continue
            legal=outgoing_legal_mask(c,int(src),dst)
            if not np.any(legal):continue
            dst=dst[legal]
            if mid=='radial_baseline':q=(c['w']*ccos[:,src])@ssin[:,dst]
            else:q=features(c,int(src),dst)[op]
            h=np.sin(np.float32(a)+np.float32(eta)*q)
            no_positive[dst[h>0]]=False
            if not np.any(no_positive[isolated]):break
        nobody=int(np.count_nonzero(no_positive[isolated]));lost=len(isolated)-nobody
        causal.append({'seed':c['seed'],'id':mid,'isolated_targets':len(isolated),
                       'no_positive_legal':nobody,'positive_but_never_selected':lost})
causal_df=pd.DataFrame(causal)
causal_df.to_csv(RESULT_DIR/'isolation_causes.csv',index=False)
display(causal_df.groupby('id',as_index=False)[['isolated_targets','no_positive_legal','positive_but_never_selected']].sum())
```

## 10. Estabilidade genética dos finalistas (teste pareado)

Mutamos um dígito quaternário do gene `connections` por $\pm1$, sem sair do intervalo $\{0,1,2,3\}$. Recalculamos o campo $A$ e os pesos harmônicos, mas mantemos fixos os papéis e $N_{\max}$ do genoma original para medir **apenas** a sensibilidade da afinidade e da seleção. Medimos Jaccard das conexões de fontes previamente selecionadas. Este teste não substitui a avaliação fenotípica de mutações que também alterem papéis neuronais.

```python
MUT_MODES=ALL_MODES  # Todas as fórmulas finalistas, sem estabilidade ausente no relatório
mut_rows=[]
for rec in FINAL_RECORDS:
    c=get_case(rec);srcs=sample_sources(c,RANK_SOURCES,667)[:MUTATION_SOURCES]
    rng=np.random.default_rng(np.random.SeedSequence([BASE_SEED,c['seed'],810]))
    for mutation_idx in range(MUTATIONS_PER_GENOME):
        g={k:v.copy() for k,v in c['genes'].items()}
        pos=int(rng.integers(64));old_digit=int(g['connections'][pos]);
        opts=[v for v in [old_digit-1,old_digit+1] if 0<=v<=3]
        g['connections'][pos]=rng.choice(opts)
        alt=c.copy();alt['A']=make_A(g,*c['size']);alt['genes']=g
        alt['w']=np.asarray([np.sum(W64[g['connections']==d],dtype=np.float32) for d in range(4)],dtype=np.float32)/np.sum(W64)
        alt['cos']=np.cos(M[:,None]*alt['A'][None,:]).astype(np.float32)
        alt['sin']=np.sin(M[:,None]*alt['A'][None,:]).astype(np.float32)
        delta=radial_delta(*c['size']);mc=M[:,None]
        rc=np.cos(mc*alt['A'][None,:]+delta[None,:]);rs=np.sin(mc*alt['A'][None,:]+delta[None,:])
        for mid in MUT_MODES:
            op,a,eta=mode_spec(mid);over=[];weight_change=[]
            old_edges=edges_by_case[c['seed']][mid]
            for src in srcs:
                dst=window_destinations(alt,int(src));legal=outgoing_legal_mask(c,int(src),dst)
                if mid=='radial_baseline':q=(alt['w']*rc[:,src])@rs[:,dst]
                else:q=features(alt,int(src),dst)[op]
                H=np.sin(np.float32(a)+np.float32(eta)*q).astype(np.float32)
                ii,keep=choose_top_positive(H,c['nmax'],legal)
                chosen=set(dst[ii[0,keep[0]]].tolist());ref=old_edges[int(src)]
                over.append(len(chosen&ref)/len(chosen|ref) if chosen|ref else 1.)
            mut_rows.append({'seed':c['seed'],'id':mid,'mutation':mutation_idx,
                'gene_position':pos,'jaccard':float(np.mean(over))})
mut_df=pd.DataFrame(mut_rows)
mut_df.to_csv(RESULT_DIR/'mutation_stability.csv',index=False)
mut_summary=mut_df.groupby('id',as_index=False).agg(jaccard=('jaccard','mean'),std=('jaccard','std'),n=('jaccard','count'))
mut_summary.to_csv(RESULT_DIR/'mutation_summary.csv',index=False)
display(mut_summary.round(4))
```

## 11. Síntese automática e critérios para recomendar uma fórmula

O menor isolamento **não basta**. Relatamos as melhores fórmulas em categorias distintas e apontamos as que mantêm pelo menos 95% das arestas do baseline, média de $H$ selecionado $\ge0.5$ e no máximo tantas saídas isoladas quanto o baseline. Esses limites são **critérios de triagem, não leis fisiológicas**; a tabela completa permanece disponível para avaliar outros limites. A recomendação exige bons resultados no *holdout* e ausência de degradação genética evidente.

```python
agg=agg.merge(mut_summary[['id','jaccard']],on='id',how='left',validate='one_to_one')
agg.rename(columns={'jaccard':'mutation_jaccard'},inplace=True)
agg.to_csv(RESULT_DIR/'final_summary.csv',index=False)
baseline=agg[agg.id==BASE].iloc[0]
feasible=agg[(agg.edge_retention>=.95)&(agg.mean_H>=.50)
             &(agg.isolated_outputs<=baseline.isolated_outputs)]
valid_candidates=feasible[feasible.id.isin(final)]
print('Baseline:',BASE,'isolamento',round(baseline.isolation_pct,3),'%')
print('Melhor isolamento bruto:')
display(agg[['id','op','a','eta','isolation_pct','mean_H','edge_retention','isolated_outputs']].head(8).round(4))
print('Candidatos que passam os filtros de pesos, arestas e outputs:',len(valid_candidates))
display(valid_candidates[['id','op','a','eta','isolation_pct','mean_H','edge_retention','isolated_outputs',
                              'mutation_jaccard','near_A_pct','cross_layer_pct']].head(10).round(4))
print('Estabilidade genética do baseline:',round(baseline.mutation_jaccard,4))
print('ATENÇÃO: a fórmula com menor isolamento pode ter estabilidade/estrutura muito piores.')
if len(valid_candidates):
    best=valid_candidates.iloc[0]
    print('MELHOR ENTRE CANDIDATOS ADMISSÍVEIS:',best.id,best.formula,
          'a=',round(best.a,5),'b=',f'{best.eta:.5f}/C',
          'isolamento=',round(best.isolation_pct,3),'%')
    sub=by_split[(by_split.id==best.id)|(by_split.id==BASE)]
    print('GENERALIZAÇÃO (separação de descoberta / holdout):')
    display(sub.pivot(index='split',columns='id',values='isolation_pct').round(3))
else:print('Nenhum candidato passou todos os filtros: não há vencedor defensável.')

questions=[
 {'question':'Trocar produto por soma ajuda?','diagnostic':'ranking por operação, resultados completos, holdout',
  'interpretation':'Uma expressão só deve ser considerada superior se melhorar cobertura sem perder pesos/conexões.'},
 {'question':'a=0 e b=1/C são competitivos?','diagnostic':'screen_ranking + parâmetros dos finalistas',
  'interpretation':'A presença na grade não garante promoção à simulação completa; examine a triagem.'},
 {'question':'O ganho se deve apenas a H fraco?','diagnostic':'peso médio, p05 e fração abaixo de 0,1',
  'interpretation':'Descartar vitórias que transformem a maioria das conexões em pesos minúsculos.'},
 {'question':'Os rankings continuam redundantes?','diagnostic':'ranking_overlap_by_genome.csv',
  'interpretation':'Menor Jaccard entre origens não garante maior cobertura; verificar ambos.'},
 {'question':'Os isolados são impossíveis ou preteridos?','diagnostic':'isolation_causes.csv',
  'interpretation':'A oportunidade H>0 pode existir sem o destino entrar em top Nmax.'},
 {'question':'A melhoria generaliza?','diagnostic':'generalization_discovery_holdout.csv e per_size_results.csv',
  'interpretation':'Vantagem restrita aos genomas de descoberta não justifica adotar fórmula.'},
 {'question':'Conexões se concentram em valores de A quase idênticos?', 'diagnostic':'final_summary.csv near_A_pct / cross_layer_pct',
  'interpretation':'Zero isolamento não é suficiente se a topologia perder diversidade e conectividade entre camadas.'},
 {'question':'Conexões sobrevivem a mutação pequena?','diagnostic':'mutation_stability.csv',
  'interpretation':'Analisar Jaccard com a distribuição por genoma, sem usar somente a média.'},
]
pd.DataFrame(questions).to_csv(RESULT_DIR/'questions_and_diagnostics.csv',index=False)
```

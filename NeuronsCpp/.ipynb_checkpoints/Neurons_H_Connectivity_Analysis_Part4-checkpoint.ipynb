# Neurons — Parte IV: escala angular $\lambda$ na função $H$

**Pergunta experimental:** multiplicar os argumentos trigonométricos por um único fator $\lambda$ reduz a concentração das conexões sem alterar $A$, o genoma, as regras de construção ou a seleção determinística dos maiores valores de $H$?

$$H_\lambda(a,b)=\sin\left(\frac{\pi}{16}+\sum_{m=1}^4 c_m\cos(m\lambda a)\sin(m\lambda b)\right).$$

**Controle:** $\lambda=1$ reproduz a função atual. Variamos **somente** $\lambda$; os valores $A_{i,j,k}$, os papéis de entrada/saída, os coeficientes genéticos $c_m$, $N_{\max}$ e a lista de candidatos permanecem fixos. Todos os destinos positivos são avaliados, então escolhemos os $N_{\max}$ maiores e removemos conexões proibidas **antes do resgate**. O notebook não modifica o projeto Godot.

**Interpretação:** a métrica prioritária é a fração de neurônios sem as conexões exigidas pelo seu papel. Para evitar a falsa solução de uma rede quase aleatória, também examinamos Gini, participação dos 10% mais conectados, número de conexões, afinidade média e estabilidade entre genomas. Um valor particular de $\lambda$ não é garantido como solução — a relação pode ser oscilatória e não monotônica.


## 0. Configuração e genomas da Parte I

```python
import json, time, io, zipfile
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

PROFILE = 'quick'  # quick: 4 genomas de 1.000 e 4.840 neurônios; full: todos os 11 genomas
# Recomenda-se quick para escolher regiões promissoras, full para verificar generalização.
LAMBDA_COARSE = [0.5, 0.75, 1.0, 1.25, 1.5, 2.0, 2.5, 3.0, 4.0, 5.0, 6.0, 8.0]
RUN_REFINEMENT = True
REFINE_COUNT = 7             # número de lambdas locais, incluindo o centro
REFINE_HALF_WIDTH = 0.6      # largura de cada lado do centro
REFINE_TOP = 2               # seleciona os 2 melhores lambdas não triviais da varredura inicial
N_INPUTS = 10
N_OUTPUTS = 10
OUTPUT_DIR = Path('neurons_h_part4_results'); OUTPUT_DIR.mkdir(exist_ok=True)
ZIP_RESULTS = Path('Neurons_H_Part4_Results.zip')
PRIOR_ZIP = Path('Neurons_H_Analysis_Results.zip')
assert PRIOR_ZIP.exists(), 'Coloque Neurons_H_Analysis_Results.zip na mesma pasta do notebook.'
with zipfile.ZipFile(PRIOR_ZIP) as z:
    records = json.loads(z.read('genomes.json'))
if PROFILE == 'quick':
    selected = [r for r in records if r['size'][0] == 10][:2] + [r for r in records if r['size'][0] == 22][:2]
else:
    selected = records
assert selected, 'Nenhum genoma foi encontrado.'
print('Genomas:',len(selected),'| Dimensões:',[r['size'] for r in selected]);print('Lambdas:',LAMBDA_COARSE)
```

## 1. Reprodução das funções matemáticas do Neurons

```python
PI32 = np.float32(np.pi)
W8 = np.sqrt(np.arange(1, 9, dtype=np.float32))
W64 = np.sqrt(np.arange(1, 65, dtype=np.float32))

def random_genome(seed):
    rng = np.random.default_rng(seed)
    # O construtor C++ gera dígitos independentes e uniformes de 0 a 3.
    return {"connections": rng.integers(0,4,64,dtype=np.int8),
            "beta": rng.integers(0,4,8,dtype=np.int8),
            "max_connections": rng.integers(0,4,8,dtype=np.int8)}

def parse_genome(dictionary):
    lengths = {"connections": 64, "beta": 8, "max_connections": 8}
    ans = {}
    for name, count in lengths.items():
        value = dictionary[name]
        if isinstance(value, str):
            if len(value) != count or not set(value) <= set("0123"):
                raise ValueError(f"{name}: esperado {count} dígitos quaternários")
            ans[name] = np.array([int(ch) for ch in value], dtype=np.int8)
        else:
            ans[name] = np.asarray(value, dtype=np.int8)
            if ans[name].shape != (count,) or not np.isin(ans[name], [0,1,2,3]).all():
                raise ValueError(f"{name}: formato inválido")
    return ans

def genome_text(g):
    return {key: "".join(map(str, val)) for key, val in g.items()}

def beta_max(g, I, J):
    beta = np.float32(1.0 + np.sum(W8*g["beta"],dtype=np.float32)
                     / np.sum(3*W8,dtype=np.float32) * np.float32((I*J/16.)**(1/3)))
    maxconn = int(np.floor(np.sum(W8*np.power(beta, g["max_connections"].astype(np.float32)),dtype=np.float32)))
    return float(beta), maxconn

def h_coefficients(g):
    digits = g["connections"]
    return np.array([np.sum(W64[digits == d],dtype=np.float32) for d in range(4)],dtype=np.float32) * (np.float32(3)*PI32 / (np.float32(4)*np.sum(W64,dtype=np.float32)))

def make_A(g, I, J, K):
    # A[k,j,i], linearizado como i + I*j + I*J*k.
    coeff = (np.float32(2)/np.float32(3))*g["connections"].astype(np.float32)-np.float32(1)
    cx = np.cos(PI32*np.arange(4,dtype=np.float32)[:,None]*
                (np.arange(I,dtype=np.float32)[None,:]/np.float32(I-1))).astype(np.float32)
    cy = np.cos(PI32*np.arange(4,dtype=np.float32)[:,None]*
                (np.arange(J,dtype=np.float32)[None,:]/np.float32(J-1))).astype(np.float32)
    cz = np.cos(PI32*np.arange(4,dtype=np.float32)[:,None]*
                (np.arange(K,dtype=np.float32)[None,:]/np.float32(K-1))).astype(np.float32)
    field = np.zeros((K,J,I), dtype=np.float32)
    for r in range(4):
        for s in range(4):
            for q in range(4):
                field += coeff[q + 4*s + 16*r] * cx[r][None,None,:] * cy[s][None,:,None] * cz[q][:,None,None]
    return np.tanh(field / np.float32(8)).astype(np.float32).ravel()

def h_direct(a, b, c):
    z = np.float32(np.pi/16)
    for m in range(1, 5):
        z = z + c[m-1]*np.cos(np.float32(m)*a)*np.sin(np.float32(m)*b)
    return np.sin(z)

def make_fast_h(A, coeffs):
    harmonics = np.arange(1,5,dtype=np.float32)[:,None]
    ac = (coeffs[:,None]*np.cos(harmonics*A[None,:])).astype(np.float32)
    bs = np.sin(harmonics*A[None,:]).astype(np.float32)
    return ac, bs

def h_from_cached(ac, bs, src, target_indices):
    z = np.full(len(target_indices), np.float32(np.pi/16), dtype=np.float32)
    for m in range(4):
        z += ac[m, src] * bs[m, target_indices]
    return np.sin(z).astype(np.float32)


def make_fast_h_lambda(A, coeffs, lam):
    # Mesma organização de float32 da versão anterior; lambda=1 replica make_fast_h.
    harmonics = (np.arange(1,5,dtype=np.float32)*np.float32(lam))[:,None]
    ac = (coeffs[:,None]*np.cos(harmonics*A[None,:])).astype(np.float32)
    bs = np.sin(harmonics*A[None,:]).astype(np.float32)
    return ac,bs

# Testes diretos da equivalência em lambda = 1
_g = random_genome(2026); _a = make_A(_g,5,5,4); _c = h_coefficients(_g)
_ac,_bs = make_fast_h(_a,_c); _lc,_ls = make_fast_h_lambda(_a,_c,1.0)
assert np.array_equal(_ac,_lc) and np.array_equal(_bs,_ls)
ix=np.arange(0,len(_a),3)
assert np.allclose(h_from_cached(_lc,_ls,0,ix),h_direct(_a[0],_a[ix],_c),atol=1e-6)
print('Validação de lambda=1: OK')
```

## 2. Seleção de conexões e métricas

Para cada origem, avaliamos todos os candidatos nas camadas $|\Delta k|\le 2$, excluímos a autoconexão, selecionamos os maiores $H_\lambda>0$ até $N_{\max}$ e só depois removemos as conexões proibidas. **Não** exigimos cobertura adicional, nem realizamos resgate; assim a métrica mede a necessidade natural de resgate. Os mesmos papéis neuronais são mantidos para todos os $\lambda$.

O índice *top-10%* é a fração das entradas recebidas pelos 10% dos destinos válidos com maior grau de entrada. O Gini refere-se a esses mesmos destinos. 

```python
def flags_by_rank(A, I, J, K, n_inputs, n_outputs):
    N = len(A)
    if n_inputs+n_outputs > N:
        raise ValueError("Número de inputs+outputs maior que número de neurônios")
    by_layer = []
    for k in range(K):
        ix = np.arange(k*I*J,(k+1)*I*J,dtype=np.int32)
        by_layer.append(ix[np.argsort(A[ix],kind="quicksort")])
    # rank 0 de cada camada; depois rank 1 de cada camada; ...
    permutation = np.stack(by_layer,axis=1).ravel()
    is_input = np.zeros(N,dtype=bool); is_output = np.zeros(N,dtype=bool)
    is_input[permutation[:n_inputs]] = True
    is_output[permutation[-n_outputs:]] = True
    return is_input, is_output

def gini_nonnegative(data):
    x = np.sort(np.asarray(data,dtype=np.float64)); n=len(x)
    if n == 0 or np.sum(x) == 0: return 0.
    return float((2*np.dot(np.arange(1,n+1),x)/(n*np.sum(x))) - (n+1)/n)

def metrics(incoming, outgoing, is_input, is_output):
    internal = ~(is_input | is_output)
    counts = {"internal":int(np.count_nonzero((incoming==0)&internal)),
              "output":int(np.count_nonzero((incoming==0)&is_output)),
              "input":int(np.count_nonzero((outgoing==0)&is_input))}
    N=len(incoming)
    eligible = ~is_input
    vals = incoming[eligible]
    tenpct = max(1,int(np.ceil(.1*len(vals))))
    total = np.sum(vals)
    top_share = float(np.partition(vals,-tenpct)[-tenpct:].sum()/total) if total else np.nan
    return {"isolated_total":sum(counts.values()),"isolation_pct":100.*sum(counts.values())/N,
            "isolated_internal":counts["internal"],"isolated_output":counts["output"],"isolated_input":counts["input"],
            "mean_incoming":float(np.mean(vals)),"median_incoming":float(np.median(vals)),
            "gini_incoming":gini_nonnegative(vals),"top_10pct_incoming_share":top_share,
            "zero_incoming_eligible_pct":100.*np.mean(vals==0),"total_valid_connections":int(np.sum(vals))}


def simulate_lambda(A,g,size,lam,roles=None):
    I,J,K=map(int,size);N=I*J*K
    is_input,is_output = flags_by_rank(A,I,J,K,N_INPUTS,N_OUTPUTS) if roles is None else roles
    _,nmax=beta_max(g,I,J)
    ac,bs=make_fast_h_lambda(A,h_coefficients(g),lam)
    incoming=np.zeros(N,np.int32); outgoing=np.zeros(N,np.int32)
    windows=[]
    for k in range(K):
        windows.append(np.arange(max(0,k-2)*I*J,min(K,k+3)*I*J,dtype=np.int32))
    total_positive=total_candidates=total_preclean=total_removed=0
    sum_h=0.0;num_h=0
    for src in np.flatnonzero(~is_output):
        cand=windows[int(src//(I*J))]
        dest=cand[cand!=src]
        h=h_from_cached(ac,bs,int(src),dest)
        eligible=np.flatnonzero(h>0)
        total_positive+=len(eligible);total_candidates+=len(dest)
        take=min(nmax,len(eligible))
        if take==0:continue
        if take<len(eligible):
            picked=eligible[np.argpartition(h[eligible],-take)[-take:]]
        else:picked=eligible
        target=dest[picked]
        valid=(~is_input[target]) & (~(is_input[src] & is_output[target]))
        kept=target[valid]
        incoming[kept]+=1;outgoing[src]+=len(kept)
        total_preclean+=len(picked);total_removed+=len(picked)-len(kept)
        sum_h+=float(h[picked][valid].sum(dtype=np.float64));num_h+=len(kept)
    result=metrics(incoming,outgoing,is_input,is_output)
    result.update(Nmax=int(nmax),positive_pct=100*total_positive/total_candidates,
       removed_pct=100*total_removed/total_preclean if total_preclean else 0.,
       mean_H=sum_h/num_h if num_h else np.nan,
       incoming=incoming)
    return result

smg=random_genome(99);sma=make_A(smg,5,5,4)
sm=simulate_lambda(sma,smg,(5,5,4),1.0)
assert 0<=sm['isolation_pct']<=100
print('Simulação de controle: OK | isolamento:',round(sm['isolation_pct'],2),'%')
```

## 3. Varredura ampla de $\lambda$: genomas idênticos para todas as condições

```python
cases=[]
for idx,r in enumerate(selected):
    g=parse_genome(r['genes']);size=tuple(r['size']);A=make_A(g,*size)
    roles=flags_by_rank(A,*size,N_INPUTS,N_OUTPUTS)
    cases.append(dict(case=f"{size[0]}x{size[1]}x{size[2]}_g{idx+1:02d}",
                      genome_seed=r['seed'],size=size,A=A,g=g,roles=roles))
rows=[]; degrees={};start=time.perf_counter()
def execute(lambdas,phase):
    for case in cases:
        for lam in lambdas:
            lam=float(lam)
            # reutilização segura: testes com lambdas coincidentes não serão repetidos
            if any(row['case']==case['case'] and abs(row['lambda']-lam)<1e-9 for row in rows): continue
            tic=time.perf_counter();m=simulate_lambda(case['A'],case['g'],case['size'],lam,case['roles'])
            row=dict(case=case['case'],genome_seed=case['genome_seed'],size=case['size'][0],
                     neurons=len(case['A']),**{'lambda':lam},phase=phase,
                     runtime_s=time.perf_counter()-tic)
            row.update({k:v for k,v in m.items() if k!='incoming'})
            rows.append(row)
            if lam in (1.0,2.0,4.0):degrees[(case['case'],lam)]=m['incoming']
            print(f"{case['case']:16} lambda={lam:5.2f} isolados={m['isolation_pct']:6.2f}% "
                  f"Gini={m['gini_incoming']:.3f} H={m['mean_H']:.4f} "
                  f"{row['runtime_s']:.1f}s",flush=True)
execute(LAMBDA_COARSE,'coarse')
raw=pd.DataFrame(rows)
display(raw.head().round(3));print('Tempo inicial:',round(time.perf_counter()-start,1),'s')
```

## 4. Refinamento local, sem escolher um $\lambda$ a partir de um único cérebro

Escolhemos até `REFINE_TOP` valores entre os **melhores da média dos genomas** quanto ao isolamento. A avaliação precisa de genomas independentes para confirmar generalização: os dados de refinamento usam os mesmos genomas e, portanto, **não são um conjunto independente de validação**. A seleção de valores baseia-se apenas no isolamento; Gini e afinidade são analisados separadamente como compromissos.

```python
coarse=pd.DataFrame(rows)
leaders=(coarse[coarse['lambda']!=1.0].groupby('lambda')['isolation_pct']
         .mean().sort_values().head(REFINE_TOP))
print('Melhores lambdas na grade inicial:',leaders.round(3).to_dict())
if RUN_REFINEMENT:
    finer=set()
    for center in leaders.index:
        finer.update(np.linspace(max(.05,center-REFINE_HALF_WIDTH),center+REFINE_HALF_WIDTH,REFINE_COUNT).round(6))
    finer=sorted(x for x in finer if all(abs(x-y)>1e-9 for y in LAMBDA_COARSE))
    print('Pontos de refinamento:',finer)
    execute(finer,'refined')
df=pd.DataFrame(rows).sort_values(['neurons','case','lambda']).reset_index(drop=True)
df.to_csv(OUTPUT_DIR/'per_genome_lambda.csv',index=False)
original=df[np.isclose(df['lambda'],1.0)].drop(columns=['lambda','phase','runtime_s'])
paired=df.merge(original,on=['case','genome_seed','size','neurons'],suffixes=('','_original'))
for name in ['isolation_pct','gini_incoming','top_10pct_incoming_share',
             'mean_H','total_valid_connections','positive_pct']:
    paired['delta_'+name]=paired[name]-paired[name+'_original']
paired.to_csv(OUTPUT_DIR/'paired_vs_lambda1.csv',index=False)
summary=df.groupby(['lambda','neurons'],as_index=False).agg(
    n_genomes=('case','nunique'),isolation=('isolation_pct','mean'),
    isolation_sd=('isolation_pct','std'),gini=('gini_incoming','mean'),
    top10=('top_10pct_incoming_share','mean'),edges=('total_valid_connections','mean'),
    mean_H=('mean_H','mean'),positive=('positive_pct','mean'))
summary.to_csv(OUTPUT_DIR/'summary_by_size_lambda.csv',index=False)
display(summary.round(3))
```

## 5. Curvas de isolamento e concentração em função de $\lambda$

```python
fig,axs=plt.subplots(1,3,figsize=(17,4.8),constrained_layout=True)
for N,s in summary.groupby('neurons'):
    s=s.sort_values('lambda')
    for ax,col in zip(axs,['isolation','gini','top10']):
        ax.plot(s['lambda'],s[col],marker='.',label=f'{N:,} neurônios')
for ax,lab in zip(axs,['Isolados antes do resgate (%)','Gini das entradas','Conexões dos 10% mais conectados']):
    ax.set_xlabel(r'$\lambda$');ax.set_ylabel(lab);ax.grid(alpha=.25)
    ax.axvline(1,color='black',linestyle='--',alpha=.45)
axs[0].legend();fig.savefig(OUTPUT_DIR/'01_isolation_concentration_vs_lambda.png',dpi=160);plt.show()
```

## 6. Troca entre cobertura, número de conexões e afinidade

```python
fig,axs=plt.subplots(1,3,figsize=(17,4.8),constrained_layout=True)
for N,s in summary.groupby('neurons'):
    s=s.sort_values('lambda')
    for ax,col in zip(axs,['mean_H','edges','positive']):
        ax.plot(s['lambda'],s[col],marker='.',label=f'{N:,} neurônios')
for ax,lab in zip(axs,['H médio das conexões válidas','Total de conexões válidas','Candidatos com H>0 (%)']):
    ax.set_xlabel(r'$\lambda$');ax.set_ylabel(lab);ax.axvline(1,color='black',ls='--',alpha=.45);ax.grid(alpha=.25)
axs[0].legend();fig.savefig(OUTPUT_DIR/'02_affinity_edges_positivity.png',dpi=160);plt.show()
```

## 7. Mudanças pareadas: melhora real ou apenas média enganosa?

```python
fig,axs=plt.subplots(1,2,figsize=(14,5),constrained_layout=True)
for case,sub in paired.groupby('case'):
    sub=sub.sort_values('lambda')
    axs[0].plot(sub['lambda'],sub['delta_isolation_pct'],alpha=.75,label=case)
    axs[1].plot(sub['lambda'],sub['delta_gini_incoming'],alpha=.75,label=case)
for ax,lab in zip(axs,['Mudança no isolamento (pontos percentuais)','Mudança no Gini']):
    ax.axhline(0,color='black',ls='--',alpha=.5);ax.axvline(1,color='black',ls=':',alpha=.4)
    ax.set_xlabel(r'$\lambda$');ax.set_ylabel(lab);ax.grid(alpha=.2)
axs[0].legend(fontsize=7,ncol=2)
fig.savefig(OUTPUT_DIR/'03_paired_effects.png',dpi=160);plt.show()
# Tabela de candidatos promissores com variação entre genomas
ranking=(paired.groupby('lambda').agg(average_delta_isolation=('delta_isolation_pct','mean'),
     worst_delta_isolation=('delta_isolation_pct','max'),
     improved_genomes=('delta_isolation_pct',lambda x:int((x<0).sum())),
     count=('case','nunique'),average_delta_gini=('delta_gini_incoming','mean'),
     average_delta_H=('delta_mean_H','mean')).reset_index()
     .sort_values('average_delta_isolation'))
ranking.to_csv(OUTPUT_DIR/'lambda_ranking.csv',index=False)
display(ranking.head(20).round(4))
```

## 8. Visualização da geometria de $H_\lambda$

À medida que $\lambda$ aumenta, crescem também as oscilações no domínio $a,b\in[-1,1]$. O gráfico ajuda a verificar se um eventual ganho corresponde a maior variedade de preferências — sem garantir, por si só, melhor cobertura.

```python
example=cases[0]; coeff=h_coefficients(example['g'])
axis=np.linspace(-1,1,180,dtype=np.float32);aa,bb=np.meshgrid(axis,axis)
choices=[1.0,2.0,4.0,6.0]
fig,axs=plt.subplots(1,4,figsize=(16,3.9),constrained_layout=True)
for ax,lam in zip(axs,choices):
    z=np.full_like(aa,np.float32(np.pi/16))
    for m in range(1,5):
        z+=coeff[m-1]*np.cos(np.float32(m*lam)*aa)*np.sin(np.float32(m*lam)*bb)
    h=np.sin(z)
    im=ax.imshow(h,origin='lower',extent=[-1,1,-1,1],cmap='coolwarm',vmin=-1,vmax=1,aspect='equal')
    ax.set_title(f'lambda = {lam:g}');ax.set_xlabel('A do destino');ax.set_ylabel('A da origem')
fig.colorbar(im,ax=axs,shrink=.85,label='H')
fig.savefig(OUTPUT_DIR/'04_H_surfaces.png',dpi=150);plt.show()
```

## 9. Leituras e limitações

- **Caso $\lambda=1$:** referência obrigatória, inclusive para verificar a reprodução dos resultados anteriores (diferenças de empates ou arredondamento podem ocorrer).
- **Não monotonicidade:** $\lambda$ maiores não são automaticamente melhores; o seno exterior também é periódico.
- **Separação de causas:** alterar $\lambda$ altera simultaneamente os perfis de afinidade, a taxa $H>0$ e a ordenação dos candidatos. As métricas adicionais ajudam a identificar qual efeito acompanha uma melhora.
- **Resgate:** não é simulado. A taxa de isolamento mede quantos neurônios necessitariam de resgate.
- **Generalização:** os melhores valores da varredura são exploratórios, sobretudo com poucos genomas; seria necessário validar em genomas adicionais antes de alterar a modelagem do Neurons.

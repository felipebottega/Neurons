# Neurons — Parte X: a simetria da origem em $H$ e a fase $\phi$

**Pergunta central.** A simetria $H(a,b)=H(-a,b)$ e a pequena derivada do cosseno perto de $a=0$ são responsáveis por parte da concentração dos rankings e do isolamento pré-resgate?

Para cada genoma (64 dígitos quaternários no gene `connections`), mantemos exatamente o campo $A$, as mesmas posições e papéis neuronais, $\beta$, $N_{\max}$, os candidatos com $|k_p-k_q|\le 2$, o limiar $H>0$, a seleção **dos maiores valores** e a filtragem posterior das conexões proibidas, tal como nas Partes anteriores.

$$H_{\phi}(p,q)=\sin\left(\frac\pi{16}+\sum_{m=1}^4c_m\cos(mA_p+\phi)\sin(mA_q)\right), \qquad c_m=\frac{3\pi}{4C}\sum_{t:U(x_t)=m}\sqrt{1+t}.$$

**Não adicionamos genes, perturbações aleatórias, novos somatórios nem seleção por pesos mínimos.** A fase $\phi$ é uma constante global, não um parâmetro genético.

### Perguntas investigadas neste mesmo notebook

1. Remover a paridade da origem realmente melhora a cobertura em 11 genomas e quatro tamanhos?
2. O sinal de $\phi$ altera os resultados? Uma melhora isolada pode refletir o viés de sinais de $A$?
3. Os rankings de origens com $|A|$ semelhante, mas sinais opostos, deixam de coincidir?
4. Que fração dos isolados tinha oportunidades positivas e perdeu na competição do top-$N_{\max}$?
5. A cobertura melhora às custas de conexões positivas muito fracas, de perda de conexões ou de outputs isolados?
6. A fase funciona sozinha ou interage com a perturbação radial e com o *shift* externo (controles separados)?
7. Pequenas mutações produzem rankings instáveis? Diferenciação genética é preservada?
8. Há relação entre derivada local de $H$, diversidade dos rankings e ganhos de cobertura?

**Limitações:** o ranking discreto não é uma função contínua do genoma nas fronteiras de troca de destino. A análise de mutações amostra 1–2 mutações por genoma, mantendo fixos os papéis input/output originais para isolar o efeito sobre $H$. Os controles radial e de *shift* externo são **diagnósticos** e não devem ser interpretados como uma única varredura de $\phi$.

```python
import io, json, zipfile, time, math
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

PROFILE='quick'              # 'quick': 4 genomas; 'full': todos os 11
BASE_SEED=20261025
N_INPUTS=N_OUTPUTS=10
SAMPLED_SOURCES=64          # fontes para comparação pareada dos rankings
RANK_PAIR_SOURCES=36       # pares de fontes por camada / genoma
STABILITY_MUTATIONS=2      # mutações de 1 dígito por genoma (amostra)
STABILITY_ORIGINS=20
OUTPUT_DIR=Path('neurons_h_part10_results'); OUTPUT_DIR.mkdir(exist_ok=True)
RESULT_ZIP=Path('Neurons_H_Part10_Results.zip')
INPUT_ZIP=Path('Neurons_H_Analysis_Results.zip')
assert INPUT_ZIP.exists(), 'Coloque Neurons_H_Analysis_Results.zip ao lado do notebook.'
with zipfile.ZipFile(INPUT_ZIP) as zf:
    records=json.loads(zf.read('genomes.json'))
    baseline_reference=pd.read_csv(io.BytesIO(zf.read('per_brain_metrics.csv')))
assert len(records)==11, f'Esperados 11 genomas; recebidos {len(records)}'
if PROFILE=='quick':
    selected=[r for r in records if r['size'][0]==10][:2]+[r for r in records if r['size'][0]==22][:2]
elif PROFILE=='full':selected=records
else:raise ValueError('PROFILE deve ser quick ou full')
SHIFT0=float(np.pi/16)
PHASES={'phi_0':0.,'phi_p16':float(np.pi/16),'phi_p8':float(np.pi/8),
        'phi_p4':float(np.pi/4),'phi_3p8':float(3*np.pi/8),
        'phi_p2':float(np.pi/2),'phi_neg_p4':float(-np.pi/4)}
CONTROL_MODES=['target_phi_p4','radial','radial_phi_p4','outer_shift_p4','phi_p4_outer_shift_p4']
MODES=list(PHASES)+CONTROL_MODES
BASELINE='phi_0'
MODE_LABELS={'phi_0':'Original (φ=0)','phi_p16':'φ=π/16','phi_p8':'φ=π/8',
    'phi_p4':'φ=π/4','phi_3p8':'φ=3π/8','phi_p2':'φ=π/2','phi_neg_p4':'φ=−π/4',
    'target_phi_p4':'fase π/4 no destino','radial':'radial da Parte V',
    'radial_phi_p4':'radial + φ=π/4','outer_shift_p4':'shift externo π/4',
    'phi_p4_outer_shift_p4':'φ=π/4 + shift π/4'}
print('Perfil:',PROFILE,'| genomas:',len(selected),'| condições:',len(MODES))
print('Dimensões:',[r['size'] for r in selected])
```

## 1. Implementação fiel de $A$, dos genes e dos papéis neuronais

A expansão tridimensional de $A$ e as fórmulas de $\beta$ e $N_{\max}$ são as mesmas da Parte IX. Todos os coeficientes são calculados em `float32`, como referência para a implementação nativa. Empates numéricos na fronteira top-$N_{\max}$ ainda podem variar em relação ao C++.

```python
PI32 = np.float32(np.pi)
W8 = np.sqrt(np.arange(1, 9, dtype=np.float32))
W64 = np.sqrt(np.arange(1, 65, dtype=np.float32))

def random_genome(seed):
    rng = np.random.default_rng(seed)
    # O construtor C++ gera dígitos independentes e uniformes de 0 a 3.
    return {"connections": rng.integers(0,4,64,dtype=np.int8),
            "beta": rng.integers(0,4,8,dtype=np.int8),
            "max_connections": rng.integers(0,4,8,dtype=np.int8)}

def parse_genome(dictionary):
    lengths = {"connections": 64, "beta": 8, "max_connections": 8}
    ans = {}
    for name, count in lengths.items():
        value = dictionary[name]
        if isinstance(value, str):
            if len(value) != count or not set(value) <= set("0123"):
                raise ValueError(f"{name}: esperado {count} dígitos quaternários")
            ans[name] = np.array([int(ch) for ch in value], dtype=np.int8)
        else:
            ans[name] = np.asarray(value, dtype=np.int8)
            if ans[name].shape != (count,) or not np.isin(ans[name], [0,1,2,3]).all():
                raise ValueError(f"{name}: formato inválido")
    return ans

def genome_text(g):
    return {key: "".join(map(str, val)) for key, val in g.items()}

def beta_max(g, I, J):
    beta = np.float32(1.0 + np.sum(W8*g["beta"],dtype=np.float32)
                     / np.sum(3*W8,dtype=np.float32) * np.float32((I*J/16.)**(1/3)))
    maxconn = int(np.floor(np.sum(W8*np.power(beta, g["max_connections"].astype(np.float32)),dtype=np.float32)))
    return float(beta), maxconn

def h_coefficients(g):
    digits = g["connections"]
    return np.array([np.sum(W64[digits == d],dtype=np.float32) for d in range(4)],dtype=np.float32) * (np.float32(3)*PI32 / (np.float32(4)*np.sum(W64,dtype=np.float32)))

def make_A(g, I, J, K):
    # A[k,j,i], linearizado como i + I*j + I*J*k.
    coeff = (np.float32(2)/np.float32(3))*g["connections"].astype(np.float32)-np.float32(1)
    cx = np.cos(PI32*np.arange(4,dtype=np.float32)[:,None]*
                (np.arange(I,dtype=np.float32)[None,:]/np.float32(I-1))).astype(np.float32)
    cy = np.cos(PI32*np.arange(4,dtype=np.float32)[:,None]*
                (np.arange(J,dtype=np.float32)[None,:]/np.float32(J-1))).astype(np.float32)
    cz = np.cos(PI32*np.arange(4,dtype=np.float32)[:,None]*
                (np.arange(K,dtype=np.float32)[None,:]/np.float32(K-1))).astype(np.float32)
    field = np.zeros((K,J,I), dtype=np.float32)
    for r in range(4):
        for s in range(4):
            for q in range(4):
                field += coeff[q + 4*s + 16*r] * cx[r][None,None,:] * cy[s][None,:,None] * cz[q][:,None,None]
    return np.tanh(field / np.float32(8)).astype(np.float32).ravel()


def flags_by_rank(A, I, J, K, n_inputs, n_outputs):
    N = len(A)
    if n_inputs+n_outputs > N:
        raise ValueError("Número de inputs+outputs maior que número de neurônios")
    by_layer = []
    for k in range(K):
        ix = np.arange(k*I*J,(k+1)*I*J,dtype=np.int32)
        by_layer.append(ix[np.argsort(A[ix],kind="quicksort")])
    # rank 0 de cada camada; depois rank 1 de cada camada; ...
    permutation = np.stack(by_layer,axis=1).ravel()
    is_input = np.zeros(N,dtype=bool); is_output = np.zeros(N,dtype=bool)
    is_input[permutation[:n_inputs]] = True
    is_output[permutation[-n_outputs:]] = True
    return is_input, is_output

def gini_nonnegative(data):
    x = np.sort(np.asarray(data,dtype=np.float64)); n=len(x)
    if n == 0 or np.sum(x) == 0: return 0.
    return float((2*np.dot(np.arange(1,n+1),x)/(n*np.sum(x))) - (n+1)/n)

def metrics(incoming, outgoing, is_input, is_output):
    internal = ~(is_input | is_output)
    counts = {"internal":int(np.count_nonzero((incoming==0)&internal)),
              "output":int(np.count_nonzero((incoming==0)&is_output)),
              "input":int(np.count_nonzero((outgoing==0)&is_input))}
    N=len(incoming)
    eligible = ~is_input
    vals = incoming[eligible]
    tenpct = max(1,int(np.ceil(.1*len(vals))))
    total = np.sum(vals)
    top_share = float(np.partition(vals,-tenpct)[-tenpct:].sum()/total) if total else np.nan
    return {"isolated_total":sum(counts.values()),"isolation_pct":100.*sum(counts.values())/N,
            "isolated_internal":counts["internal"],"isolated_output":counts["output"],"isolated_input":counts["input"],
            "mean_incoming":float(np.mean(vals)),"median_incoming":float(np.median(vals)),
            "gini_incoming":gini_nonnegative(vals),"top_10pct_incoming_share":top_share,
            "zero_incoming_eligible_pct":100.*np.mean(vals==0),"total_valid_connections":int(np.sum(vals))}
```

## 2. Decomposição analítica exata: componente par e componente ímpar

A identidade trigonométrica mostra que a nova fase não acrescenta harmônicos; ela combina componentes preexistentes:

$$z_\phi(a,b)=\cos\phi\;z_{\rm par}(a,b)-\sin\phi\;z_{\rm ímpar}(a,b),$$
$$z_{\rm par}(a,b)=\sum_m c_m\cos(ma)\sin(mb),\qquad z_{\rm ímpar}(a,b)=\sum_m c_m\sin(ma)\sin(mb).$$

Assim, $z_\phi(-a,b)=z_\phi(a,b)$ somente em circunstâncias particulares ou quando $\sin\phi=0$. Para $a$ pequeno, a derivada $\partial z_\phi/\partial a$ passa a ser não nula em geral, **mas ainda depende dos valores de $b$**. Esta célula confere a identidade usando os 64 termos originais, sem depender da seleção de conexões.

```python
HARMONICS=np.arange(1,5,dtype=np.float32)[:,None]
M=np.arange(1,5,dtype=np.float32)

def radial_delta(I,J,K):
    # Mesma normalização individual dos eixos da Parte V.
    k,j,i=np.indices((K,J,I),dtype=np.float32)
    rr=np.sqrt((i/max(1,I-1))**2+(j/max(1,J-1))**2+(k/max(1,K-1))**2)
    return ((8/np.sqrt(3))*rr-4).astype(np.float32).ravel()

def cache_for_mode(case,mode, A_override=None, coeff_override=None):
    A=case['A'] if A_override is None else A_override
    coeff=case['coeff'] if coeff_override is None else coeff_override
    phi=PHASES.get(mode,0.)
    srcdelta=0.; dstdelta=0.; shift=SHIFT0
    if mode in ('radial','radial_phi_p4'):
        srcdelta=case['radial'];dstdelta=case['radial']
    if mode in ('radial_phi_p4','phi_p4_outer_shift_p4'):phi=float(np.pi/4)
    if mode=='target_phi_p4':dstdelta=float(np.pi/4)
    if mode in ('outer_shift_p4','phi_p4_outer_shift_p4'):shift=float(np.pi/4)
    ac=(coeff[:,None]*np.cos(HARMONICS*A[None,:]+srcdelta+np.float32(phi))).astype(np.float32)
    bs=np.sin(HARMONICS*A[None,:]+dstdelta).astype(np.float32)
    return ac,bs,np.float32(shift)

def affinity(ac,bs,shift,src,dest):
    z=np.zeros(len(dest),dtype=np.float32)
    for m in range(4):z+=ac[m,src]*bs[m,dest]
    return np.sin(z+shift).astype(np.float32)

def z_direct64(a,b,genes,phi):
    u=genes['connections'].astype(np.float32)+1
    w=np.sqrt(np.arange(1,65,dtype=np.float32))
    return np.float32(3*np.pi/(4*np.sum(w,dtype=np.float32)))*np.sum(
        w*np.cos(a*u+np.float32(phi))*np.sin(b*u),dtype=np.float32)

c=h_coefficients(random_genome(321));rg=np.random.default_rng(921)
a=rg.uniform(-1,1,200);b=rg.uniform(-1,1,200)
for phi in PHASES.values():
    ze=np.sum(c[:,None]*np.cos(M[:,None]*a)*np.sin(M[:,None]*b),axis=0)
    zo=np.sum(c[:,None]*np.sin(M[:,None]*a)*np.sin(M[:,None]*b),axis=0)
    zd=np.sum(c[:,None]*np.cos(M[:,None]*a+phi)*np.sin(M[:,None]*b),axis=0)
    assert np.max(np.abs(zd-(np.cos(phi)*ze-np.sin(phi)*zo)))<2e-6
_g=random_genome(214);_c=h_coefficients(_g)
for phi in PHASES.values():
    for a0,b0 in [(.31,-.26),(-.14,.17),(.72,.02)]:
        grouped=np.sum(_c*np.cos(M*a0+phi)*np.sin(M*b0))
        assert abs(grouped-z_direct64(np.float32(a0),np.float32(b0),_g,phi))<3e-6
assert np.max(abs(np.sin(M*.21)-np.sin(M*.21)))==0
print('Identidade par/ímpar e equivalência com os 64 genes: OK')
```

## 3. Construção dos 11 casos — mesmos genomas, papéis e candidatos

```python
cases=[]
for ix,rec in enumerate(selected):
    size=tuple(rec['size']);genes=parse_genome(rec['genes']);A=make_A(genes,*size)
    roles=flags_by_rank(A,*size,N_INPUTS,N_OUTPUTS)
    beta,nmax=beta_max(genes,size[0],size[1]);coeff=h_coefficients(genes)
    c={'case':f'{size[0]}x{size[1]}x{size[2]}_g{ix+1:02d}',
       'genome_seed':rec['seed'],'size':size,'genes':genes,'A':A,'roles':roles,
       'coeff':coeff,'beta':beta,'nmax':nmax,'radial':radial_delta(*size)}
    c['cache']={mode:cache_for_mode(c,mode) for mode in MODES}
    cases.append(c)
assert len(cases)==len(selected)
assert all(np.isclose(c['coeff'].sum(),3*np.pi/4,atol=1e-5) for c in cases)
print([(c['case'],c['nmax']) for c in cases])
```

## 4. Seleção idêntica em todas as condições

Para cada origem não-output, enumeramos os destinos em camadas a até duas posições de distância, removemos apenas o próprio neurônio, selecionamos até $N_{\max}$ dos **maiores $H>0$** e *depois* descartamos os destinos proibidos (inputs, ou outputs no caso de origem input). O comparador de scores é idêntico em todas as fases. Os rankings em empates `float32` podem diferir entre ambientes.

```python
def choose_scores(scores,dest,src,nmax,inputs,outputs):
    pos=np.flatnonzero(scores>0)
    take=min(nmax,len(pos))
    if take==0:sel=np.empty(0,dtype=np.int32)
    elif take<len(pos):sel=pos[np.argpartition(scores[pos],-take)[-take:]]
    else:sel=pos
    legal=(~inputs[dest]) & (~(inputs[src]&outputs[dest]))
    legal_idx=sel[legal[sel]]
    return dest[legal_idx],scores[legal_idx],pos,legal,sel

# Sanidade com toy-case e escolha dos maiores, não dos mínimos
v=np.array([.1,.8,-.3,.45,.05],dtype=np.float32)
t=np.arange(5);f=np.zeros(5,dtype=bool)
chosen,weights,eligible,legal,sel=choose_scores(v,t,0,2,f,f)
assert set(chosen.tolist())=={1,3} and np.all(weights>0)
print('Seleção top positiva e filtro posterior: OK')
```

## 5. Experimento completo: isolamento, concentração, saídas, pesos e oportunidades

Para diagnosticar **por que** um destino permaneceu isolado, distinguimos: (i) nenhum $H>0$ legal; (ii) pelo menos um $H>0$ legal, mas nenhuma seleção no top. Contabilizamos esses casos **separadamente para cada fase**. Os inputs sem saídas são medidos à parte, como no builder original.

```python
def run_case(case):
    I,J,K=case['size'];N=len(case['A']);L=I*J
    inp,out=case['roles'];nmax=case['nmax']
    sources=np.flatnonzero(~out)
    windows=[np.arange(max(0,k-2)*L,min(K,k+3)*L,dtype=np.int32) for k in range(K)]
    rng=np.random.default_rng(np.random.SeedSequence([BASE_SEED,case['genome_seed'],444]))
    sample=set(rng.choice(sources,size=min(SAMPLED_SOURCES,len(sources)),replace=False).tolist())
    indeg={m:np.zeros(N,dtype=np.int32) for m in MODES}
    outdeg={m:np.zeros(N,dtype=np.int32) for m in MODES}
    opportunities={m:np.zeros(N,dtype=np.int32) for m in MODES}
    selected_h={m:[] for m in MODES}
    sampled_edges={m:{} for m in MODES}
    counter={m:{'potential':0,'positive':0,'selected_pre_filter':0,'selected_valid':0,
                'sum_H':0.,'discarded':0} for m in MODES}
    for src in sources:
        dst=windows[int(src//L)];dst=dst[dst!=src]
        for mode in MODES:
            ac,bs,shift=case['cache'][mode]
            h=affinity(ac,bs,shift,int(src),dst)
            chosen,weights,eligible,legal,sel=choose_scores(h,dst,int(src),nmax,inp,out)
            # Atualiza oportunidades somente em destinos legalmente conectáveis.
            opportunities[mode][dst[(h>0)&legal]]+=1
            indeg[mode][chosen]+=1
            outdeg[mode][src]+=len(chosen)
            d=counter[mode]
            d['potential']+=len(dst)
            d['positive']+=len(eligible)
            d['selected_pre_filter']+=len(sel)
            d['selected_valid']+=len(chosen)
            d['sum_H']+=float(weights.sum(dtype=np.float64))
            d['discarded']+=len(sel)-len(chosen)
            if src in sample:
                sampled_edges[mode][int(src)]=set(chosen.tolist())
                selected_h[mode].extend(weights.tolist())
    rows=[];causes=[]
    for mode in MODES:
        meas=metrics(indeg[mode],outdeg[mode],inp,out);d=counter[mode]
        opportunity=opportunities[mode];eligible=~inp
        n_no=int(np.sum(eligible&(opportunity==0)))
        n_lost=int(np.sum(eligible&(opportunity>0)&(indeg[mode]==0)))
        n_seen=int(np.sum(eligible&(indeg[mode]>0)))
        assert n_no+n_lost+n_seen==int(np.sum(eligible))
        rows.append({'case':case['case'],'seed':case['genome_seed'],'neurons':N,
            'size_x':I,'size_y':J,'size_z':K,'mode':mode,
            'phase':PHASES.get(mode,np.nan), 'Nmax':nmax,'beta':case['beta'],
            **meas,'positive_candidate_pct':100*d['positive']/d['potential'],
            'mean_selected_H':d['sum_H']/d['selected_valid'] if d['selected_valid'] else np.nan,
            'filter_discarded_pct':100*d['discarded']/d['selected_pre_filter'] if d['selected_pre_filter'] else np.nan})
        causes.append({'case':case['case'],'mode':mode,'eligible_destinations':int(np.sum(eligible)),
            'no_legal_positive_opportunity':n_no,'positive_but_not_selected':n_lost,
            'selected_destinations':n_seen,
            'isolated_outputs_with_positive_opportunity':int(np.sum(out&(opportunity>0)&(indeg[mode]==0))),
            'positive_opportunities_median':float(np.median(opportunity[eligible]))})
    return rows,causes,sampled_edges,selected_h,indeg,opportunities

# Verificação de um cérebro pequeno antes da análise dos 11 genomas
_toy=cases[0].copy();_g=_toy['genes'];_toy['size']=(5,5,4);_toy['A']=make_A(_g,5,5,4)
_toy['roles']=flags_by_rank(_toy['A'],5,5,4,N_INPUTS,N_OUTPUTS)
_toy['radial']=radial_delta(5,5,4);_toy['cache']={m:cache_for_mode(_toy,m) for m in MODES}
_chk=run_case(_toy)
assert len(_chk[0])==len(MODES) and all(x['eligible_destinations']==90 for x in _chk[1])
print('Simulação pequena / contagem de oportunidades: OK')
```

## 6. Execução pareada e reprodução obrigatória do $H$ original

```python
all_rows=[];all_causes=[];edges_by_case={};weights_by_case={};indeg_by_case={}
t0=time.perf_counter()
for c in cases:
    start=time.perf_counter()
    rows,causes,edges,weights,indeg,opps=run_case(c)
    all_rows+=rows;all_causes+=causes
    edges_by_case[c['case']]=edges;weights_by_case[c['case']]=weights;indeg_by_case[c['case']]=indeg
    ref=next(r for r in rows if r['mode']==BASELINE)
    phi=next(r for r in rows if r['mode']=='phi_p4')
    print(f"{c['case']}: original={ref['isolation_pct']:.2f}%; φ=π/4={phi['isolation_pct']:.2f}%; t={time.perf_counter()-start:.1f}s",flush=True)
print('Tempo total:',round(time.perf_counter()-t0,1),'s')
df=pd.DataFrame(all_rows);causes_df=pd.DataFrame(all_causes)
summary=df.groupby('mode',as_index=False).agg(n_genomes=('case','nunique'),
    isolation_pct=('isolation_pct','mean'),gini=('gini_incoming','mean'),
    top10=('top_10pct_incoming_share','mean'),isolated_outputs=('isolated_output','sum'),
    isolated_inputs=('isolated_input','sum'),mean_selected_H=('mean_selected_H','mean'),
    total_edges=('total_valid_connections','sum'),positive_pct=('positive_candidate_pct','mean'))
summary['label']=summary['mode'].map(MODE_LABELS)
summary=summary.sort_values('isolation_pct')
df.to_csv(OUTPUT_DIR/'per_genome_modes.csv',index=False)
summary.to_csv(OUTPUT_DIR/'summary_modes.csv',index=False)
causes_df.to_csv(OUTPUT_DIR/'isolation_causes_by_mode.csv',index=False)
# Reproduzir exatamente a Parte I em cada genoma e confirmar que mudar a fase não alterou indevidamente o baseline.
check=df[df['mode']==BASELINE].merge(baseline_reference,left_on='seed',right_on='seed',validate='one_to_one')
max_isolation_error=float(abs(check['isolation_pct']-check['isolation_H_pct']).max())
edge_error=abs(check['total_valid_connections']-check['valid_edges_H'])
print('Validação da referência: erro máximo de isolamento:',max_isolation_error,
      'p.p.; diferença máxima de arestas:',int(edge_error.max()))
assert max_isolation_error<.3, 'Falha na reprodução do isolamento da Parte I!'
assert (edge_error<=np.maximum(10,check['valid_edges_H']*.005)).all(), 'Falha na reprodução das arestas!'
display(summary.round(4))
```

## 7. Pergunta A — qual fase reduz isolamento, sem enfraquecer demais os pesos?

A análise compara fases apenas dentro da família de origem (`phi_*`). Cada genoma tem peso igual nas médias. **Não confundir** a amplitude do campo com o número total de conexões; o número de conexões e os outputs isolados são informados separadamente.

## 8. Pergunta B — quebra da simetria: $H(a,b)$ versus $H(-a,b)$

A identidade original é exata e independe do histograma. Medimos tanto a separação de afinidades para pares sintéticos simétricos $\pm a$ quanto a mudança nos conjuntos selecionados por origens reais com sinais opostos e $|A|$ semelhante. A segunda medição é condicionada à existência desses pares na mesma camada, e não deve ser confundida com a primeira.

```python
rng=np.random.default_rng(BASE_SEED+202)
q_a=rng.uniform(.03,.7,2500);q_b=rng.uniform(-1,1,2500)
sep_rows=[]
for phi in PHASES.values():
    c=cases[0]['coeff'].astype(np.float64)
    def zz(a,b):return sum(c[m-1]*np.cos(m*a+phi)*np.sin(m*b) for m in range(1,5))
    hp=np.sin(SHIFT0+zz(q_a,q_b));hn=np.sin(SHIFT0+zz(-q_a,q_b))
    sep_rows.append({'phase':phi,'phase_pi':phi/np.pi,'mean_abs_H_sign_separation':float(np.mean(abs(hp-hn))),
        'median_abs_H_sign_separation':float(np.median(abs(hp-hn)))})
symmetry=pd.DataFrame(sep_rows)
assert abs(symmetry.loc[symmetry.phase==0,'mean_abs_H_sign_separation'].iloc[0])<1e-12
pairs=[];overlap=[]
for case in cases:
    key=case['case'];A=case['A'];I,J,K=case['size'];L=I*J
    origin_set=sorted(edges_by_case[key][BASELINE].keys())
    bylayer={k:np.array([s for s in origin_set if s//L==k],dtype=np.int32) for k in range(K)}
    rng=np.random.default_rng(np.random.SeedSequence([BASE_SEED,case['genome_seed'],778]))
    for layer,sources in bylayer.items():
        neg=sources[A[sources]<0];pos=sources[A[sources]>=0]
        if len(neg)==0 or len(pos)==0:continue
        # Comparar pares com sinais opostos e |A| próximos no mesmo plano.
        chosen_neg=rng.choice(neg,size=min(RANK_PAIR_SOURCES,len(neg)),replace=False)
        for src in chosen_neg:
            dst_src=int(pos[np.argmin(np.abs(abs(A[pos])-abs(A[src])))])
            pairs.append((key,int(src),dst_src,float(abs(abs(A[src])-abs(A[dst_src])))))
            for mode in MODES:
                a=edges_by_case[key][mode][int(src)];b=edges_by_case[key][mode][dst_src]
                overlap.append({'case':key,'mode':mode,'src1':int(src),'src2':dst_src,
                    'delta_absA':float(abs(abs(A[src])-abs(A[dst_src]))),
                    'jaccard':len(a&b)/len(a|b) if a|b else 1.})
pair_df=pd.DataFrame(overlap)
pair_df.to_csv(OUTPUT_DIR/'opposite_sign_similar_absA_overlap.csv',index=False)
symmetry.to_csv(OUTPUT_DIR/'analytic_sign_separation.csv',index=False)
fig,axs=plt.subplots(1,2,figsize=(13,4),constrained_layout=True)
axs[0].plot(symmetry.phase_pi,symmetry.mean_abs_H_sign_separation,'o-')
axs[0].set(xlabel='φ/π',ylabel='média |H(a,b)−H(−a,b)|',title='Separação algébrica do sinal')
by_mode=pair_df.groupby('mode')['jaccard'].mean()
for mode in PHASES:axs[1].bar(MODE_LABELS[mode],by_mode[mode])
axs[1].tick_params(axis='x',labelrotation=55)
axs[1].set(ylabel='Jaccard médio',title='Origens reais, |A| próximo e sinais opostos')
fig.savefig(OUTPUT_DIR/'02_sign_symmetry_and_rank_overlap.png',dpi=150);plt.show()
print('Pares fonte opostos encontrados:',len(pairs))
display(symmetry.round(4));display(by_mode.to_frame('jaccard').round(4))
```

## 9. Pergunta C — diferenças reais de rankings e competição pelo top-$N_{\max}$

A comparação utiliza exatamente as mesmas fontes amostradas por genoma em todos os modos. Medimos Jaccard entre os conjuntos escolhidos em duas origens da mesma camada e Jaccard das arestas **da mesma origem** antes e depois de introduzir $\phi$. Um Jaccard menor entre duas origens, por si só, não prova que a cobertura global melhorou.

```python
rank_rows=[];changed_rows=[]
for case in cases:
    key=case['case'];I,J,K=case['size'];L=I*J
    edges=edges_by_case[key];sources=list(edges[BASELINE].keys())
    rng=np.random.default_rng(np.random.SeedSequence([BASE_SEED,case['genome_seed'],868]))
    # Pares de fontes fixos para todos os modos: evitar variabilidade amostral na comparação.
    fixed_pairs=[]
    for layer in range(K):
        srcs=np.array([s for s in sources if s//L==layer],dtype=np.int32)
        if len(srcs)<2:continue
        fixed_pairs.extend([tuple(map(int,rng.choice(srcs,size=2,replace=False)))
                            for _ in range(RANK_PAIR_SOURCES)])
    for mode in MODES:
        baseline_js=[];cross_js=[]
        for src in sources:
            x=edges[BASELINE][src];y=edges[mode][src]
            cross_js.append(len(x&y)/len(x|y) if x|y else 1.)
        for a,b in fixed_pairs:
            aa=edges[mode][a];bb=edges[mode][b]
            if aa|bb:baseline_js.append(len(aa&bb)/len(aa|bb))
        rank_rows.append({'case':key,'mode':mode,'n_pairs':len(baseline_js),
            'mean_same_layer_jaccard':float(np.mean(baseline_js)) if baseline_js else np.nan,
            'mean_edge_jaccard_vs_original':float(np.mean(cross_js))})
rank_df=pd.DataFrame(rank_rows)
rank_df.to_csv(OUTPUT_DIR/'rank_overlap_by_mode.csv',index=False)
metrics_df=rank_df.groupby('mode')[['mean_same_layer_jaccard','mean_edge_jaccard_vs_original']].mean()
fig,ax=plt.subplots(figsize=(10,4),constrained_layout=True)
order=list(PHASES)
ax.bar(np.arange(len(order)),metrics_df.loc[order,'mean_same_layer_jaccard'])
ax.set_xticks(range(len(order)),[MODE_LABELS[x] for x in order],rotation=45,ha='right')
ax.set(ylabel='Jaccard médio entre origens',title='Preferências compartilhadas, mesma camada')
fig.savefig(OUTPUT_DIR/'03_shared_ranking.png',dpi=150);plt.show()
display(metrics_df.round(4))
```

## 11. Pergunta E — interação da fase com o campo radial e com o *shift* externo

Os dois controles são propositadamente separados:

- `radial`: $\cos(mA_p+\delta_p)\sin(mA_q+\delta_q)$, com $\delta_p=\frac8{\sqrt3}\sqrt{(i/(I-1))^2+(j/(J-1))^2+(k/(K-1))^2}-4$.
- `radial_phi_p4`: acrescenta $\phi=\pi/4$ **apenas ao cosseno da origem**, além da radial em origem e destino.
- `outer_shift_p4`: mantém $\phi=0$ e altera apenas o deslocamento externo para $s=\pi/4$.
- `phi_p4_outer_shift_p4`: combina a fase de origem com $s=\pi/4$, sem radial.

A comparação inclui controles individuais e combinados, para distinguir benefício aditivo, interação ou conflito.

```python
interaction=['phi_0','phi_p4','radial','radial_phi_p4','outer_shift_p4','phi_p4_outer_shift_p4']
display(summary.set_index('mode').loc[interaction,
    ['label','isolation_pct','gini','isolated_outputs','mean_selected_H','total_edges']].round(4))
fig,axs=plt.subplots(1,2,figsize=(13,4.5),constrained_layout=True)
vals=summary.set_index('mode').loc[interaction]
axs[0].bar(range(len(vals)),vals.isolation_pct)
axs[0].set_xticks(range(len(vals)),vals.label,rotation=45,ha='right')
axs[0].set(ylabel='isolamento (%)',title='Fases, radial e shift externo')
axs[1].bar(range(len(vals)),vals.mean_selected_H)
axs[1].set_xticks(range(len(vals)),vals.label,rotation=45,ha='right')
axs[1].set(ylabel='média H selecionado',title='Custo potencial em pesos sinápticos')
fig.savefig(OUTPUT_DIR/'05_phase_interactions.png',dpi=150);plt.show()
```

## 12. Pergunta F — sensibilidade genética e sobrevivência das conexões após mutação

Alteramos um dígito do gene `connections` por $\pm1$ dentro de $\{0,1,2,3\}$; recalculamos **$A$ e os coeficientes $c_m$**. A fase e a radial (quando aplicável) permanecem fixas, assim como os papéis input/output e $N_{\max}$ para isolar a mudança de $H$. Reavaliamos as mesmas fontes e destinos candidatos antes e depois, medindo alteração média de $H$ e Jaccard das arestas legais escolhidas.

```python
MUT_MODES=['phi_0','phi_p8','phi_p4','phi_p2','phi_neg_p4','radial','radial_phi_p4']
mut_rows=[]
for case in cases:
    I,J,K=case['size'];L=I*J;inp,out=case['roles']
    rng=np.random.default_rng(np.random.SeedSequence([BASE_SEED,case['genome_seed'],333]))
    sources=np.flatnonzero(~out)
    chosen=rng.choice(sources,size=min(STABILITY_ORIGINS,len(sources)),replace=False)
    windows=[np.arange(max(0,k-2)*L,min(K,k+3)*L,dtype=np.int32) for k in range(K)]
    for mu in range(STABILITY_MUTATIONS):
        genes={k:v.copy() for k,v in case['genes'].items()}
        idx=int(rng.integers(0,64));old_digit=int(genes['connections'][idx])
        direction=int(rng.choice([-1,1]))
        if old_digit==0:direction=1
        if old_digit==3:direction=-1
        genes['connections'][idx]=old_digit+direction
        mutant_A=make_A(genes,I,J,K);mutant_coeff=h_coefficients(genes)
        mutated={m:cache_for_mode(case,m,A_override=mutant_A,coeff_override=mutant_coeff) for m in MUT_MODES}
        for mode in MUT_MODES:
            oldc=case['cache'][mode];newc=mutated[mode]
            js=[];dh=[]
            for src in chosen:
                src=int(src);dest=windows[src//L];dest=dest[dest!=src]
                h0=affinity(*oldc,src,dest);h1=affinity(*newc,src,dest)
                a,_,_,_,_=choose_scores(h0,dest,src,case['nmax'],inp,out)
                b,_,_,_,_=choose_scores(h1,dest,src,case['nmax'],inp,out)
                aa=set(a.tolist());bb=set(b.tolist())
                js.append(len(aa&bb)/len(aa|bb) if aa|bb else 1.)
                dh.append(float(np.mean(np.abs(h0-h1))))
            mut_rows.append({'case':case['case'],'mode':mode,'mutation':mu,
                'gene_index':idx,'old_digit':old_digit,'new_digit':int(genes['connections'][idx]),
                'sampled_origins':len(chosen),'jaccard':float(np.mean(js)),
                'mean_abs_H_change':float(np.mean(dh))})
mut_df=pd.DataFrame(mut_rows)
mut_df.to_csv(OUTPUT_DIR/'mutation_stability.csv',index=False)
mut_summary=mut_df.groupby('mode')[['jaccard','mean_abs_H_change']].agg(['mean','std'])
fig,axs=plt.subplots(1,2,figsize=(13,4.3),constrained_layout=True)
order=MUT_MODES
for ax,col in zip(axs,['jaccard','mean_abs_H_change']):
    vals=[mut_df[mut_df['mode']==m][col].mean() for m in order]
    ax.bar(range(len(vals)),vals)
    ax.set_xticks(range(len(vals)),[MODE_LABELS[x] for x in order],rotation=45,ha='right')
    ax.set(ylabel=col,title='Mutações de um dígito · '+col)
fig.savefig(OUTPUT_DIR/'06_mutation_stability.png',dpi=150);plt.show()
display(mut_summary.round(4))
```

## 13. Pergunta G — derivada local e redundância perto de $A\approx0$

A derivada do somatório com relação ao escalar de origem $a$ é

$$\frac{\partial z_\phi}{\partial a}=-\sum_m mc_m\sin(ma+\phi)\sin(mb).$$

Isso é exatamente zero em $a=0$ quando $\phi=0$. Para $\phi\ne0$, a derivada normalmente deixa de ser zero, **mas permanece pequena para destinos com $b\approx0$**. Medimos as sensibilidades de $z$ amostrando destinos reais, distinguindo alteração algébrica de ganho comprovado de conectividade.

```python
sensitivity=[]
for case in cases:
    A=case['A'];rng=np.random.default_rng(np.random.SeedSequence([BASE_SEED,case['genome_seed'],858]))
    candidates=rng.choice(len(A),size=min(2500,len(A)),replace=False)
    b=A[candidates];coeff=case['coeff'].astype(np.float64)
    for mode,phi in PHASES.items():
        da=np.zeros(len(candidates),dtype=np.float64)
        for m in range(1,5):da-=m*coeff[m-1]*np.sin(phi)*np.sin(m*b)
        sensitivity.append({'case':case['case'],'mode':mode,'phase':phi,
            'mean_abs_dz_da_at_a0':float(np.mean(abs(da))),
            'p90_abs_dz_da_at_a0':float(np.quantile(abs(da),.90)),
            'fraction_targets_absA_le_01':float(np.mean(abs(b)<=.1))})
sens=pd.DataFrame(sensitivity)
sens.to_csv(OUTPUT_DIR/'sensitivity_near_zero.csv',index=False)
display(sens.groupby('mode')[['mean_abs_dz_da_at_a0','fraction_targets_absA_le_01']].mean().round(5))
```

## 14. Síntese automática: resultados, perguntas seguintes e decisões condicionais

O notebook deve gerar respostas para as hipóteses **na mesma execução**, não exigir outro notebook para apenas descobrir qual diagnóstico faltou. A síntese evita declarar vitória baseada exclusivamente em isolamento: exige examinar saídas, pesos positivos, número de conexões e mutação. As conclusões são comparações observacionais entre fórmulas; não demonstram aprendizagem ou cognição.

```python
x=summary.set_index('mode');baseline=x.loc[BASELINE]
sweep=summary[summary['mode'].isin(PHASES)].sort_values('isolation_pct')
best=sweep.iloc[0];bestmode=best['mode']
reports=[]
def report(question,evidence,reading):reports.append({'questao':question,'medida':evidence,'interpretacao_condicional':reading})
report('Quebrar a paridade reduz isolamento?',
    f"melhor fase={MODE_LABELS[bestmode]}, isolamento={best['isolation_pct']:.2f}% vs {baseline.isolation_pct:.2f}% original",
    'Uma melhora concomitante à redução da sobreposição entre rankings seria compatível com a hipótese. Não comprova causalidade exclusiva da paridade.')
report('O sinal da fase importa?',
    f"φ=+π/4: {x.loc['phi_p4'].isolation_pct:.2f}%; φ=−π/4: {x.loc['phi_neg_p4'].isolation_pct:.2f}%",
    'Diferença sinalizada sugere assimetria da população de A e/ou do destino sin(mb), mesmo com intensidade de quebra semelhante.')
report('Os destinos isolados tiveram oportunidades positivas?',
    f"Original preteridos={int(causes_sum.set_index('mode').loc['phi_0','positive_but_not_selected'])}; melhor fase preteridos={int(causes_sum.set_index('mode').loc[bestmode,'positive_but_not_selected'])}",
    'Se os preteridos dominam, a competição pelos rankings continua central mesmo quando a simetria é quebrada.')
report('A quebra de paridade altera rankings de origens semelhantes?',
    f"Jaccard de pares reais: original={pair_df[pair_df['mode']=='phi_0'].jaccard.mean():.3f}, melhor fase={pair_df[pair_df['mode']==bestmode].jaccard.mean():.3f}",
    'Jaccard menor pode significar diferenciação, mas cobertura deve ser avaliada separadamente.')
report('A fase prejudica os pesos ou outputs?',
    f"Original H médio={baseline.mean_selected_H:.3f}, outputs isolados={int(baseline.isolated_outputs)}; melhor fase H médio={best['mean_selected_H']:.3f}, outputs isolados={int(best['isolated_outputs'])}",
    'Não adotar fórmula que produza ganho aparente de cobertura às custas de pesos excessivamente baixos ou muitas saídas sem entradas.')
report('A radial e o shift externo interagem com a fase?',
    f"radial={x.loc['radial'].isolation_pct:.2f}%, radial+φ={x.loc['radial_phi_p4'].isolation_pct:.2f}%, shift π/4={x.loc['outer_shift_p4'].isolation_pct:.2f}%, φ+shift π/4={x.loc['phi_p4_outer_shift_p4'].isolation_pct:.2f}%",
    'Comparar controles pareados antes de somar mecanismos; não supor benefícios cumulativos.')
report('A estabilidade genética diminuiu?',
    f"Jaccard mutação: original={mut_df[mut_df['mode']=='phi_0'].jaccard.mean():.3f}; melhor fase testada={mut_df[mut_df['mode']==bestmode].jaccard.mean():.3f}" if bestmode in MUT_MODES else 'Melhor fase fora da amostra de mutação; rever MUT_MODES',
    'Evidência exploratória: poucas mutações; roles e Nmax mantidos fixos durante o teste.')
report_df=pd.DataFrame(reports);report_df.to_csv(OUTPUT_DIR/'interpretation_and_next_questions.csv',index=False)
for r in reports:
    print('\nPERGUNTA:',r['questao']);print('MEDIDA:',r['medida']);print('CRITÉRIO:',r['interpretacao_condicional'])
```

# Neurons — Parte XII: auditoria final de $H$ e decisão de engenharia

**Último experimento nesta série.** Congelamos as candidatas da Parte XI. Não há busca de novos parâmetros nem escolha por um único número. A tarefa é avaliar se a cobertura excepcional se transforma em **caminhos úteis entre inputs e outputs**, sem homofilia degenerada, empates numéricos problemáticos ou fragilidade evolutiva excessiva.

**Distinção decisiva:** caminhos num grafo e transporte em um passeio ponderado **não** demonstram cognição, estabilidade da dinâmica biológica artificial ou desempenho em labirintos. Este notebook pode escolher uma **formulação final para implementação e teste funcional no Neurons**, mas não demonstrar que ela produz inteligência.

### Protocolo congelado

* Genomas: os mesmos 11 da Parte XI, inclusive os 4 previamente reservados (não retreinar neste holdout).
* Neurônios, máscaras de input/output, $A$, $N_{\max}$, janela $|k_p-k_q|\le 2$: mesmos algoritmos das Partes IX–XI.
* Seleção: os maiores $H>0$ **antes** do filtro de arestas proibidas e **antes** do resgate.
* Fórmulas: original, radial, `p0689`, `p0579`, `p1384`, `p0199` e `p0328` (esta última como controle de baixa taxa de isolamento).
* Métricas novas: alcançabilidade dirigida, componentes fortes/fracas, reciprocidade, corredores input–output, caminhada ponderada **apenas como proxy**, precisão FP32/FP64, sensibilidade a empates, mutações pareadas e reclassificação input/output.
* Política de decisão: filtros e pesos definidos **antes** da avaliação. Se ninguém satisfizer os critérios, a resposta será **nenhuma vencedora comprovada**, acompanhada da melhor candidata provisória e dos motivos.

A avaliação cobre a **topologia antes do resgate**. Isso evita que o resgate esconda problemas da função $H$; o comportamento da dinâmica real deverá ser validado dentro do Godot.

## 1. Reprodução do genoma, campo $A$, papéis neuronais e seleção

```python
PI32 = np.float32(np.pi)
W8 = np.sqrt(np.arange(1, 9, dtype=np.float32))
W64 = np.sqrt(np.arange(1, 65, dtype=np.float32))

def random_genome(seed):
    rng = np.random.default_rng(seed)
    # O construtor C++ gera dígitos independentes e uniformes de 0 a 3.
    return {"connections": rng.integers(0,4,64,dtype=np.int8),
            "beta": rng.integers(0,4,8,dtype=np.int8),
            "max_connections": rng.integers(0,4,8,dtype=np.int8)}

def parse_genome(dictionary):
    lengths = {"connections": 64, "beta": 8, "max_connections": 8}
    ans = {}
    for name, count in lengths.items():
        value = dictionary[name]
        if isinstance(value, str):
            if len(value) != count or not set(value) <= set("0123"):
                raise ValueError(f"{name}: esperado {count} dígitos quaternários")
            ans[name] = np.array([int(ch) for ch in value], dtype=np.int8)
        else:
            ans[name] = np.asarray(value, dtype=np.int8)
            if ans[name].shape != (count,) or not np.isin(ans[name], [0,1,2,3]).all():
                raise ValueError(f"{name}: formato inválido")
    return ans

def genome_text(g):
    return {key: "".join(map(str, val)) for key, val in g.items()}

def beta_max(g, I, J):
    beta = np.float32(1.0 + np.sum(W8*g["beta"],dtype=np.float32)
                     / np.sum(3*W8,dtype=np.float32) * np.float32((I*J/16.)**(1/3)))
    maxconn = int(np.floor(np.sum(W8*np.power(beta, g["max_connections"].astype(np.float32)),dtype=np.float32)))
    return float(beta), maxconn

def h_coefficients(g):
    digits = g["connections"]
    return np.array([np.sum(W64[digits == d],dtype=np.float32) for d in range(4)],dtype=np.float32) * (np.float32(3)*PI32 / (np.float32(4)*np.sum(W64,dtype=np.float32)))

def make_A(g, I, J, K):
    # A[k,j,i], linearizado como i + I*j + I*J*k.
    coeff = (np.float32(2)/np.float32(3))*g["connections"].astype(np.float32)-np.float32(1)
    cx = np.cos(PI32*np.arange(4,dtype=np.float32)[:,None]*
                (np.arange(I,dtype=np.float32)[None,:]/np.float32(I-1))).astype(np.float32)
    cy = np.cos(PI32*np.arange(4,dtype=np.float32)[:,None]*
                (np.arange(J,dtype=np.float32)[None,:]/np.float32(J-1))).astype(np.float32)
    cz = np.cos(PI32*np.arange(4,dtype=np.float32)[:,None]*
                (np.arange(K,dtype=np.float32)[None,:]/np.float32(K-1))).astype(np.float32)
    field = np.zeros((K,J,I), dtype=np.float32)
    for r in range(4):
        for s in range(4):
            for q in range(4):
                field += coeff[q + 4*s + 16*r] * cx[r][None,None,:] * cy[s][None,:,None] * cz[q][:,None,None]
    return np.tanh(field / np.float32(8)).astype(np.float32).ravel()


def flags_by_rank(A, I, J, K, n_inputs, n_outputs):
    N = len(A)
    if n_inputs+n_outputs > N:
        raise ValueError("Número de inputs+outputs maior que número de neurônios")
    by_layer = []
    for k in range(K):
        ix = np.arange(k*I*J,(k+1)*I*J,dtype=np.int32)
        by_layer.append(ix[np.argsort(A[ix],kind="quicksort")])
    # rank 0 de cada camada; depois rank 1 de cada camada; ...
    permutation = np.stack(by_layer,axis=1).ravel()
    is_input = np.zeros(N,dtype=bool); is_output = np.zeros(N,dtype=bool)
    is_input[permutation[:n_inputs]] = True
    is_output[permutation[-n_outputs:]] = True
    return is_input, is_output

def gini_nonnegative(data):
    x = np.sort(np.asarray(data,dtype=np.float64)); n=len(x)
    if n == 0 or np.sum(x) == 0: return 0.
    return float((2*np.dot(np.arange(1,n+1),x)/(n*np.sum(x))) - (n+1)/n)

def metrics(incoming, outgoing, is_input, is_output):
    internal = ~(is_input | is_output)
    counts = {"internal":int(np.count_nonzero((incoming==0)&internal)),
              "output":int(np.count_nonzero((incoming==0)&is_output)),
              "input":int(np.count_nonzero((outgoing==0)&is_input))}
    N=len(incoming)
    eligible = ~is_input
    vals = incoming[eligible]
    tenpct = max(1,int(np.ceil(.1*len(vals))))
    total = np.sum(vals)
    top_share = float(np.partition(vals,-tenpct)[-tenpct:].sum()/total) if total else np.nan
    return {"isolated_total":sum(counts.values()),"isolation_pct":100.*sum(counts.values())/N,
            "isolated_internal":counts["internal"],"isolated_output":counts["output"],"isolated_input":counts["input"],
            "mean_incoming":float(np.mean(vals)),"median_incoming":float(np.median(vals)),
            "gini_incoming":gini_nonnegative(vals),"top_10pct_incoming_share":top_share,
            "zero_incoming_eligible_pct":100.*np.mean(vals==0),"total_valid_connections":int(np.sum(vals))}
```

## 2. Coeficientes e operações de $H$ — mesma matemática da Parte XI

```python
OPS={k:k for k in ['cos*sin','sin+sin','sin+cos','sin*cos','cos*cos','cs-sc']}
M=np.arange(1,5,dtype=np.float32)
C=float(np.sum(W64,dtype=np.float32))

def get_case(record):
    I,J,K=tuple(record['size']);g=parse_genome(record['genes'])
    A=make_A(g,I,J,K);inp,out=flags_by_rank(A,I,J,K,N_INPUTS,N_OUTPUTS)
    beta,nmax=beta_max(g,I,J)
    gw=np.array([np.sum(W64[g['connections']==d],dtype=np.float32) for d in range(4)],dtype=np.float32)
    weights=gw/np.sum(W64,dtype=np.float32)
    cos=np.cos(M[:,None]*A[None,:]).astype(np.float32)
    sin=np.sin(M[:,None]*A[None,:]).astype(np.float32)
    return {'seed':record['seed'],'case':f'{I}x{J}x{K}_{record["seed"]}',
            'size':(I,J,K),'genes':g,'A':A,'input':inp,'output':out,
            'beta':beta,'nmax':nmax,'w':weights,'cos':cos,'sin':sin}

def features(case,src,dst):
    w=case['w'];cx=case['cos'][:,src];sx=case['sin'][:,src]
    cy=case['cos'][:,dst];sy=case['sin'][:,dst]
    cs=(w*cx)@sy;sc=(w*sx)@cy;ss=(w*sx)@sy;cc=(w*cx)@cy
    cp=float(w@cx);sp=float(w@sx)
    cq=w@cy;sq=w@sy
    return {
        'cos*sin':cs, 'sin*cos':sc, 'sin*sin':ss, 'cos*cos':cc,
        'cos+sin':cp+sq, 'sin+cos':sp+cq,
        'sin+sin':sp+sq, 'cos+cos':cp+cq,
        'cos-sin':cp-sq, 'sin-cos':sp-cq,
        'sin-sin':sp-sq, 'cos-cos':cp-cq,
        'cs+sc':cs+sc,'cs-sc':cs-sc,'cc+ss':cc+ss,'cc-ss':cc-ss}

def window_destinations(case,src):
    I,J,K=case['size'];L=I*J;k=int(src//L)
    dest=np.arange(max(0,k-2)*L,min(K,k+3)*L,dtype=np.int32)
    return dest[dest!=src]

def choose_top_positive(H,nmax,legal_mask=None):
    # H: [config, destinations]. Seleção ANTES da filtragem de input/output.
    H=np.asarray(H,dtype=np.float32)
    if H.ndim==1:H=H[None,:]
    take=min(nmax,H.shape[1]);ids=np.arange(H.shape[0])[:,None]
    if take==0:return np.empty((H.shape[0],0),dtype=np.int32),np.empty((H.shape[0],0),dtype=bool)
    selected=np.argpartition(H,-take,axis=1)[:,-take:]
    keep=H[ids,selected]>0
    if legal_mask is not None:keep &= legal_mask[selected]
    return selected,keep

def outgoing_legal_mask(case,src,dst):
    return (~case['input'][dst]) & (~(case['input'][src] & case['output'][dst]))

TEST_CASE=get_case(RECORDS[0]);rng=np.random.default_rng(120)
for _ in range(60):
    src=int(rng.integers(len(TEST_CASE['A'])));dst=np.array(rng.integers(len(TEST_CASE['A']),size=8))
    v=features(TEST_CASE,src,dst);assert set(OPS).issubset(set(v))
    X=M[:,None]*TEST_CASE['A'][src];Y=M[:,None]*TEST_CASE['A'][dst]
    w=TEST_CASE['w'][:,None]
    assert np.max(np.abs(v['cs+sc']-np.sum(w*np.sin(X+Y),axis=0)))<2e-6
    assert np.max(np.abs(v['cc+ss']-np.sum(w*np.cos(X-Y),axis=0)))<2e-6
    assert np.max(np.abs(v['cs-sc']-np.sum(w*np.sin(Y-X),axis=0)))<2e-6
    assert np.max(np.abs(v['cc-ss']-np.sum(w*np.cos(X+Y),axis=0)))<2e-6
# Exatamente a forma original quando a=π/16 e η=3π/4
src=21;dst=window_destinations(TEST_CASE,src)[:100]
q=features(TEST_CASE,src,dst)['cos*sin']
new=np.sin(np.float32(np.pi/16)+np.float32(3*np.pi/4)*q)
old=np.sin(np.float32(np.pi/16)+np.sum(h_coefficients(TEST_CASE['genes'])[:,None]*TEST_CASE['cos'][:,src,None]*TEST_CASE['sin'][:,dst],axis=0))
assert np.max(abs(new-old))<3e-6
print('Identidades trigonométricas e formulação original: OK')
```

## 3. Montagem integral dos grafos — sem resgate

Geramos uma matriz esparsa dirigida $W\in\mathbb{R}_{\ge0}^{N\times N}$, na qual $W_{pq}$ é o peso $H(p,q)$ **se** a aresta foi escolhida; caso contrário, é zero. Atribuição de input/output e seleção top-$N_{\max}$ seguem a versão de referência, inclusive o filtro *posterior* dos destinos proibidos.

**Importante:** o resultado de `np.argpartition` em empates pode diferir do desempate da implementação C++; a análise FP32/FP64 abaixo quantifica a sensibilidade, mas uma validação final na GDExtension ainda é necessária.

```python
def radial_delta(I,J,K,dtype=np.float32):
    k,j,i=np.indices((K,J,I),dtype=dtype)
    rr=np.sqrt((i/max(I-1,1))**2+(j/max(J-1,1))**2+(k/max(K-1,1))**2)
    return ((8/np.sqrt(3))*rr-4).astype(dtype).ravel()

def q_arrays(c,dtype=np.float32):
    A=c['A'].astype(dtype);w=c['w'].astype(dtype)
    M2=np.arange(1,5,dtype=dtype)
    cc=np.cos(M2[:,None]*A); ss=np.sin(M2[:,None]*A)
    arr={'cos':cc,'sin':ss,'w':w,'S':w@ss,'C':w@cc}
    if dtype==np.float64:
        dg=c['genes']['connections']; w64=np.sqrt(np.arange(1,65,dtype=np.float64));
        arr['w']=np.array([w64[dg==d].sum()/w64.sum() for d in range(4)])
        arr['S']=arr['w']@ss;arr['C']=arr['w']@cc
    if 'radial_baseline' in MODES:
        delta=radial_delta(*c['size'],dtype=dtype)
        arr['rc']=np.cos(M2[:,None]*A+delta[None,:])
        arr['rs']=np.sin(M2[:,None]*A+delta[None,:])
    return arr

def q_eval(c,src,dst,op,arr):
    w=arr['w'];ss=arr['sin'];cc=arr['cos']
    if op=='cos*sin':return (w*cc[:,src])@ss[:,dst]
    if op=='sin*cos':return (w*ss[:,src])@cc[:,dst]
    if op=='cos*cos':return (w*cc[:,src])@cc[:,dst]
    if op=='sin+sin':return arr['S'][src]+arr['S'][dst]
    if op=='sin+cos':return arr['S'][src]+arr['C'][dst]
    if op=='sin(Y-X)':return (w*cc[:,src])@ss[:,dst]-(w*ss[:,src])@cc[:,dst]
    if op=='radial':return (w*arr['rc'][:,src])@arr['rs'][:,dst]
    raise ValueError(op)

def scores(c,src,dst,mid,arr):
    op,a,eta=MODES[mid]
    return np.sin(np.asarray(a,dtype=arr['w'].dtype)+np.asarray(eta,dtype=arr['w'].dtype)
                  *q_eval(c,int(src),dst,op,arr))

def graph_from_H(c,mid):
    N=len(c['A']);src_all=[];dst_all=[];weight_all=[]
    arr=q_arrays(c)
    for src in np.flatnonzero(~c['output']):
        dst=window_destinations(c,int(src))
        H=np.asarray(scores(c,src,dst,mid,arr),dtype=np.float32)
        chosen,keep=choose_top_positive(H,c['nmax'],outgoing_legal_mask(c,int(src),dst))
        ix=chosen[0,keep[0]]
        if not len(ix):continue
        src_all.append(np.full(len(ix),src,dtype=np.int32))
        dst_all.append(dst[ix]);weight_all.append(H[ix])
    rows=np.concatenate(src_all);cols=np.concatenate(dst_all);weights=np.concatenate(weight_all)
    graph=sparse.csr_matrix((weights,(rows,cols)),shape=(N,N),dtype=np.float32)
    graph.sum_duplicates();graph.sort_indices()
    assert graph.nnz==len(rows)
    assert np.all(graph.data>0)
    assert np.all(np.asarray(graph[:,c['input']].sum(axis=0))==0)
    assert np.all(np.asarray(graph[c['output'],:].sum(axis=1))==0)
    return graph

# Reprodução quantitativa: referência deve coincidir em 11 genomas no perfil full.
test_case=get_case(RECORDS[0]); test_graph=graph_from_H(test_case,'p0047')
test_in=np.diff(test_graph.tocsc().indptr)
test_out=np.diff(test_graph.indptr)
test_met=metrics(test_in,test_out,test_case['input'],test_case['output'])
ref_row=prior_per_genome[(prior_per_genome.seed==test_case['seed'])&(prior_per_genome.id=='p0047')].iloc[0]
assert abs(test_met['isolation_pct']-ref_row.isolation_pct)<0.3
assert abs(test_graph.nnz-ref_row.total_valid_connections)<max(10,.005*ref_row.total_valid_connections)
print('Reprodução baseline:',test_case['case'],test_met['isolation_pct'],'% isolamento;',test_graph.nnz,'arestas')
del test_graph
```

## 4. Saúde estrutural: caminhos dirigidos, ilhas, reciprocidade e mistura

* `reachable_outputs_pct`: fração de outputs que podem receber algum sinal de algum input em um grafo dirigido, **sem assumir disparos**.
* `reachable_input_output_pairs_pct`: fração dos 100 pares input-output conectados por caminho.
* `input_output_corridor_pct`: fração de neurônios internos **ao mesmo tempo** alcançáveis a partir dos inputs e capazes de alcançar um output.
* `strong_largest_frac_pct`: tamanho relativo do maior componente fortemente conexo, **não** interpretado automaticamente como qualidade cognitiva.
* `near_A_pct`: proporção de arestas com $|A_p-A_q|<0{,}05$, diagnosticando homofilia exagerada.
* `reciprocity_pct`: fração de arestas que têm uma aresta de retorno.
* `walk_output_mass_12`: massa que chega aos outputs após 12 passos numa caminhada estocástica ponderada; **proxy topológico, não dinâmica neural**. Outputs absorventes, demais sumidouros perdem massa.

```python
def assess_graph(c,graph,mid):
    N=graph.shape[0];I,J,K=c['size'];L=I*J
    inp=np.flatnonzero(c['input']);out=np.flatnonzero(c['output']);internal=~(c['input']|c['output'])
    d_in=csgraph.shortest_path(graph,directed=True,unweighted=True,indices=inp)
    d_out=csgraph.shortest_path(graph.T.tocsr(),directed=True,unweighted=True,indices=out)
    reach_in=np.isfinite(d_in).any(axis=0)
    can_reach_out=np.isfinite(d_out).any(axis=0)
    io=d_in[:,out]
    w_n,labels_w=csgraph.connected_components(graph,directed=True,connection='weak')
    s_n,labels_s=csgraph.connected_components(graph,directed=True,connection='strong')
    largest_w=np.bincount(labels_w,minlength=w_n).max()/N
    largest_s=np.bincount(labels_s,minlength=s_n).max()/N
    adjacency=graph.copy();adjacency.data=np.ones_like(adjacency.data)
    reciprocal=adjacency.multiply(adjacency.T).nnz/graph.nnz if graph.nnz else 0.
    src=np.repeat(np.arange(N,dtype=np.int32),np.diff(graph.indptr))
    dst=graph.indices
    da=np.abs(c['A'][src]-c['A'][dst])
    cross=(src//L != dst//L)
    # Markov simplificado: cada fonte transmite uma unidade repartida pelo peso;
    # os outputs absorvem. Não usa limiares, polaridade ou plasticidade.
    row_sums=np.asarray(graph.sum(axis=1)).ravel()
    inverse=np.divide(1.,row_sums,out=np.zeros_like(row_sums),where=row_sums>0)
    P=sparse.diags(inverse)@graph
    P=P+sparse.csr_matrix((np.ones(len(out)),(out,out)),shape=(N,N))
    prob=np.zeros(N,dtype=np.float64);prob[inp]=1./len(inp)
    walk_results={}
    for step in range(1,WALK_STEPS+1):
        prob=np.asarray(P.T@prob).ravel()
        if step in (2,4,8,12):walk_results[f'walk_output_mass_{step}']=float(prob[out].sum())
    gin=np.diff(graph.tocsc().indptr);gout=np.diff(graph.indptr)
    core=metrics(gin,gout,c['input'],c['output'])
    res={'id':mid,'seed':c['seed'],'case':c['case'],'neurons':N,
         'held_out':c['seed'] in HOLDOUT_SEEDS,'nmax':c['nmax'],'edges':graph.nnz,
         'reachable_outputs_pct':100*np.mean(np.isfinite(io).any(axis=0)),
         'reachable_input_output_pairs_pct':100*np.mean(np.isfinite(io)),
         'unreachable_outputs':int(np.sum(~np.isfinite(io).any(axis=0))),
         'median_io_distance':float(np.median(io[np.isfinite(io)])) if np.isfinite(io).any() else np.nan,
         'p90_io_distance':float(np.percentile(io[np.isfinite(io)],90)) if np.isfinite(io).any() else np.nan,
         'input_reachable_internal_pct':100*np.mean(reach_in[internal]),
         'input_output_corridor_pct':100*np.mean((reach_in&can_reach_out)[internal]),
         'weak_component_count':w_n,'strong_component_count':s_n,
         'weak_largest_frac_pct':100*largest_w,'strong_largest_frac_pct':100*largest_s,
         'reciprocity_pct':100*reciprocal,
         'near_A_pct':100*np.mean(da<.05),'mean_abs_A_diff':float(np.mean(da)),
         'cross_layer_pct':100*np.mean(cross),
         'weight_mean':float(np.mean(graph.data)),
         'weight_p05':float(np.quantile(graph.data,.05)),
         'weight_below_0_1_pct':100*np.mean(graph.data<.1),
         'weight_above_0_999_pct':100*np.mean(graph.data>=.999),
         'weight_exact_one_pct':100*np.mean(graph.data==1.),
         **walk_results, **core}
    return res

# Um gráfico pequeno permite inspecionar a semântica de alcance e a fórmula de massa.
G=sparse.csr_matrix((np.array([.8,.7,.5],dtype=np.float32),([0,1,2],[1,2,3])),shape=(4,4))
assert csgraph.shortest_path(G,indices=[0],directed=True,unweighted=True)[0,3]==3
assert np.isinf(csgraph.shortest_path(G,indices=[3],directed=True,unweighted=True)[0,0])
print('Testes dos caminhos dirigidos: OK')
```

## 6. Precisão numérica, empates e dependência da implementação

**Este é um possível ponto de falha das soluções com $H\approx1$.** Avaliamos fontes fixas e equivalentes nos mesmos genomas e comparamos seleção FP32, FP64 (mesmo $A$ em FP32 para isolar a precisão de $H$), desempate lexicográfico por índice e desempate aleatório **somente entre afinidades exatamente iguais**. Não é correto interpretar esta amostra como prova de equivalência com o desempate do C++: ela indica risco.

```python
def sampled_sources(c,limit,tag):
    rng=np.random.default_rng(np.random.SeedSequence([BASE_SEED,int(c['seed']),int(tag)]))
    candidates=np.flatnonzero(~c['output'])
    if len(candidates)<=limit:return candidates
    return np.sort(rng.choice(candidates,size=limit,replace=False))

def reference_top(h,dst,nmax,legal):
    idx,keep=choose_top_positive(np.asarray(h,dtype=np.float32),nmax,legal)
    return set(dst[idx[0,keep[0]]].tolist())

def sorted_top(h,dst,nmax,legal,rank_noise=None):
    if rank_noise is None:rank_noise=dst # tie break by spatial index
    order=np.lexsort((rank_noise,-np.asarray(h)))[:min(nmax,len(h))]
    return set(dst[order[(h[order]>0)&legal[order]]].tolist())

def jaccard(a,b):return len(a&b)/len(a|b) if a|b else 1.

numeric=[]
for rec in EVAL_RECORDS:
    c=get_case(rec);f32=q_arrays(c,np.float32);f64=q_arrays(c,np.float64)
    for mid in MODES:
        for src in sampled_sources(c,RANK_SOURCES,1207):
            dst=window_destinations(c,int(src));legal=outgoing_legal_mask(c,int(src),dst)
            h32=np.asarray(scores(c,src,dst,mid,f32),dtype=np.float32)
            h64=np.asarray(scores(c,src,dst,mid,f64),dtype=np.float64)
            take=min(c['nmax'],len(dst));original=reference_top(h32,dst,c['nmax'],legal)
            best64=sorted_top(h64,dst,c['nmax'],legal)
            lex32=sorted_top(h32,dst,c['nmax'],legal)
            threshold=np.partition(h32,-take)[-take] if take else np.nan
            count_gt=int(np.count_nonzero(h32>threshold))
            tie_count=int(np.count_nonzero(h32==threshold))
            tie_at_cutoff=(count_gt<take and count_gt+tie_count>take)
            # Desempate verdadeiramente aleatório apenas dentro de blocos exatamente iguais:
            rng=np.random.default_rng(np.random.SeedSequence([BASE_SEED,int(c['seed']),int(src),sum(map(ord,mid))]))
            rnd=sorted_top(h32,dst,c['nmax'],legal,rng.random(len(dst)))
            numeric.append({'id':mid,'seed':c['seed'],'source':int(src),
               'jaccard_fp32_fp64':jaccard(original,best64),
               'jaccard_fp32_lexicographic':jaccard(original,lex32),
               'jaccard_fp32_random_ties':jaccard(original,rnd),
               'ties_straddle_cutoff':int(tie_at_cutoff),
               'ties_at_cutoff':tie_count if tie_at_cutoff else 0,
               'fraction_H_exact_one':float(np.mean(h32==1.0)),
               'min_max_H_difference':float(np.max(abs(h32.astype(np.float64)-h64)))})
num_df=pd.DataFrame(numeric);num_df.to_csv(RESULT_DIR/'numerical_stability_by_source.csv',index=False)
num_summary=num_df.groupby('id',as_index=False).agg(fp_agreement=('jaccard_fp32_fp64','mean'),
   random_tie_agreement=('jaccard_fp32_random_ties','mean'),
   lex_agreement=('jaccard_fp32_lexicographic','mean'),
   cutoff_ties_pct=('ties_straddle_cutoff',lambda x:100*x.mean()),
   exact_one_fraction=('fraction_H_exact_one','mean'))
num_summary.to_csv(RESULT_DIR/'numerical_stability_summary.csv',index=False)
display(num_summary.round(4))
```

## 7. Pequenas mutações genéticas pareadas — estabilidade dos rankings

Mutações quaternárias $\pm1$ em um único dos 64 genes de `connections`, mantendo a mesma escolha de origem e os mesmos destinos candidatos. Registramos o efeito em $H$ e o Jaccard das conexões selecionadas. **O teste primário mantém os papéis input/output fixos**, para isolar a continuidade de $H$; em paralelo registramos quantos neurônios mudariam de papel se a classificação fosse recalculada.

```python
mut_rows=[]
for rec in EVAL_RECORDS:
    c=get_case(rec);srcs=sampled_sources(c,MUTATION_SOURCES,7711)
    rng=np.random.default_rng(np.random.SeedSequence([BASE_SEED,int(c['seed']),919]))
    f_orig=q_arrays(c)
    for m_idx in range(MUTATIONS_PER_GENOME):
        g={name:val.copy() for name,val in c['genes'].items()}
        p=int(rng.integers(64));old=int(g['connections'][p]);choices=[v for v in (old-1,old+1) if 0<=v<=3]
        g['connections'][p]=int(rng.choice(choices))
        alt=c.copy();alt['genes']=g;alt['A']=make_A(g,*c['size'])
        alt['w']=np.array([np.sum(W64[g['connections']==d],dtype=np.float32)/np.sum(W64,dtype=np.float32)
                           for d in range(4)],dtype=np.float32)
        flags_in,flags_out=flags_by_rank(alt['A'],*c['size'],N_INPUTS,N_OUTPUTS)
        role_changes=int(np.sum(flags_in!=c['input'])+np.sum(flags_out!=c['output']))
        f_alt=q_arrays(alt)
        for mid in MODES:
            jac=[];deltaH=[]
            for src in srcs:
                dst=window_destinations(c,int(src));legal=outgoing_legal_mask(c,int(src),dst)
                h0=np.asarray(scores(c,src,dst,mid,f_orig),dtype=np.float32)
                h1=np.asarray(scores(alt,src,dst,mid,f_alt),dtype=np.float32)
                selected0=reference_top(h0,dst,c['nmax'],legal)
                selected1=reference_top(h1,dst,c['nmax'],legal)
                jac.append(jaccard(selected0,selected1))
                deltaH.append(float(np.mean(np.abs(h0-h1))))
            mut_rows.append({'id':mid,'seed':c['seed'],'mutation':m_idx,'gene_pos':p,
                'digit_before':old,'digit_after':int(g['connections'][p]),
                'jaccard':float(np.mean(jac)), 'mean_abs_delta_H':float(np.mean(deltaH)),
                'input_output_role_changes':role_changes})
mut_df=pd.DataFrame(mut_rows)
mut_df.to_csv(RESULT_DIR/'mutation_paired.csv',index=False)
mut_summary=mut_df.groupby('id',as_index=False).agg(mutation_jaccard=('jaccard','mean'),
    mutation_jaccard_p10=('jaccard',lambda s:np.quantile(s,.1)),
    mean_delta_H=('mean_abs_delta_H','mean'),
    mean_role_changes=('input_output_role_changes','mean'))
mut_summary.to_csv(RESULT_DIR/'mutation_summary.csv',index=False)
display(mut_summary.round(4))
```

## 8. Critérios de decisão **pré-especificados** e análise de sensibilidade

Os limites abaixo são **critérios de engenharia heurísticos**, não leis biológicas. Evitam que uma única métrica (isolamento) produza uma vitória espúria. Há dois níveis de conclusão:

* **Aceitável para integração experimental:** passa pelos filtros definidos e supera a referência no objetivo de isolamento.
* **Solução cognitiva definitiva:** **não pode ser afirmada** apenas com este notebook. Exige teste com o motor de dinâmica e tarefas de generalização.

Para tornar explícita a diferença entre “vencedor” e “preferência arbitrária de métricas”, mostramos também um ranking Pareto e a sensibilidade a pesos do escore. **Nenhum parâmetro de $H$ será reajustado depois de olhar o holdout.**

```python
THRESHOLDS={
    'isolation_pct':5.0,            # cobertura inicial >=95%
    'isolated_outputs':0,          # outputs devem receber entrada
    'reachable_input_output_pairs_pct':95.,
    'input_output_corridor_pct':65.,
    'mean_abs_A_diff':.20,         # impede preferência quase exclusiva por A idêntico
    'near_A_pct':25.,
    'weight_p05':.25,
    'mutation_jaccard':.25,
    'fp_agreement':.95,
    'cutoff_ties_pct':5.,
}
agg=full.groupby('id',as_index=False).agg(
    n_genomes=('seed','nunique'), isolation_pct=('isolation_pct','mean'),
    max_isolation_pct=('isolation_pct','max'), isolated_outputs=('isolated_output','sum'),
    reachable_input_output_pairs_pct=('reachable_input_output_pairs_pct','mean'),
    min_reachable_outputs_pct=('reachable_outputs_pct','min'),
    input_output_corridor_pct=('input_output_corridor_pct','mean'),
    near_A_pct=('near_A_pct','mean'), mean_abs_A_diff=('mean_abs_A_diff','mean'),
    weight_p05=('weight_p05','mean'),mean_weight=('weight_mean','mean'),
    reciprocity_pct=('reciprocity_pct','mean'),
    walk_output_mass_12=('walk_output_mass_12','mean'),
    gini=('gini_incoming','mean'),largest_scc_pct=('strong_largest_frac_pct','mean'),
    edge_total=('edges','sum'))
agg=agg.merge(num_summary,on='id').merge(mut_summary,on='id')
criteria={
 'isolation_pct':lambda r:r.isolation_pct<=THRESHOLDS['isolation_pct'],
 'isolated_outputs':lambda r:r.isolated_outputs<=THRESHOLDS['isolated_outputs'],
 'reachable_input_output_pairs_pct':lambda r:r.reachable_input_output_pairs_pct>=THRESHOLDS['reachable_input_output_pairs_pct'],
 'input_output_corridor_pct':lambda r:r.input_output_corridor_pct>=THRESHOLDS['input_output_corridor_pct'],
 'mean_abs_A_diff':lambda r:r.mean_abs_A_diff>=THRESHOLDS['mean_abs_A_diff'],
 'near_A_pct':lambda r:r.near_A_pct<=THRESHOLDS['near_A_pct'],
 'weight_p05':lambda r:r.weight_p05>=THRESHOLDS['weight_p05'],
 'mutation_jaccard':lambda r:r.mutation_jaccard>=THRESHOLDS['mutation_jaccard'],
 'fp_agreement':lambda r:r.fp_agreement>=THRESHOLDS['fp_agreement'],
 'cutoff_ties_pct':lambda r:r.cutoff_ties_pct<=THRESHOLDS['cutoff_ties_pct']}
passes=[]
for _,r in agg.iterrows():
    failed=[key for key,rule in criteria.items() if not rule(r)]
    passes.append({'id':r.id,'pass':not failed,'failed_count':len(failed),'failed_conditions':'; '.join(failed)})
agg=agg.merge(pd.DataFrame(passes),on='id')
# Score declarado: nenhuma métrica isolada pode dominar sem passar pelos filtros.
from numpy import clip
agg['quality_score']=(
   .27*(1-clip(agg.isolation_pct/60.,0,1))+
   .20*clip(agg.reachable_input_output_pairs_pct/100.,0,1)+
   .16*clip(agg.input_output_corridor_pct/100.,0,1)+
   .13*clip(agg.mutation_jaccard,0,1)+
   .09*(1-clip(agg.near_A_pct/100.,0,1))+
   .10*clip(agg.fp_agreement,0,1)+
   .05*clip(agg.walk_output_mass_12,0,1))
agg=agg.sort_values(['pass','quality_score','id'],ascending=[False,False,True])
agg.to_csv(RESULT_DIR/'decision_scorecard.csv',index=False)
display(agg[['id','pass','failed_conditions','isolation_pct','reachable_input_output_pairs_pct',
             'input_output_corridor_pct','near_A_pct','mutation_jaccard','fp_agreement','cutoff_ties_pct',
             'quality_score']].round(3))

# Pareto em isolamento, estabilidade, cobertura de pares e robustez numérica;
# sem transformar heurísticas de escore em verdades absolutas.
objective_cols=['isolation_pct','mutation_jaccard','reachable_input_output_pairs_pct','fp_agreement','near_A_pct']
mat=agg[objective_cols].to_numpy(float)*np.array([-1,1,1,1,-1])
pareto=np.array([not any(np.all(mat[j]>=mat[i]) and np.any(mat[j]>mat[i])
                        for j in range(len(mat)) if j!=i) for i in range(len(mat))])
agg['pareto']=pareto
agg.to_csv(RESULT_DIR/'decision_scorecard.csv',index=False)
print('Fronteira Pareto:',agg.loc[agg.pareto,'id'].tolist())
```

```python
# Partição por genoma congelada antes dos testes da Parte XI
split=full.copy();split['split']=np.where(split.held_out,'holdout','discovery')
validation=split.groupby(['id','split'],as_index=False).agg(isolation_pct=('isolation_pct','mean'),
     pair_reach_pct=('reachable_input_output_pairs_pct','mean'),
     corridor_pct=('input_output_corridor_pct','mean'),
     near_A_pct=('near_A_pct','mean'),weight_mean=('weight_mean','mean'))
validation.to_csv(RESULT_DIR/'discovery_vs_holdout.csv',index=False)
display(validation.round(3))

fig,axes=plt.subplots(2,2,figsize=(13,9),constrained_layout=True)
order=list(MODES); summary=agg.set_index('id').loc[order]
for ax,key,label in zip(axes.ravel(),
    ['isolation_pct','reachable_input_output_pairs_pct','mutation_jaccard','near_A_pct'],
    ['Isolamento antes do resgate (%)','Pares input-output alcançáveis (%)',
     'Jaccard após uma mutação','Arestas entre valores A próximos (%)']):
    ax.bar(range(len(order)),summary[key]);ax.set_xticks(range(len(order)),order,rotation=35,ha='right');ax.set_ylabel(label)
fig.suptitle(f'Parte XII — perfil {PROFILE}, {len(EVAL_RECORDS)} genomas')
fig.savefig(RESULT_DIR/'01_final_comparison.png',dpi=160);plt.show()

fig,axes=plt.subplots(1,2,figsize=(13,4.8),constrained_layout=True)
axes[0].scatter(agg.isolation_pct,agg.mutation_jaccard)
axes[1].scatter(agg.near_A_pct,agg.reachable_input_output_pairs_pct)
for _,r in agg.iterrows():
    axes[0].annotate(r.id,(r.isolation_pct,r.mutation_jaccard),xytext=(3,3),textcoords='offset points',fontsize=8)
    axes[1].annotate(r.id,(r.near_A_pct,r.reachable_input_output_pairs_pct),xytext=(3,3),textcoords='offset points',fontsize=8)
axes[0].set(xlabel='Isolamento (%)',ylabel='Jaccard após mutação',title='Cobertura versus estabilidade')
axes[1].set(xlabel='Homofilia |ΔA|<0,05 (%)',ylabel='Pares input-output alcançáveis (%)',
            title='Homofilia versus comunicação dirigida')
fig.savefig(RESULT_DIR/'02_tradeoffs.png',dpi=160);plt.show()

is_full=PROFILE=='full' and len(EVAL_RECORDS)==11
passing=agg[(agg['pass'])&(agg.id!='p0047')]
if not is_full:
    STATUS='INCONCLUSIVO — avaliação rápida, não decisão nos 11 genomas'
    candidate=agg.iloc[0]
elif len(passing):
    STATUS='CANDIDATA SELECIONADA PARA INTEGRAÇÃO — NÃO PROVA DE COGNIÇÃO'
    candidate=passing.iloc[0]
else:
    STATUS='NENHUMA FÓRMULA PASSOU TODOS OS CRITÉRIOS — SEM SOLUÇÃO DEFINITIVA COMPROVADA'
    # Melhor compromisso observado, mas não declarar como aprovada.
    candidate=agg.loc[agg.id!='p0047'].sort_values(['failed_count','quality_score'],ascending=[True,False]).iloc[0]

op,a,eta=MODES[candidate.id]
form={'cos*sin':'cos(X_t) sin(Y_t)','radial':'cos(X_t+delta_p) sin(Y_t+delta_q)',
      'sin+sin':'sin(X_t)+sin(Y_t)','sin+cos':'sin(X_t)+cos(Y_t)',
      'sin(Y-X)':'sin(Y_t-X_t)','sin*cos':'sin(X_t) cos(Y_t)','cos*cos':'cos(X_t) cos(Y_t)'}[op]
cpp_expression={
 'cos*sin':'cosf(m * Ap) * sinf(m * Aq)',
 'sin+sin':'sinf(m * Ap) + sinf(m * Aq)',
 'sin+cos':'sinf(m * Ap) + cosf(m * Aq)',
 'sin(Y-X)':'sinf(m * (Aq - Ap))',
 'sin*cos':'sinf(m * Ap) * cosf(m * Aq)',
 'cos*cos':'cosf(m * Ap) * cosf(m * Aq)',
 'radial':'cosf(m * Ap + delta_p) * sinf(m * Aq + delta_q)'}[op]
report=[f'# Resultado da Parte XII ({PROFILE})', '',f'**Status:** {STATUS}', '',
        f'**Melhor compromisso sob os critérios declarados:** {candidate.id}',
        f'**Fórmula:** H = sin({a:.12g} + ({eta:.12g}/C) × Σ[t=0..63] sqrt(1+t) × ({form}))',
        'Onde X_t=U(x_t) A_p; Y_t=U(x_t) A_q; C=Σ[t]sqrt(1+t).',
        '',f'Isolamento médio: {candidate.isolation_pct:.3f}%; ' +
           f'pares input-output: {candidate.reachable_input_output_pairs_pct:.2f}%; ' +
           f'Jaccard mutação: {candidate.mutation_jaccard:.4f}; '+
           f'near-A: {candidate.near_A_pct:.2f}%; '+
           f'FP32/FP64: {candidate.fp_agreement:.4f}.',
        f'Filtros não atendidos: {candidate.failed_conditions or "nenhum"}.',
        '', '## Limites desta decisão',
        f'- Testes realizados em {len(EVAL_RECORDS)} genomas e em grafos pré-resgate; não há ensaio de cognição.',
        '- Passeio ponderado não é a dinâmica de disparos do BrainDynamicsNative.',
        '- Empates FP32 e regras de resgate devem ser comparados com a GDExtension antes de substituir H.',
        '- Os limiares são decisões de engenharia; examine a fronteira de Pareto.']
(RESULT_DIR/'recommendation.md').write_text('\n'.join(report),encoding='utf-8')
# Snippet de implementação: não é patch automático; usuário insere apenas após validação.
snippet=f"""// Parte XII: {candidate.id}; status: {STATUS}
// X_t=(1+x_t)*Ap, Y_t=(1+x_t)*Aq, w_t=sqrtf(1+t).
// C = sum_t sqrtf(1+t); compute C once per genome.
float sum = 0.0f;
for (int t = 0; t < 64; ++t) {{
    const float m = 1.0f + static_cast<float>(connections_gene[t]);
    sum += sqrtf(1.0f + static_cast<float>(t)) * ({cpp_expression});
}}
float H = sinf({a:.12g}f + ({eta:.12g}f / C) * sum);
// C++ snippet pressupõe Ap, Aq, C, connections_gene; para radial também delta_p/delta_q.
"""
(RESULT_DIR/'candidate_cpp_snippet.txt').write_text(snippet,encoding='utf-8')
print('\n'.join(report[:8]))
if not is_full: print('ATENÇÃO: Execute PROFILE="full" para uma decisão final nos 11 genomas.')
```

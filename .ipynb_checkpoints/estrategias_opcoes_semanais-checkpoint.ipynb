# 1. Trava de Alta com PUT

A ideia é montar travas de alta com put de BOVA11 semanal, montando na segunda-feira e desmontando na sexta-feira. 

## 1.1. Matemática

Seja 
$$P_v = \text{prêmio da PUT vendida},$$
$$P_c = \text{prêmio da PUT comprada},$$
$$C = P_v - P_c = \text{crédito (líquido) recebido},$$
$$P_v^{stop} = \text{valor do prêmio de } P_v \text{ na stoploss},$$
$$P_c^{stop} = \text{valor do prêmio de } P_c \text{ na stoploss}.$$

No caso de uma loss, queremos que a perda seja no máximo $A$ vezes o crédito. Ou seja,
$$\text{perda} \leq A C = A (P_v - P_c).$$

O valor da stoploss é um multiplicador de $P_v$ a ser acionado automaticamente no homebroker. Seja $m$ este multiplicador e note que o resultado líquido é dado por
$$\text{resultado} = C - mP_v + P_c + (P_c^{stop} - P_c) = $$
$$ = P_v - P_c - mP_v + P_c + (P_c^{stop} - P_c) = $$
$$ = P_v - P_c - mP_v + P_c^{stop} = $$
$$ = (1 - m) P_v + (P_c^{stop} - P_c).$$

O multiplicador da stoploss satifaz $m > 1$, de modo que o valor acima é negativo, representando a perda líquida da operação. Vamos trabalhar com a perda em valores positivos para poder comparar com $AC$. Neste caso, queremos que

$$-(1 - m) P_v - (P_c^{stop} - P_c) \leq A (P_v - P_c)$$
$$m - 1 \leq \frac{ A (P_v - P_c) + (P_c^{stop} - P_c)}{P_v}$$
$$m \leq 1 + \frac{ A (P_v - P_c) + (P_c^{stop} - P_c)}{P_v}.$$

Assumindo o cenário mais conservador em que $P_c^{stop} - P_c \approx 0$, a fórmula pode ser simplificada para

$$m \leq 1 + \frac{ A (P_v - P_c)}{P_v} = 1 + \frac{ A C}{P_v}.$$

Daí segue que a fórmula para a stoploss é $P_v^{stop} = P_v + A (P_v - P_c)$. Ela vem da formulação para $m$, ou seja, 
$$P_v^{stop} = m P_v = \left(1 + A \frac{(P_v - P_c)}{P_v}\right) P_v = P_v + A (P_v - P_c).$$

## 1.2. Estratégia

**1:** Montar a operação na segunda-feira.

**2:** O strike da perna vendida e da perna comprada devem ter diferença de 1 real. Deve ser escolhido o strike mais baixo cujo crédito seja maior ou igual à 20 centavos.

**3:** Venda uma quantidade tal que a perda máxima seja igual o total no DEBB11. 

**4:** Quando o prêmio da opção vendida utrapassar $P_v + 4 (P_v - P_c)$, desfazer a operação. Utilize ordem de compra stop na perna vendida para garantir a saída da operação automaticamente. Você pode acompanhar o mercado antes de decidir desfazer a perna comprada. Se for uma queda considerável, a saída da operação ainda pode ser com lucro.

**5:** Encerrar a operação quando:
 - O lucro atingir 80% do lucro inicial.
 - For quinta-feira e o lucro atingir 50% do lucro inicial.
 - For o fim do pregão da quinta-feira, o ativo está ITM e não bateu na stoploss.
 - For sexta-feira de manhã (desmontar antes das 13:00 do jeito que está, não importa se é lucro ou prejuízo). 

**6:** Nunca deixar esta operação ir a exercício.

**7:** Assim que receber o prêmio, colocar no DEBB11. Esse dinheiro será utilizado para desmontes. Caso o total no DEBB11 chegue a R$ 100.000,00, todo lucro de opções será para investir normalmente em outros ativos. 

# 2. Venda coberta

**4:** Você deve ter o dinheiro para cobrir a put e deve ter os ativos para cobrir a call.

**5:** Assim que receber o prêmio, colocar no DEBB11. Esse dinheiro será utilizado para desmontes. Caso o total no DEBB11 chegue a R$ 100.000,00, todo lucro de opções será para investir normalmente em outros ativos. 

# 3. Riscos 

 1. O maior risco na trava de alta com put é no caso de uma queda grande e brusca. Esteja sempre atento para desmontar a operação, aceitando o prejuízo.
 2. O maior risco na parte dos dividendos sintéticos (call coberta) é quando o ativo sobre muito rápido. O prêmio pode passar ficar bem alto nestes casos. Não tente fazer rolagem, você corre o risco de estar apostando contra tendência e vai só perder mais dinheiro assim. Bateu no strike, encerra e pronto.
 3. Desmontar a operação e esquecer de cancelar a ordem stop é um risco constante. Sempre olhe as ordens abertas.
 4. A maior burrada é querer contar com a sorte. Toda vez que eu contei com o mercado indo em alguma direção para poder lucrar mais ou reduzir prejuízo, ele foi na direção contrária e eu me ferrei. Então fica aqui a dica de ouro:
$$\textbf{sempre siga a estratégia à risca e nunca conte com o mercado.}$$

# 4. Compra Stop

Assim que você montar a perna vendida, deverá também criar uma ordem de compra stop para desfazer essa perna no caso de uma desvalorização, isso foi mencionado no item 4 acima. A ideia é sair rápido da operação quando ela ficar ruim e manter a perna comprada para ganhar com a inversão do mercado. É importante saber como funciona o operacional para montar uma compra stop, por isso teremos esse pequeno tutorial.

A Ordem de Compra Stop é utilizada para realizar compras acima do preço que está sendo negociado o ativo no mercado. O preenchimento da Ordem de Compra Stop é semelhante a ordem de Compra, sendo que nessa boleta é adicionado um novo campo.

<p align="center">
    <img width="600" src="compra_stop.png">
</p>

O item *Stop* será o disparo da ordem de compra, enviando uma ordem com a configuração feita no campo *Preço*. No exemplo da imagem, como podemos observar na boleta, o ativo está sendo negociado no mercado ao preço de $R\$\ 19,05$, mas queremos comprar o ativo apenas caso ele atinja o valor de $R\$\ 19,50$. Então preenchemos o campo *Stop* com nosso preço de disparo, e o no campo *Preço*, colocamos o máximo que aceitamos pagar nesse ativo. Então quando atingir $R\$\ 19,50$, será enviado uma ordem de Compra ao mercado no preço de $R\$\ 19,60$. Caso o mercado esteja abaixo de $R\$\ 19,60$, nossa ordem será executada.
